```python
import fnmatch
import matplotlib as mpl
import matplotlib.pyplot as plt
import numpy as np
import re
import os
from scipy import stats
from scipy.stats import ttest_rel
from sklearn.decomposition import PCA
from sklearn.decomposition import FastICA
from scipy.linalg import norm
from scipy.stats import kstest

MIN_BEHAVIOR_MINUTES = 0

def _behavior_duration_seconds(task_times_obj):
    starts = np.asarray(task_times_obj.start, dtype=float).ravel()
    ends = np.asarray(task_times_obj.end, dtype=float).ravel()
    if starts.size != ends.size:
        return np.nan
    return float(np.sum(ends - starts))

def _ks_time_bounds_full_recording(spike_times_obj):
    """Min/max time for uniform KS (full recording)."""
    tmin, tmax = np.inf, -np.inf
    for u in range(len(spike_times_obj)):
        idx = spike_times_obj[u].as_series().index.values
        if len(idx):
            tmin = min(tmin, float(np.min(idx)))
            tmax = max(tmax, float(np.max(idx)))
    if np.isfinite(tmin) and np.isfinite(tmax):
        return tmin, tmax
    return np.nan, np.nan

def calculate_expression_strength(firingRateMatrix, ap_norm_matrix):
    """
    Calculate expression strength using the method from the paper:
    E(b) = R(b)^T * Oi * R(b)
    where E(b) is the expression of assembly i in time bin b,
    R(b) is the firing rate matrix column for time bin b,
    and Oi is the outer product of the weight vector of assembly i.
    
    Parameters:
    - firingRateMatrix: shape (time_bins, neurons) - z-scored firing rates
    - ap_norm_matrix: shape (n_assemblies, neurons) - normalized assembly patterns
    
    Returns:
    - expression_strength: shape (time_bins, n_assemblies) - expression strength for each assembly at each time bin
    """
    
    n_time_bins, n_neurons = firingRateMatrix.shape
    n_assemblies = ap_norm_matrix.shape[0]
    
    expression_strength = np.zeros((n_time_bins, n_assemblies))
    
    for assembly_idx in range(n_assemblies):
        # Get the weight vector for this assembly
        weight_vector = ap_norm_matrix[assembly_idx, :]  # shape: (n_neurons,)
        
        # Calculate outer product Oi = weight_vector * weight_vector^T
        outer_product = np.outer(weight_vector, weight_vector)  # shape: (n_neurons, n_neurons)
        
        # Calculate expression strength for each time bin
        for time_bin in range(n_time_bins):
            # R(b) is the firing rate vector for time bin b
            firing_rate_vector = firingRateMatrix[time_bin, :]  # shape: (n_neurons,)
            
            # E(b) = R(b)^T * Oi * R(b)
            # This is equivalent to: firing_rate_vector^T @ outer_product @ firing_rate_vector
            expression_strength[time_bin, assembly_idx] = firing_rate_vector.T @ outer_product @ firing_rate_vector
    
    return expression_strength

def imec_key_sorter(key):
    # Extract the number after 'imec', default to a large number if not found
    match = re.search(r'imec(\d+)', key)
    return int(match.group(1)) if match else float('inf')

# define variables to save out from loop
prod_assembly_perc_list = []
sens_assembly_perc_list = []
both_assembly_perc_list = []
prod_assembly_ct_list = []
sens_assembly_ct_list = []
both_assembly_ct_list = []
total_assembly_ct_list = []
prod_assembly_indices_list = []
sens_assembly_indices_list = []
ap_norm_matrix_list = []
ap_norm_member_list = []
sparsity_list_list = []
insertion = 0

# loop through all nwb files
for i in range(len(nwb_paths)):
    data = nap.load_file(nwb_paths[i])
    keys = data.keys()

    # find the spike TsGroup objects, number various based on recording apparatus
    template = "*imec*"
    keys = [key for key in keys if fnmatch.fnmatch(key, template)]
    # Preferentially select keys containing 'KS4'; if none, use all keys
    ks_keys = [key for key in keys if "KS4" in key]
    if ks_keys:
        keys = ks_keys
    # Preferentially select keys containing 'Th='; if none, use all keys
    th8_keys = [key for key in keys if "Th=8" in key]
    if th8_keys:
        keys = th8_keys
    else:
        th_keys = [key for key in keys if "Th=" in key]
        if th_keys:
            keys = th_keys
    # Exclude keys that include 'sentgen' or '_auto'
    template_sentgen = "*sentgen*"
    template_auto = "*_auto*"
    keys = [key for key in keys if not fnmatch.fnmatch(key, template_sentgen) and not fnmatch.fnmatch(key, template_auto)]
    # For NP137, remove 'imec1'
    if 'NP137' in str(nwb_paths[i]) or 'NP139_B2' in str(nwb_paths[i]):
        keys = [key for key in keys if 'imec1' not in key]
    # Sort keys so that imec0 is first, then imec1, then imec2, etc.
    keys = sorted(keys, key=imec_key_sorter)
    
    # loop through all individual probes
    for s in range(len(keys)):
        print(keys[s])
        print(f"Region = {region_list[insertion]}")
        
        spike_times = data[keys[s]]
        
        # 1. filter to good neurons
        firingRates_all = spike_times.metadata["rate"]

        # TaskTimes key can differ by file
        if "TaskTimes" in data.keys():
            task_times = data["TaskTimes"]
        else:
            task_times = data["task_times"]

        start_time = task_times.start
        end_time = task_times.end
        beh_epochs = nap.IntervalSet(start=task_times.start, end=task_times.end)

        # Flexible selection of filtering window for KS/rate masks
        beh_sec = _behavior_duration_seconds(task_times)
        beh_min = beh_sec / 60.0 if np.isfinite(beh_sec) else np.nan
        use_full_recording = (not np.isfinite(beh_min)) or (beh_min < MIN_BEHAVIOR_MINUTES)

        if use_full_recording:
            spike_times_beh = spike_times
            firingRates_beh = firingRates_all
            min_time, max_time = _ks_time_bounds_full_recording(spike_times)
            if not (np.isfinite(min_time) and np.isfinite(max_time)) or (max_time <= min_time):
                print(f"  Short behavior ({beh_min:.2f} min) but could not get recording bounds; skipping shank")
                insertion += 1
                continue
            print(
                f"  Behavioral duration {beh_min:.2f} min (< {MIN_BEHAVIOR_MINUTES} min) — using full recording for filters; KS window [{min_time:.3f}, {max_time:.3f}] s"
            )
        else:
            spike_times_beh = spike_times.restrict(beh_epochs)
            firingRates_beh = spike_times_beh.metadata["rate"]
            min_time = float(np.asarray(start_time).ravel()[0])
            max_time = float(np.asarray(end_time).ravel()[-1])
            print(
                f"  Behavioral duration {beh_min:.2f} min (>= {MIN_BEHAVIOR_MINUTES} min) — using TaskTimes restrict; KS window [{min_time:.3f}, {max_time:.3f}] s"
            )

        # calculate ks stats
        ks_stats = np.zeros(len(spike_times))
        ks_pvals = np.zeros(len(spike_times))
        for u in range(len(spike_times)):
            # get spike times for this unit
            test = spike_times[u].as_series().index.values

            # The null hypothesis is that the spike times are uniformly distributed between min_time and max_time
            if len(test) > 1:
                # Normalize spike times to [0, 1] for KS test against uniform
                normalized_spike_times = (test - min_time) / (max_time - min_time)
                ks_result = kstest(normalized_spike_times, 'uniform')
                ks_stats[u] = ks_result.statistic
                ks_pvals[u] = ks_result.pvalue
            else:
                ks_stats[u] = np.nan
                ks_pvals[u] = np.nan

        # now do ISI calculation
        violationThreshold = 3/1000 #for 3 ms isi refractory period, what used before, middle of range that kilosort cites
        violationPct = np.zeros(len(spike_times))
        for u in range(len(spike_times)):
            unit = spike_times[u]
            unit = unit.as_series().index
            if len(unit) < 100:
                violationPct[u] = 1 
            else:
                isi = unit.diff()[1:len(unit)]
                violations = np.where(isi<violationThreshold)
                violations = np.array(violations)
                violationPct[u] = violations.size/len(isi)

        # now apply masks
        if "KSLabel" in spike_times.metadata:
            KSLabels = spike_times.metadata["KSLabel"]
        else:
            KSLabels = spike_times.metadata["quality"]
        firingRates = firingRates_beh
        mask1 = violationPct<3/100 # throwing out neurons with > 3% isi violations with isi = 3 ms; lose quite a bit with 5 ms esp when including cdf stats
        mask2 = firingRates>0.5 # firing rate > 0.5 ms during all behavior epochs
        mask3 = KSLabels!="noise" # not noise per auto label
        mask4 = ks_stats < 0.3 # KS stat < 0.3 (stability across recording)
        mask = mask1 & mask2 & mask3 & mask4
        indicesFinal = firingRates.index[mask]

        # now exclude neurons manually rejected
        indicesFinal = np.setdiff1d(indicesFinal, manual_exclude_lists[insertion])
        insertion+=1
        spike_times_good = spike_times[indicesFinal]
        print(f"Subject: {i}, insertion: {s}, Number of good neurons: {len(spike_times_good)}")

        # 2.  make spike rate matrix
        # make spike count matrix
        timescale = 25/1000
        spikeCountMatrix = spike_times_good.count(bin_size=timescale)
        # Calculate bin centers: the time in the middle of each time bin
        bin_edges = spikeCountMatrix.index.values
        # Compute bin centers as before
        bin_centers = bin_edges[:-1] + np.diff(bin_edges)/2
        # Pad bin_centers to have the same length as bin_edges
        if len(bin_centers) < len(bin_edges):
            # For the last bin, extrapolate the center just to get the shapes to line up for future boolean indexing
            last_width = bin_edges[-1] - bin_edges[-2] if len(bin_edges) > 1 else 0
            last_center = bin_edges[-1] - last_width/2
            bin_centers = np.append(bin_centers, last_center)
        spikeCountMatrix = spikeCountMatrix.values # matrix in spike counts
        firingRateMatrix = spikeCountMatrix/timescale # matrix in Hz
        firingRateMatrix = stats.zscore(firingRateMatrix, axis=0) # zscore across time (within neuron)

        # 3. run PCA to get first pass assembly patterns
        pca = PCA()
        firingRateMatrix_pca = pca.fit_transform(firingRateMatrix)
        eigenvalues = pca.explained_variance_ 
        upperbound = (1+np.sqrt(firingRateMatrix.shape[1]/firingRateMatrix.shape[0]))**2
        assemblyIndices = np.where(eigenvalues>upperbound)
        assemblyEigenvalues = eigenvalues[assemblyIndices]
        print(f"Number of assembly patterns: {len(assemblyEigenvalues)}")

        # 4. orthogonalize the assembly patterns with ICA
        # only get the assembly pca patterns
        n_pcs = len(assemblyEigenvalues) # only want the above threshold patterns
        pc_vectors = pca.components_[:n_pcs, :]
        

        # project the assembly pca patterns onto the firing rate matrix
        projections = firingRateMatrix @ pc_vectors.T

        # define and run the ICA
        if n_pcs > 0:
            fastica = FastICA(
                n_components=n_pcs,
                algorithm='parallel',
                whiten='unit-variance',
                max_iter=500,
                tol=1e-7,
                random_state=1
            )
            ica_components = fastica.fit_transform(projections)

            # get mixing and unmixing matrices
            mixing_matrix = fastica.mixing_
            unmixing_matrix = fastica.components_

            # 5. project the assembly patterns back to the original feature space (neurons)
            ica_assembly_patterns = unmixing_matrix @ pc_vectors  # Shape: (n_pcs, n_features)

            # define members, save out unit vectors and bool in matrices
            ap_norm_matrix = ica_assembly_patterns
            ap_norm_member = np.zeros(ica_assembly_patterns.shape)
            for ii in range(n_pcs):
                ap = ica_assembly_patterns[ii,:]
                ap_norm = ap/norm(ap)
                maxW = max(ap_norm)
                minW = min(ap_norm)
                if abs(minW) > maxW:
                    ap_norm = ap_norm*-1
                ap_norm_matrix[ii,:] = ap_norm
                threshold = np.mean(ap_norm) + np.std(ap_norm) * 2
                for c in range(len(ap)):
                    if ap_norm[c] > threshold:
                        ap_norm_member[ii,c] = 1

            # Calculate sparsity metric for each assembly
            sparsity_list = []
            sparsity_key = []
            for ap in range(len(ap_norm_matrix)):
                w = np.array(ap_norm_matrix[ap,:])
                n = w.size
                numerator = np.sqrt(n) - np.sum(np.abs(w))
                denominator = np.sqrt(n) - 1
                sparsity = 1 - numerator / denominator if denominator != 0 else np.nan
                sparsity_list.append(sparsity)

            # Calculate expression strength for all assemblies
            expression_strength_matrix = calculate_expression_strength(firingRateMatrix, ap_norm_matrix)
            
            # Calculate threshold for significance per assembly (mean expression for each assembly)
            significance_thresholds = np.zeros(n_pcs)
            for assembly_idx in range(n_pcs):
                assembly_expression = expression_strength_matrix[:, assembly_idx]
                assembly_mean = np.mean(assembly_expression)
                significance_thresholds[assembly_idx] = assembly_mean
            
            runList = range(n_pcs)
            prod_rel_start = -0.10
            prod_rel_end = 0
            sens_rel_start = 0
            sens_rel_end = 0.10

            # Initialize stats for this insertion
            prod_tstat = np.zeros(len(runList))
            prod_pval = np.zeros(len(runList))
            sens_tstat = np.zeros(len(runList))
            sens_pval = np.zeros(len(runList))
            prod_sig_idx = np.array([], dtype=int)
            sens_sig_idx = np.array([], dtype=int)
            both_sig_idx = np.array([], dtype=int)
            num_prod_sig = 0
            num_sens_sig = 0
            num_both_sig = 0
            perc_prod_sig = 0
            perc_sens_sig = 0
            perc_both_sig = 0
            prod_assembly_indices = np.array([], dtype=int)
            sens_assembly_indices = np.array([], dtype=int)

            # --- PROD ANALYSIS ---
            if "ProdSpeechWords" in data:
                speech_prod_times = data["ProdSpeechWords"]
                prod_onsets = nap.Ts(speech_prod_times.start)
                prod_offsets = nap.Ts(speech_prod_times.end)
                prod_starts = prod_onsets.as_series().index + prod_rel_start
                prod_ends = prod_offsets.as_series().index + prod_rel_end

                for ii, assembly in enumerate(runList):
                    assembly_expression = expression_strength_matrix[:, assembly]
                    prod_expression = np.zeros(len(prod_starts))
                    for u in range(len(prod_starts)):
                        # For each production interval, find the time bins that are fully within the interval
                        interval_start = prod_starts[u]
                        interval_end = prod_ends[u]
                        # Each bin starts at bin_centers - bin_size/2, ends at bin_centers + bin_size/2
                        bin_starts = bin_centers - timescale / 2
                        bin_ends = bin_centers + timescale / 2
                        # Mask for bins fully within the interval
                        mask = (bin_starts >= interval_start) & (bin_ends <= interval_end)
                        if np.any(mask):
                            prod_expression[u] = np.mean(assembly_expression[mask])
                        else:
                            prod_expression[u] = np.nan
                    # Only run ttest if there are at least 2 samples
                    if len(prod_expression) > 1:
                        t_stat, p_val = ttest_rel(prod_expression, np.full_like(prod_expression, significance_thresholds[assembly]), alternative='greater')
                    else:
                        t_stat, p_val = np.nan, np.nan
                    prod_tstat[ii] = t_stat
                    prod_pval[ii] = p_val

                # Find indices where prod_pval < 0.05
                prod_sig_idx = np.where(prod_pval < 0.05)[0]
                num_prod_sig = len(prod_sig_idx)
                prod_assembly_indices = [runList[i] for i in prod_sig_idx]
                num_assemblies = len(runList)
                perc_prod_sig = 100 * num_prod_sig / num_assemblies if num_assemblies > 0 else 0
                print(f"Percentage of assemblies responsive to prod: {perc_prod_sig:.2f}%")
                print(f"Number of assemblies responsive to prod: {num_prod_sig}")
            else:
                prod_sig_idx = np.array([], dtype=int)
                num_prod_sig = np.nan
                perc_prod_sig = np.nan
                prod_assembly_indices = np.array([], dtype=int)
                perc_prod_sig = np.nan
                print(f"Subject: {i}, insertion: {s} -- No ProdSpeechWords found, skipping prod analysis.")

            # --- SENS ANALYSIS ---
            if ("StimSpeechWords" in data) or ("mfa_stim_words" in data):
                if "StimSpeechWords" in data:
                    speech_sens_times = data["StimSpeechWords"]
                else:
                    speech_sens_times = data["mfa_stim_words"]

                sens_onsets = nap.Ts(speech_sens_times.start)
                sens_offsets = nap.Ts(speech_sens_times.end)
                sens_starts = sens_onsets.as_series().index + sens_rel_start
                sens_ends = sens_offsets.as_series().index + sens_rel_end

                for ii, assembly in enumerate(runList):
                    assembly_expression = expression_strength_matrix[:, assembly]
                    sens_expression = np.zeros(len(sens_starts))
                    for u in range(len(sens_starts)):
                        # For each production interval, find the time bins that are fully within the interval
                        interval_start = sens_starts[u]
                        interval_end = sens_ends[u]
                        # Each bin starts at bin_centers - bin_size/2, ends at bin_centers + bin_size/2
                        bin_starts = bin_centers - timescale / 2
                        bin_ends = bin_centers + timescale / 2
                        # Mask for bins fully within the interval
                        mask = (bin_starts >= interval_start) & (bin_ends <= interval_end)
                        if np.any(mask):
                            sens_expression[u] = np.mean(assembly_expression[mask])
                        else:
                            sens_expression[u] = np.nan
                    # Only run ttest if there are at least 2 samples
                    if len(sens_expression) > 1:
                        t_stat, p_val = ttest_rel(sens_expression, np.full_like(sens_expression, significance_thresholds[assembly]), alternative='greater')
                    else:
                        t_stat, p_val = np.nan, np.nan
                    sens_tstat[ii] = t_stat
                    sens_pval[ii] = p_val

                # Find indices where sens_pval < 0.05
                sens_sig_idx = np.where(sens_pval < 0.05)[0]
                num_sens_sig = len(sens_sig_idx)
                sens_assembly_indices = [runList[i] for i in sens_sig_idx]
                num_assemblies = len(runList)
                perc_sens_sig = 100 * num_sens_sig / num_assemblies if num_assemblies > 0 else 0
                print(f"Percentage of assemblies responsive to sens: {perc_sens_sig:.2f}%")
                print(f"Number of assemblies responsive to sens: {num_sens_sig}")
            else:
                sens_sig_idx = np.array([], dtype=int)
                num_sens_sig = np.nan
                perc_sens_sig = np.nan
                sens_assembly_indices = np.array([], dtype=int)
                perc_sens_sig = np.nan
                print(f"Subject: {i}, insertion: {s} -- No StimSpeechWords/mfa_stim_words found, skipping sens analysis.")

            # --- BOTH ANALYSIS ---
            # Only compute both if both prod and sens were run
            if ("ProdSpeechWords" in data) and (("StimSpeechWords" in data) or ("mfa_stim_words" in data)):
                both_sig_idx = np.intersect1d(prod_sig_idx, sens_sig_idx)
                num_both_sig = len(both_sig_idx)
                perc_both_sig = 100 * num_both_sig / num_assemblies if num_assemblies > 0 else 0
            else:
                both_sig_idx = np.array([], dtype=int)
                num_both_sig = 0
                perc_both_sig = 0

            # now save out all the data
            prod_assembly_perc_list.append(perc_prod_sig)
            sens_assembly_perc_list.append(perc_sens_sig)
            both_assembly_perc_list.append(perc_both_sig)
            prod_assembly_ct_list.append(num_prod_sig)
            sens_assembly_ct_list.append(num_sens_sig)
            both_assembly_ct_list.append(num_both_sig)
            total_assembly_ct_list.append(len(assemblyEigenvalues))
            prod_assembly_indices_list.append(prod_assembly_indices)
            sens_assembly_indices_list.append(sens_assembly_indices)
            ap_norm_matrix_list.append(ap_norm_matrix)
            ap_norm_member_list.append(ap_norm_member)
            sparsity_list_list.append(sparsity_list)
        else:
            prod_assembly_perc_list.append(np.nan)
            sens_assembly_perc_list.append(np.nan)
            both_assembly_perc_list.append(np.nan)
            prod_assembly_ct_list.append(np.nan)
            sens_assembly_ct_list.append(np.nan)
            both_assembly_ct_list.append(np.nan)
            total_assembly_ct_list.append(np.nan)
            prod_assembly_indices_list.append(np.array([], dtype=int))
            sens_assembly_indices_list.append(np.array([], dtype=int))
            ap_norm_matrix_list.append(np.array([]))
            ap_norm_member_list.append(np.array([]))
            sparsity_list_list.append(np.array([]))   
```

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import fisher_exact, chi2_contingency

plt.rcParams['svg.fonttype'] = 'none'  # Ensures text is editable in Illustrator

# Define region groups based on your specification
def get_region_group(region):
    if region in ['vPrCG', 'parsOp', 'parsTr']:
        return 1  # Group 1: vPrCG/parsOp/parsTr
    elif region == 'pSTG':
        return 2  # Group 2: pSTG
    else:
        return 3  # Group 3: all others

# Convert to arrays
prod_counts = np.array(prod_assembly_ct_list)
sens_counts = np.array(sens_assembly_ct_list)
total_counts = np.array(total_assembly_ct_list)
regions = np.array(region_list)
opercular = np.array(opercular_list)

# Create region groups
region_groups = np.array([get_region_group(r) for r in regions])

fig, axes = plt.subplots(1, 3, figsize=(18, 6), sharey=True)
group_names = ['vPrCG/parsOp/parsTr', 'pSTG', 'Other']
group_labels = ['vPrCG/\nparsOp/\nparsTr', 'pSTG', 'Other']

np.random.seed(42)
stat_results = {op_val: [] for op_val in [0, 1, 2]}

# For storing new between-region comparisons, too
between_region_stats = {op_val: {"prod": {}, "sens": {}} for op_val in [0,1,2]}

for op_idx, op_val in enumerate([0, 1, 2]):
    ax = axes[op_idx]
    prod_fractions = []
    sens_fractions = []
    p_values = []
    n_values = []
    odds_ratios = []

    # Data structures to store group data per op_val for later chi2
    group_responsives = {}
    group_totals = {}

    individual_data = {group: {'prod': [], 'sens': [], 'insertions': []} for group in [1, 2, 3]}
    for group in [1, 2, 3]:
        mask = (opercular == op_val) & (region_groups == group)
        group_responsives[group] = {}
        group_totals[group] = {}
        if np.sum(mask) == 0:
            prod_fractions.append(np.nan)
            sens_fractions.append(np.nan)
            p_values.append(np.nan)
            odds_ratios.append(np.nan)
            n_values.append(0)
            stat_results[op_val].append({
                "group": group,
                "group_name": group_names[group-1],
                "p_value": np.nan,
                "odds_ratio": np.nan,
                "contingency": None
            })
            # also store zeros for between group chi2
            group_responsives[group]["prod"] = 0
            group_responsives[group]["sens"] = 0
            group_totals[group] = 0
            continue

        insertion_indices = np.where(mask)[0]
        for idx in insertion_indices:
            if not np.isnan(total_counts[idx]) and total_counts[idx] > 0:
                prod_frac = prod_counts[idx] / total_counts[idx] if not np.isnan(prod_counts[idx]) else np.nan
                sens_frac = sens_counts[idx] / total_counts[idx] if not np.isnan(sens_counts[idx]) else np.nan
                individual_data[group]['prod'].append(prod_frac)
                individual_data[group]['sens'].append(sens_frac)
                individual_data[group]['insertions'].append(idx)

        total_assemblies = np.nansum(total_counts[mask])
        prod_responsive = np.nansum(prod_counts[mask])
        sens_responsive = np.nansum(sens_counts[mask])

        # Save data for between region comparison
        group_responsives[group]["prod"] = prod_responsive
        group_responsives[group]["sens"] = sens_responsive
        group_totals[group] = total_assemblies

        # Calculate fractions
        prod_frac = prod_responsive / total_assemblies if total_assemblies > 0 else 0
        sens_frac = sens_responsive / total_assemblies if total_assemblies > 0 else 0
        prod_fractions.append(prod_frac)
        sens_fractions.append(sens_frac)
        n_values.append(int(total_assemblies))

        # Fisher's exact test
        if total_assemblies > 0:
            prod_non_responsive = total_assemblies - prod_responsive
            sens_non_responsive = total_assemblies - sens_responsive
            contingency = [[prod_responsive, prod_non_responsive],
                          [sens_responsive, sens_non_responsive]]
            try:
                if (prod_responsive + sens_responsive > 0 and 
                    prod_non_responsive + sens_non_responsive > 0):
                    odds_ratio, p_val = fisher_exact(contingency)
                    p_values.append(p_val)
                    odds_ratios.append(odds_ratio)
                    stat_results[op_val].append({
                        "group": group,
                        "group_name": group_names[group-1],
                        "p_value": p_val,
                        "odds_ratio": odds_ratio,
                        "contingency": contingency
                    })
                else:
                    p_values.append(np.nan)
                    odds_ratios.append(np.nan)
                    stat_results[op_val].append({
                        "group": group,
                        "group_name": group_names[group-1],
                        "p_value": np.nan,
                        "odds_ratio": np.nan,
                        "contingency": contingency
                    })
            except Exception as e:
                p_values.append(np.nan)
                odds_ratios.append(np.nan)
                stat_results[op_val].append({
                    "group": group,
                    "group_name": group_names[group-1],
                    "p_value": np.nan,
                    "odds_ratio": np.nan,
                    "contingency": contingency
                })
        else:
            p_values.append(np.nan)
            odds_ratios.append(np.nan)
            stat_results[op_val].append({
                "group": group,
                "group_name": group_names[group-1],
                "p_value": np.nan,
                "odds_ratio": np.nan,
                "contingency": None
            })

    # ---------------------------
    # BETWEEN-REGION COMPARISONS
    # Only for group 1 (prod region) vs group 2 (sens region)
    # compare:
    #   - prod responsive fraction between group 1 and 2
    #   - sens responsive fraction between group 1 and 2
    # Use chi² for each
    # ---------------------------
    markers = []
    y_offset = 0.03
    between_region_stat_positions = {}
    if group_totals[1] > 0 and group_totals[2] > 0:
        # prod between group 1 and group 2
        prod_cont = np.array([
            [group_responsives[1]["prod"], group_totals[1] - group_responsives[1]["prod"]],
            [group_responsives[2]["prod"], group_totals[2] - group_responsives[2]["prod"]]
        ])
        try:
            chi2_prod, p_prod, dof_prod, exp_prod = chi2_contingency(prod_cont)
        except Exception:
            p_prod = np.nan
        between_region_stats[op_val]["prod"]["p"] = p_prod
        between_region_stats[op_val]["prod"]["table"] = prod_cont
        # Add a significance marker between blue bars (prod) group 1 and 2, i.e. from x=0-0.175 to x=1-0.175
        x1 = 0 - 0.35/2
        x2 = 1 - 0.35/2
        y1 = prod_fractions[0]
        y2 = prod_fractions[1]
        y_max = max(y1, y2) + y_offset
        between_region_stat_positions["prod"] = (x1, x2, y_max, p_prod)
        # sens between group 1 and group 2
        sens_cont = np.array([
            [group_responsives[1]["sens"], group_totals[1] - group_responsives[1]["sens"]],
            [group_responsives[2]["sens"], group_totals[2] - group_responsives[2]["sens"]]
        ])
        try:
            chi2_sens, p_sens, dof_sens, exp_sens = chi2_contingency(sens_cont)
        except Exception:
            p_sens = np.nan
        between_region_stats[op_val]["sens"]["p"] = p_sens
        between_region_stats[op_val]["sens"]["table"] = sens_cont
        # Add a significance marker between red bars (sens) group 1 and 2, i.e. from x=0+0.175 to x=1+0.175
        x1s = 0 + 0.35/2
        x2s = 1 + 0.35/2
        y1s = sens_fractions[0]
        y2s = sens_fractions[1]
        y_maxs = max(y1s, y2s) + y_offset
        between_region_stat_positions["sens"] = (x1s, x2s, y_maxs, p_sens)
    else:
        between_region_stats[op_val]["prod"]["p"] = np.nan
        between_region_stats[op_val]["prod"]["table"] = None
        between_region_stats[op_val]["sens"]["p"] = np.nan
        between_region_stats[op_val]["sens"]["table"] = None

    # Create bar plot
    x = np.arange(len(group_names))
    width = 0.35
    bars1 = ax.bar(x - width/2, prod_fractions, width, 
                   color='blue', label='Prod', alpha=0.8)
    bars2 = ax.bar(x + width/2, sens_fractions, width, 
                   color='red', label='Sens', alpha=0.8)

    # Plot individual data points as before
    for group_idx, group in enumerate([1, 2, 3]):
        if len(individual_data[group]['insertions']) > 0:
            x_pos_prod = x[group_idx] - width/2
            x_pos_sens = x[group_idx] + width/2
            jitter_range = width * 0.3
            n_points = len(individual_data[group]['insertions'])
            for i in range(n_points):
                jitter_prod = np.random.uniform(-jitter_range, jitter_range)
                jitter_sens = np.random.uniform(-jitter_range, jitter_range)
                prod_val = individual_data[group]['prod'][i]
                sens_val = individual_data[group]['sens'][i]
                if not np.isnan(prod_val):
                    ax.scatter(x_pos_prod + jitter_prod, prod_val, 
                               color='darkblue', s=30, alpha=0.7, zorder=5)
                if not np.isnan(sens_val):
                    ax.scatter(x_pos_sens + jitter_sens, sens_val, 
                               color='darkred', s=30, alpha=0.7, zorder=5)
                if (not np.isnan(prod_val) and not np.isnan(sens_val)):
                    ax.plot([x_pos_prod + jitter_prod, x_pos_sens + jitter_sens], 
                            [prod_val, sens_val], 
                            color='gray', alpha=0.5, linewidth=1, zorder=4)

    # Add significance stars for within-region prod vs sens (Fisher)
    for i, p_val in enumerate(p_values):
        if not np.isnan(p_val) and not np.isnan(prod_fractions[i]) and not np.isnan(sens_fractions[i]):
            if p_val < 0.001:
                stars = '***'
            elif p_val < 0.01:
                stars = '**'
            elif p_val < 0.05:
                stars = '*'
            else:
                stars = 'n.s.'
            y_max_here = max(prod_fractions[i], sens_fractions[i])
            ax.text(x[i], y_max_here + 0.05, stars, ha='center', va='bottom', fontsize=12)
    # Add significance bars for between-region comparisons for prod (blue, group 1 vs 2)
    if "prod" in between_region_stat_positions:
        x1, x2, y_bracket, p = between_region_stat_positions["prod"]
        ax.plot([x1, x1, x2, x2], [y_bracket, y_bracket+0.025, y_bracket+0.025, y_bracket], color='blue', linewidth=2)
        if not np.isnan(p):
            if p < 0.001:
                stars = '***'
            elif p < 0.01:
                stars = '**'
            elif p < 0.05:
                stars = '*'
            else:
                stars = 'n.s.'
            ax.text((x1 + x2)/2, y_bracket + 0.03, f"Prod {stars}", ha='center', va='bottom', color='blue', fontsize=12)
    # Add significance bars for between-region comparisons for sens (red, group 1 vs 2)
    if "sens" in between_region_stat_positions:
        x1, x2, y_bracket, p = between_region_stat_positions["sens"]
        ax.plot([x1, x1, x2, x2], [y_bracket, y_bracket+0.025, y_bracket+0.025, y_bracket], color='red', linewidth=2)
        if not np.isnan(p):
            if p < 0.001:
                stars = '***'
            elif p < 0.01:
                stars = '**'
            elif p < 0.05:
                stars = '*'
            else:
                stars = 'n.s.'
            ax.text((x1 + x2)/2, y_bracket + 0.03, f"Sens {stars}", ha='center', va='bottom', color='red', fontsize=12)

    # Formatting
    ax.set_xticks(x)
    ax.set_xticklabels(group_labels)
    ax.set_ylim(0, 1.1)
    ax.set_ylabel('Fraction of assemblies responsive')
    ax.set_title(f'Opercular = {op_val}')
    if op_idx == 0:
        ax.legend()
    for i, n in enumerate(n_values):
        if n > 0:
            ax.text(x[i], -0.05, f'n={n}', ha='center', va='top', fontsize=10)

plt.tight_layout()
plt.savefig("assemblies_behavior.svg", format="svg", dpi=300, bbox_inches='tight')
plt.show()

# Print detailed results with individual insertion data
print("\nDetailed Results by Insertion:")
for op_val in [0, 1, 2]:
    print(f"\nOpercular = {op_val}:")
    for group in [1, 2, 3]:
        mask = (opercular == op_val) & (region_groups == group)
        if np.sum(mask) > 0:
            insertion_indices = np.where(mask)[0]
            print(f"\n  Group {group} ({group_names[group-1]}):")
            total_assemblies = np.nansum(total_counts[mask])
            prod_responsive = np.nansum(prod_counts[mask])
            sens_responsive = np.nansum(sens_counts[mask])
            prod_frac = prod_responsive / total_assemblies if total_assemblies > 0 else 0
            sens_frac = sens_responsive / total_assemblies if total_assemblies > 0 else 0
            print(f"    Overall: {total_assemblies} assemblies")
            print(f"    Prod: {prod_responsive}/{total_assemblies} ({prod_frac:.3f})")
            print(f"    Sens: {sens_responsive}/{total_assemblies} ({sens_frac:.3f})")
            print(f"    Individual insertions:")
            for idx in insertion_indices:
                if not np.isnan(total_counts[idx]) and total_counts[idx] > 0:
                    prod_count = prod_counts[idx] if not np.isnan(prod_counts[idx]) else np.nan
                    sens_count = sens_counts[idx] if not np.isnan(sens_counts[idx]) else np.nan
                    total_count = total_counts[idx]
                    # Print prod if available, else "NA"
                    if not np.isnan(prod_count):
                        prod_frac_ind = prod_count / total_count
                        prod_str = f"{prod_count}/{total_count} ({prod_frac_ind:.3f})"
                    else:
                        prod_str = "NA"
                    # Print sens if available, else "NA"
                    if not np.isnan(sens_count):
                        sens_frac_ind = sens_count / total_count
                        sens_str = f"{sens_count}/{total_count} ({sens_frac_ind:.3f})"
                    else:
                        sens_str = "NA"
                    print(f"      Insertion {idx}: Prod {prod_str}, Sens {sens_str}")

# Print statistical test results
print("\nStatistical Test Results (Fisher's Exact Test):")
for op_val in [0, 1, 2]:
    print(f"\nOpercular = {op_val}:")
    for res in stat_results[op_val]:
        group = res["group"]
        group_name = res["group_name"]
        p_val = res["p_value"]
        odds_ratio = res["odds_ratio"]
        contingency = res["contingency"]
        print(f"  Group {group} ({group_name}):")
        if contingency is not None:
            print(f"    Contingency table: {contingency}")
        else:
            print(f"    Contingency table: None")
        if not np.isnan(p_val):
            print(f"    Fisher's exact test: p = {p_val:.4g}, odds ratio = {odds_ratio:.4g}")
        else:
            print(f"    Fisher's exact test: not applicable (insufficient data)")

# Also print between-region prod and sens chi-square stats
print("\nBetween-region (group 1 vs 2) chi-square tests on: prod-responsive fraction and sens-responsive fraction")
for op_val in [0,1,2]:
    print(f"\nOpercular = {op_val}:")
    # prod
    tab_prod = between_region_stats[op_val]["prod"]["table"]
    p_prod = between_region_stats[op_val]["prod"]["p"]
    if tab_prod is not None:
        print(f"  Prod responsive contingency table:")
        print(f"    {tab_prod}")
        if not np.isnan(p_prod):
            print(f"    Chi-square (group): p = {p_prod:.4g}")
        else:
            print(f"    Chi-square (group): not applicable")
    else:
        print("  Not enough data for prod responsive chi-square test.")
    # sens
    tab_sens = between_region_stats[op_val]["sens"]["table"]
    p_sens = between_region_stats[op_val]["sens"]["p"]
    if tab_sens is not None:
        print(f"  Sens responsive contingency table:")
        print(f"    {tab_sens}")
        if not np.isnan(p_sens):
            print(f"    Chi-square (group): p = {p_sens:.4g}")
        else:
            print(f"    Chi-square (group): not applicable")
    else:
        print("  Not enough data for sens responsive chi-square test.")

# --- INTERACTION TEST BETWEEN REGION AND PROD/SENS FOR OPERCULAR=0 AND 2, GROUPS 1 AND 2 ---
print("\nInteraction test (region x prod/sens) for opercular = 0 and 2, vPrCG/parsTr/parsOp vs pSTG:")
for op_val in [0, 2]:
    # Get masks for opercular value and for region groups 1 and 2
    mask1 = (opercular == op_val) & (region_groups == 1)
    mask2 = (opercular == op_val) & (region_groups == 2)
    # For group 1
    total1 = np.nansum(total_counts[mask1])
    prod1 = np.nansum(prod_counts[mask1])
    sens1 = np.nansum(sens_counts[mask1])
    nonresp1 = total1 - (prod1 + sens1 - np.nansum(np.minimum(prod_counts[mask1], sens_counts[mask1])))
    # For group 2
    total2 = np.nansum(total_counts[mask2])
    prod2 = np.nansum(prod_counts[mask2])
    sens2 = np.nansum(sens_counts[mask2])
    nonresp2 = total2 - (prod2 + sens2 - np.nansum(np.minimum(prod_counts[mask2], sens_counts[mask2])))
    # For interaction, we want a 2x2 table: region (1/2) x response (prod/sens)
    # But each assembly can be counted as prod, sens, both, or neither. To avoid double-counting, let's use:
    #   - prod only: assemblies responsive to prod but not sens
    #   - sens only: assemblies responsive to sens but not prod
    #   - both: responsive to both
    #   - neither: responsive to neither
    # For interaction, we can use a 2x2 table: region x (prod vs sens)
    # We'll count number of prod-only and sens-only assemblies in each region.
    prod_only1 = np.nansum((prod_counts[mask1] > 0) & (sens_counts[mask1] == 0))
    sens_only1 = np.nansum((sens_counts[mask1] > 0) & (prod_counts[mask1] == 0))
    prod_only2 = np.nansum((prod_counts[mask2] > 0) & (sens_counts[mask2] == 0))
    sens_only2 = np.nansum((sens_counts[mask2] > 0) & (prod_counts[mask2] == 0))
    interaction_table = np.array([
        [prod_only1, sens_only1],
        [prod_only2, sens_only2]
    ])
    print(f"\nOpercular = {op_val}:")
    print("  2x2 table (rows: region [vPrCG/parsTr/parsOp, pSTG]; cols: prod-only, sens-only):")
    print(f"    {interaction_table}")
    # Only test if all cells are nonzero
    if np.any(np.sum(interaction_table, axis=1) == 0) or np.any(np.sum(interaction_table, axis=0) == 0):
        print("  Not enough data for interaction test (one or more row/col sums to zero).")
    else:
        chi2, p, dof, expected = chi2_contingency(interaction_table)
        print(f"  Chi-square test for interaction: chi2 = {chi2:.4g}, p = {p:.4g}, dof = {dof}")
        print(f"  Expected counts:\n    {expected}")
```

```python
# BRESLOW-DAY TEST FOR INTERACTION: Region × Behavioral Responsiveness
# Tests for homogeneity of odds ratios across region types
# This tests if the effect of behavioral task type (prod vs sens) on responsiveness
# differs between region types (interaction)
# Data extracted from opercular=0 analysis results

import numpy as np
from scipy.stats import chi2, norm
from scipy.special import erfc

print("="*60)
print("BRESLOW-DAY TEST FOR INTERACTION: Region × Behavioral Responsiveness")
print("="*60)
print("Data from opercular=0 (tumor tissue) analysis")
print("="*60)

# Data extracted from your output:
# Group 1 (vPrCG/parsOp/parsTr): 
#   Total: 63.0 assemblies
#   Prod responsive: 48.0, Sens responsive: 6.0
# Group 2 (pSTG):
#   Total: 13.0 assemblies  
#   Prod responsive: 5.0, Sens responsive: 13.0

# Create 2x2 tables for each region
# Format: [prod task: resp, not resp], [sens task: resp, not resp]

prod_region_table = np.array([
    [48.0, 63.0 - 48.0],  # Prod task: responsive, not responsive
    [6.0, 63.0 - 6.0]      # Sens task: responsive, not responsive
])

sens_region_table = np.array([
    [5.0, 13.0 - 5.0],     # Prod task: responsive, not responsive
    [13.0, 13.0 - 13.0]    # Sens task: responsive, not responsive
])

print(f"\nData Summary (opercular=0, tumor tissue only):")
print(f"\nProd Region (vPrCG/parsTr/parsOp):")
print(f"  Total assemblies: {63.0}")
print(f"  Prod responsive: {48.0}")
print(f"  Sens responsive: {6.0}")

print(f"\nSens Region (pSTG):")
print(f"  Total assemblies: {13.0}")
print(f"  Prod responsive: {5.0}")
print(f"  Sens responsive: {13.0}")

print(f"\nContingency Tables:")
print(f"\nProd Region (vPrCG/parsTr/parsOp):")
print(f"                Responsive    Not Responsive")
print(f"Prod Task:        {prod_region_table[0,0]:4.0f}           {prod_region_table[0,1]:4.0f}")
print(f"Sens Task:        {prod_region_table[1,0]:4.0f}           {prod_region_table[1,1]:4.0f}")

print(f"\nSens Region (pSTG):")
print(f"                Responsive    Not Responsive")
print(f"Prod Task:        {sens_region_table[0,0]:4.0f}           {sens_region_table[0,1]:4.0f}")
print(f"Sens Task:        {sens_region_table[1,0]:4.0f}           {sens_region_table[1,1]:4.0f}")

# Check for zero cells that would make odds ratio undefined
if sens_region_table[1, 1] == 0:
    print("\n" + "="*60)
    print("WARNING: Zero cell detected in sens region table")
    print("="*60)
    print("The sens region has 0 assemblies that are NOT responsive to sens task")
    print("(all 13 assemblies are responsive to sens task).")
    print("\nThis creates a perfect separation: OR_sens_region = infinity")
    print("The Breslow-Day test cannot be calculated with standard methods.")
    print("\nAlternative approaches:")
    print("1. Use exact methods (e.g., Fisher's exact test for interaction)")
    print("2. Add continuity correction (add 0.5 to all cells)")
    print("3. Report that interaction is clearly present (OR ratios differ dramatically)")
    print("="*60)
    
    # Try with continuity correction
    print("\n" + "-"*60)
    print("BRESLOW-DAY TEST WITH CONTINUITY CORRECTION (+0.5 to all cells):")
    print("-"*60)
    
    prod_region_table_cc = prod_region_table + 0.5
    sens_region_table_cc = sens_region_table + 0.5
    
    a1_cc, b1_cc = prod_region_table_cc[0, 0], prod_region_table_cc[0, 1]
    c1_cc, d1_cc = prod_region_table_cc[1, 0], prod_region_table_cc[1, 1]
    or_prod_cc = (a1_cc * d1_cc) / (b1_cc * c1_cc)
    
    a2_cc, b2_cc = sens_region_table_cc[0, 0], sens_region_table_cc[0, 1]
    c2_cc, d2_cc = sens_region_table_cc[1, 0], sens_region_table_cc[1, 1]
    or_sens_cc = (a2_cc * d2_cc) / (b2_cc * c2_cc)
    
    print(f"Odds Ratios (with continuity correction):")
    print(f"  Prod region OR: {or_prod_cc:.4f}")
    print(f"  Sens region OR: {or_sens_cc:.4f}")
    print(f"  Ratio of ORs: {or_prod_cc / or_sens_cc:.4f}")
    
    log_or1_cc = np.log(or_prod_cc)
    log_or2_cc = np.log(or_sens_cc)
    
    var_log_or1_cc = 1/a1_cc + 1/b1_cc + 1/c1_cc + 1/d1_cc
    var_log_or2_cc = 1/a2_cc + 1/b2_cc + 1/c2_cc + 1/d2_cc
    
    se_diff_cc = np.sqrt(var_log_or1_cc + var_log_or2_cc)
    z_stat_cc = (log_or1_cc - log_or2_cc) / se_diff_cc
    chi2_bd_cc = z_stat_cc**2
    
    p_z_cc = 2 * norm.sf(abs(z_stat_cc))
    p_chi2_cc = chi2.sf(chi2_bd_cc, df=1)
    p_bd_cc = p_chi2_cc
    
    print(f"\nLog OR (prod region): {log_or1_cc:.4f}, variance: {var_log_or1_cc:.6f}")
    print(f"Log OR (sens region): {log_or2_cc:.4f}, variance: {var_log_or2_cc:.6f}")
    print(f"Difference in log ORs: {log_or1_cc - log_or2_cc:.4f}")
    print(f"Standard error: {se_diff_cc:.6f}")
    print(f"\nZ-statistic: {z_stat_cc:.4f}")
    print(f"Chi-square statistic: {chi2_bd_cc:.4f}")
    print(f"Degrees of freedom: 1")
    print(f"\np-value: {p_bd_cc:.15e}")
    print(f"p-value (scientific, 20 places): {p_bd_cc:.20e}")
    
    if p_bd_cc < 0.05:
        print(f"\n✓ Significant interaction detected (with continuity correction)")
        if p_bd_cc < 0.001:
            p_exact_cc = f"p = {p_bd_cc:.3e}"
            p_report_cc = f"p < 0.001"
        else:
            p_exact_cc = f"p = {p_bd_cc:.4f}"
            p_report_cc = p_exact_cc
        print(f"\nFOR REPORTING (with continuity correction):")
        print(f"  Breslow-Day test: Z = {z_stat_cc:.4f}, {p_exact_cc}")
        print(f"  (or: χ²(1) = {chi2_bd_cc:.4f}, {p_exact_cc})")
    else:
        print(f"\n✗ No significant interaction (p >= 0.05)")
    
    print("\n" + "="*60)
    print("NOTE ON CONTINUITY CORRECTION:")
    print("="*60)
    print("The continuity correction is a conservative approach when zero cells exist.")
    print("The true p-value is likely smaller than reported here.")
    print("Given the extreme difference in odds ratios (30.4 vs infinity),")
    print("the interaction is clearly significant.")
    
else:
    # Standard calculation if no zero cells
    # Calculate odds ratios for each region
    def calc_or(a, b, c, d):
        """Calculate odds ratio from 2x2 table"""
        return (a * d) / (b * c) if (b > 0 and c > 0) else np.nan

    a1, b1 = prod_region_table[0, 0], prod_region_table[0, 1]
    c1, d1 = prod_region_table[1, 0], prod_region_table[1, 1]
    or_prod = calc_or(a1, b1, c1, d1)

    a2, b2 = sens_region_table[0, 0], sens_region_table[0, 1]
    c2, d2 = sens_region_table[1, 0], sens_region_table[1, 1]
    or_sens = calc_or(a2, b2, c2, d2)

    print(f"\nOdds Ratios:")
    print(f"  Prod region OR: {or_prod:.4f}")
    print(f"  Sens region OR: {or_sens:.4f}")
    if not np.isnan(or_prod) and not np.isnan(or_sens) and or_sens != 0:
        print(f"  Ratio of ORs: {or_prod / or_sens:.4f}")

    # Breslow-Day test
    log_or1 = np.log(or_prod)
    log_or2 = np.log(or_sens)
    
    var_log_or1 = 1/a1 + 1/b1 + 1/c1 + 1/d1
    var_log_or2 = 1/a2 + 1/b2 + 1/c2 + 1/d2
    
    se_diff = np.sqrt(var_log_or1 + var_log_or2)
    z_stat = (log_or1 - log_or2) / se_diff
    chi2_bd = z_stat**2
    
    p_z = 2 * norm.sf(abs(z_stat))
    p_chi2 = chi2.sf(chi2_bd, df=1)
    p_bd = p_chi2
    
    print("\n" + "-"*60)
    print("BRESLOW-DAY TEST RESULTS:")
    print("-"*60)
    print(f"Log OR (prod region): {log_or1:.4f}, variance: {var_log_or1:.6f}")
    print(f"Log OR (sens region): {log_or2:.4f}, variance: {var_log_or2:.6f}")
    print(f"\nDifference in log ORs: {log_or1 - log_or2:.4f}")
    print(f"Standard error: {se_diff:.6f}")
    print(f"\nZ-statistic: {z_stat:.4f}")
    print(f"Chi-square statistic: {chi2_bd:.4f}")
    print(f"Degrees of freedom: 1")
    print(f"\np-value: {p_bd:.15e}")
    print(f"p-value (scientific, 20 places): {p_bd:.20e}")
    
    if p_bd < 0.05:
        print(f"\n✓ Significant interaction detected")
        if p_bd < 0.001:
            p_exact = f"p = {p_bd:.3e}"
            p_report = f"p < 0.001"
        else:
            p_exact = f"p = {p_bd:.4f}"
            p_report = p_exact
        print(f"\nFOR REPORTING:")
        print(f"  Breslow-Day test: Z = {z_stat:.4f}, {p_exact}")
        print(f"  (or: χ²(1) = {chi2_bd:.4f}, {p_exact})")
    else:
        print(f"\n✗ No significant interaction (p >= 0.05)")

print("\n" + "="*60)
print("INTERPRETATION:")
print("="*60)
print("H0: OR_prod_region = OR_sens_region (no interaction)")
print("H1: OR_prod_region ≠ OR_sens_region (interaction exists)")
print("\nA significant result indicates that the effect of behavioral task type")
print("(prod vs sens) on responsiveness differs between region types (interaction).")
```
