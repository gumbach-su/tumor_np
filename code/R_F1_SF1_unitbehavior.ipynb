```python
import fnmatch
import matplotlib as mpl
import matplotlib.pyplot as plt
import numpy as np
from scipy import stats
from sklearn.decomposition import PCA
from sklearn.decomposition import FastICA
from scipy.linalg import norm
from scipy.stats import kstest
from scipy.stats import ttest_rel
import re

def imec_key_sorter(key):
    # Extract the number after 'imec', default to a large number if not found
    match = re.search(r'imec(\d+)', key)
    return int(match.group(1)) if match else float('inf')

# ---- Behavior-duration-aware filtering (mirrors R_AIproduction.ipynb) ----
# When behavioral epochs are too short to give a stable rate / KS window,
# fall back to the full recording instead of the TaskTimes-restricted view.
MIN_BEHAVIOR_MINUTES = 0

def _behavior_duration_seconds(task_times_obj):
    starts = np.asarray(task_times_obj.start, dtype=float).ravel()
    ends   = np.asarray(task_times_obj.end,   dtype=float).ravel()
    if starts.size != ends.size:
        return np.nan
    return float(np.sum(ends - starts))

def _ks_time_bounds_full_recording(spike_times_obj):
    """Min / max spike time across all units (KS uniform null on the full recording)."""
    tmin, tmax = np.inf, -np.inf
    for u in range(len(spike_times_obj)):
        idx = spike_times_obj[u].as_series().index.values
        if len(idx):
            tmin = min(tmin, float(np.min(idx)))
            tmax = max(tmax, float(np.max(idx)))
    if np.isfinite(tmin) and np.isfinite(tmax):
        return tmin, tmax
    return np.nan, np.nan

# define variables to save out from loop
prod_neuron_indices_list = []
sens_neuron_indices_list = []
prod_neuron_perc_list = []
sens_neuron_perc_list = []
both_neuron_perc_list = []
prod_neuron_ct_list =  []
sens_neuron_ct_list = []
both_neuron_ct_list = []
total_neuron_ct_list = []
tumor_id_list = []
infiltrated_id_list = []
insertion_id_list = []
i_save = 0
insertion = 0

# loop through all nwb files
for i in range(len(nwb_paths)):
    data = nap.load_file(nwb_paths[i])
    keys = data.keys()

    # find the spike TsGroup objects, number various based on recording apparatus
    template = "*imec*"
    keys = [key for key in keys if fnmatch.fnmatch(key, template)]
    # Preferentially select keys containing 'KS4'; if none, use all keys
    ks_keys = [key for key in keys if "KS4" in key]
    if ks_keys:
        keys = ks_keys
    # Preferentially select keys containing 'Th='; if none, use all keys
    th8_keys = [key for key in keys if "Th=8" in key]
    if th8_keys:
        keys = th8_keys
    else:
        th_keys = [key for key in keys if "Th=" in key]
        if th_keys:
            keys = th_keys
    # Drop sentgen and *_auto* shanks (match R_AIproduction)
    template_sentgen = "*sentgen*"
    template_auto    = "*_auto*"
    keys = [key for key in keys if not fnmatch.fnmatch(key, template_sentgen) and not fnmatch.fnmatch(key, template_auto)]
    # For NP137, remove 'imec1'
    if 'NP137' in str(nwb_paths[i]) or 'NP139_B2' in str(nwb_paths[i]):
        keys = [key for key in keys if 'imec1' not in key]
    # Sort keys so that imec0 is first, then imec1, then imec2, etc.
    keys = sorted(keys, key=imec_key_sorter)

    for s in range(len(keys)):
        print(keys[s])
        print(region_list[insertion])
        spike_times = data[keys[s]]
        
        # 1. filter to good neurons
        firingRates_all = spike_times.metadata["rate"]

        # task_times key may be capitalized differently across sessions
        if "TaskTimes" in data.keys():
            task_times = data["TaskTimes"]
        else:
            task_times = data["task_times"]
        start_time = task_times.start
        end_time   = task_times.end
        beh_epochs = nap.IntervalSet(start=task_times.start, end=task_times.end)

        # Decide whether behavior is long enough; if not, use full recording for filters/KS.
        beh_sec = _behavior_duration_seconds(task_times)
        beh_min = beh_sec / 60.0 if np.isfinite(beh_sec) else np.nan
        use_full_recording = (not np.isfinite(beh_min)) or (beh_min < MIN_BEHAVIOR_MINUTES)

        if use_full_recording:
            spike_times_beh = spike_times
            firingRates_beh = firingRates_all
            min_time, max_time = _ks_time_bounds_full_recording(spike_times)
            if not (np.isfinite(min_time) and np.isfinite(max_time)) or (max_time <= min_time):
                print(f"  Short behavior ({beh_min:.2f} min) but could not get recording bounds; skipping shank")
                insertion += 1
                continue
            print(f"  Behavioral duration {beh_min:.2f} min (< {MIN_BEHAVIOR_MINUTES} min) — using full recording for filters; KS window [{min_time:.3f}, {max_time:.3f}] s")
        else:
            spike_times_beh = spike_times.restrict(beh_epochs)
            firingRates_beh = spike_times_beh.metadata["rate"]
            min_time = float(np.asarray(start_time).ravel()[0])
            max_time = float(np.asarray(end_time).ravel()[-1])
            print(f"  Behavioral duration {beh_min:.2f} min (>= {MIN_BEHAVIOR_MINUTES} min) — using TaskTimes restrict; KS window [{min_time:.3f}, {max_time:.3f}] s")

        # KS test against uniform on the chosen [min_time, max_time] window
        ks_stats = np.zeros(len(spike_times))
        ks_pvals = np.zeros(len(spike_times))
        for u in range(len(spike_times)):
            test = spike_times[u].as_series().index.values
            if len(test) > 1:
                normalized_spike_times = (test - min_time) / (max_time - min_time)
                ks_result = kstest(normalized_spike_times, 'uniform')
                ks_stats[u] = ks_result.statistic
                ks_pvals[u] = ks_result.pvalue
            else:
                ks_stats[u] = np.nan
                ks_pvals[u] = np.nan

        # now do ISI calculation
        violationThreshold = 3/1000 #for 3 ms isi refractory period, what used before, middle of range that kilosort cites
        violationPct = np.zeros(len(spike_times))
        for u in range(len(spike_times)):
            unit = spike_times[u]
            unit = unit.as_series().index
            if len(unit) < 100:
                violationPct[u] = 1 
            else:
                isi = unit.diff()[1:len(unit)]
                violations = np.where(isi<violationThreshold)
                violations = np.array(violations)
                violationPct[u] = violations.size/len(isi)

        # now apply masks
        if "KSLabel" in spike_times.metadata:
            KSLabels = spike_times.metadata["KSLabel"]
        else:
            KSLabels = spike_times.metadata["quality"]
        firingRates = firingRates_beh
        mask1 = violationPct<3/100 # throwing out neurons with > 3% isi violations with isi = 3 ms; lose quite a bit with 5 ms esp when including cdf stats
        mask2 = firingRates>0.5 # firing rate > 0.5 ms during all behavior epochs
        mask3 = KSLabels!="noise" # not noise per auto label
        mask4 = ks_stats < 0.3 # KS stat < 0.3 (stability across recording)
        mask = mask1 & mask2 & mask3 & mask4
        indicesFinal = firingRates.index[mask]

        # now exclude neurons manually rejected
        indicesFinal = np.setdiff1d(indicesFinal, manual_exclude_lists[insertion])
        insertion+=1
        spike_times_good = spike_times_beh[indicesFinal]
        print(f"Subject: {i}, insertion: {s}, Number of good neurons: {len(spike_times_good)}")

        runList = indicesFinal
        prod_rel_start = -0.10
        prod_rel_end = 0
        sens_rel_start = 0
        sens_rel_end = 0.10

        # Initialize stats for this insertion
        prod_tstat = np.zeros(len(runList))
        prod_pval = np.zeros(len(runList))
        sens_tstat = np.zeros(len(runList))
        sens_pval = np.zeros(len(runList))
        prod_sig_idx = np.array([], dtype=int)
        sens_sig_idx = np.array([], dtype=int)
        both_sig_idx = np.array([], dtype=int)
        num_prod_sig = 0
        num_sens_sig = 0
        num_both_sig = 0
        perc_prod_sig = 0
        perc_sens_sig = 0
        perc_both_sig = 0
        prod_neuron_indices = np.array([], dtype=int)
        sens_neuron_indices = np.array([], dtype=int)

        # --- PROD ANALYSIS ---
        if "ProdSpeechWords" in data:
            speech_prod_times = data["ProdSpeechWords"]
            prod_onsets = nap.Ts(speech_prod_times.start)
            prod_offsets = nap.Ts(speech_prod_times.end)
            prod_starts = prod_onsets.as_series().index + prod_rel_start
            prod_ends = prod_offsets.as_series().index + prod_rel_end

            for ii, neuron in enumerate(runList):
                spikes = spike_times_good[neuron]
                prod_firing_rates = np.zeros(len(prod_starts))
                threshold_firing_rate = spike_times_good[neuron].rate 
                for u in range(len(prod_starts)):
                    prod_interval = nap.IntervalSet(start=prod_starts[u], end=prod_ends[u])
                    spikes_this = spikes.restrict(prod_interval)
                    prod_count = len(spikes_this)
                    prod_rate = prod_count / (prod_ends[u] - prod_starts[u])    
                    prod_firing_rates[u] = prod_rate
                # Only run ttest if there are at least 2 samples
                if len(prod_firing_rates) > 1:
                    t_stat, p_val = ttest_rel(prod_firing_rates, np.full_like(prod_firing_rates, threshold_firing_rate), alternative='greater')
                else:
                    t_stat, p_val = np.nan, np.nan
                prod_tstat[ii] = t_stat
                prod_pval[ii] = p_val

            # Find indices where prod_pval < 0.05
            prod_sig_idx = np.where(prod_pval < 0.05)[0]
            num_prod_sig = len(prod_sig_idx)
            prod_neuron_indices = runList[prod_sig_idx]
            num_neurons = len(runList)
            perc_prod_sig = 100 * num_prod_sig / num_neurons if num_neurons > 0 else 0
            print(f"Percentage of neurons responsive to prod: {perc_prod_sig:.2f}%")
            print(f"Number of neurons responsive to prod: {num_prod_sig}")
        else:
            prod_sig_idx = np.array([], dtype=int)
            num_prod_sig = np.nan
            perc_prod_sig = np.nan
            prod_neuron_indices = np.array([], dtype=int)
            perc_prod_sig = np.nan
            print(f"Subject: {i}, insertion: {s} -- No ProdSpeechWords found, skipping prod analysis.")

        # --- SENS ANALYSIS ---
        if ("StimSpeechWords" in data) or ("mfa_stim_words" in data):
            if "StimSpeechWords" in data:
                speech_sens_times = data["StimSpeechWords"]
            else:
                speech_sens_times = data["mfa_stim_words"]
                          
            sens_onsets = nap.Ts(speech_sens_times.start)
            sens_offsets = nap.Ts(speech_sens_times.end)
            sens_starts = sens_onsets.as_series().index + sens_rel_start
            sens_ends = sens_offsets.as_series().index + sens_rel_end

            for ii, neuron in enumerate(runList):
                spikes = spike_times_good[neuron]
                sens_firing_rates = np.zeros(len(sens_starts))
                threshold_firing_rate = spike_times_good[neuron].rate 
                for u in range(len(sens_starts)):
                    sens_interval = nap.IntervalSet(start=sens_starts[u], end=sens_ends[u])
                    spikes_this = spikes.restrict(sens_interval)
                    sens_count = len(spikes_this)
                    sens_rate = sens_count / (sens_ends[u] - sens_starts[u])
                    sens_firing_rates[u] = sens_rate
                # Only run ttest if there are at least 2 samples
                if len(sens_firing_rates) > 1:
                    t_stat, p_val = ttest_rel(sens_firing_rates, np.full_like(sens_firing_rates, threshold_firing_rate), alternative='greater')
                else:
                    t_stat, p_val = np.nan, np.nan
                sens_tstat[ii] = t_stat
                sens_pval[ii] = p_val

            # Find indices where sens_pval < 0.05
            sens_sig_idx = np.where(sens_pval < 0.05)[0]
            num_sens_sig = len(sens_sig_idx)
            sens_neuron_indices = runList[sens_sig_idx]
            num_neurons = len(runList)
            perc_sens_sig = 100 * num_sens_sig / num_neurons if num_neurons > 0 else 0
            print(f"Percentage of neurons responsive to sens: {perc_sens_sig:.2f}%")
            print(f"Number of neurons responsive to sens: {num_sens_sig}")
        else:
            sens_sig_idx = np.array([], dtype=int)
            num_sens_sig = np.nan
            perc_sens_sig = np.nan
            sens_neuron_indices = np.array([], dtype=int)
            perc_sens_sig = np.nan
            print(f"Subject: {i}, insertion: {s} -- No StimSpeechWords found, skipping sens analysis.")

        # --- BOTH ANALYSIS ---
        # Only compute both if both prod and sens were run
        if ("ProdSpeechWords" in data) and ("StimSpeechWords" in data):
            both_sig_idx = np.intersect1d(prod_sig_idx, sens_sig_idx)
            num_both_sig = len(both_sig_idx)
            perc_both_sig = 100 * num_both_sig / num_neurons if num_neurons > 0 else 0
        else:
            both_sig_idx = np.array([], dtype=int)
            num_both_sig = 0
            perc_both_sig = 0

        # now save out all the data
        prod_neuron_perc_list.append(perc_prod_sig)
        sens_neuron_perc_list.append(perc_sens_sig)
        both_neuron_perc_list.append(perc_both_sig)
        prod_neuron_ct_list.append(num_prod_sig)
        sens_neuron_ct_list.append(num_sens_sig)
        both_neuron_ct_list.append(num_both_sig)
        total_neuron_ct_list.append(len(runList))
        prod_neuron_indices_list.append(prod_neuron_indices)
        sens_neuron_indices_list.append(sens_neuron_indices)
```

```python
# this version uses all the data not use the pairwise data - i think this will likely be the one to use in the final plot, still thinking, result is the same but just more data here
# Create bar plot with 2 groups of 2 bars each, showing percentages
import scipy.stats as stats
from scipy.stats import chi2_contingency, fisher_exact
import matplotlib
import matplotlib.pyplot as plt
import numpy as np

# Configure matplotlib for Jupyter notebook display
%matplotlib inline
plt.rcParams['figure.figsize'] = (12, 8)
plt.rcParams['svg.fonttype'] = 'none'  # Ensures text is editable in Illustrator

# Function to perform significance tests
def perform_significance_tests(set1_matches, set2_matches, set3_matches, opercular_condition):
    print(f"\n=== SIGNIFICANCE TESTS FOR OPERCULAR = {opercular_condition} ===")

    def get_counts(indices):
        prod_indices = [i for i in indices if not np.isnan(prod_neuron_ct_list[i])]
        prod_total = sum(total_neuron_ct_list[i] for i in prod_indices)
        prod_count = sum(prod_neuron_ct_list[i] for i in prod_indices)
        sens_indices = [i for i in indices if not np.isnan(sens_neuron_ct_list[i])]
        sens_total = sum(total_neuron_ct_list[i] for i in sens_indices)
        sens_count = sum(sens_neuron_ct_list[i] for i in sens_indices)
        return prod_count, prod_total, sens_count, sens_total

    prod1, total1_prod, sens1, total1_sens = get_counts(set1_matches)
    prod2, total2_prod, sens2, total2_sens = get_counts(set2_matches)
    prod3, total3_prod, sens3, total3_sens = get_counts(set3_matches)

    print(f"Set1 (vPrCG/parsTr/parsOp): Prod={prod1}/{total1_prod}, Sens={sens1}/{total1_sens}")
    print(f"Set2 (pSTG): Prod={prod2}/{total2_prod}, Sens={sens2}/{total2_sens}")
    print(f"Set3 (Other): Prod={prod3}/{total3_prod}, Sens={sens3}/{total3_sens}")

    results = {}

    print("\n1. PROD VS SENS WITHIN EACH REGION GROUP:")

    if total1_prod > 0 and total1_sens > 0:
        contingency1 = [
            [prod1, total1_prod - prod1],
            [sens1, total1_sens - sens1]
        ]
        chi2_1, p_val_1, dof_1, expected_1 = chi2_contingency(contingency1)
        print(f"   Set1 (vPrCG/parsTr/parsOp): χ²={chi2_1:.4f}, p={p_val_1:.4f}")
        results['set1_prod_vs_sens'] = {'chi2': chi2_1, 'p_val': p_val_1}

    if total2_prod > 0 and total2_sens > 0:
        contingency2 = [
            [prod2, total2_prod - prod2],
            [sens2, total2_sens - sens2]
        ]
        chi2_2, p_val_2, dof_2, expected_2 = chi2_contingency(contingency2)
        print(f"   Set2 (pSTG): χ²={chi2_2:.4f}, p={p_val_2:.4f}")
        results['set2_prod_vs_sens'] = {'chi2': chi2_2, 'p_val': p_val_2}

    if total3_prod > 0 and total3_sens > 0:
        contingency3 = [
            [prod3, total3_prod - prod3],
            [sens3, total3_sens - sens3]
        ]
        chi2_3, p_val_3, dof_3, expected_3 = chi2_contingency(contingency3)
        print(f"   Set3 (Other): χ²={chi2_3:.4f}, p={p_val_3:.4f}")
        results['set3_prod_vs_sens'] = {'chi2': chi2_3, 'p_val': p_val_3}

    print("\n2. PROD: vPrCG/parsTr/parsOp vs pSTG:")
    if total1_prod > 0 and total2_prod > 0:
        contingency_prod = [
            [prod1, total1_prod - prod1],
            [prod2, total2_prod - prod2]
        ]
        chi2_prod, p_val_prod, dof_prod, expected_prod = chi2_contingency(contingency_prod)
        print(f"   χ²={chi2_prod:.4f}, p={p_val_prod:.4f}")
        results['prod_set1_vs_set2'] = {'chi2': chi2_prod, 'p_val': p_val_prod}

    print("\n3. SENS: vPrCG/parsTr/parsOp vs pSTG:")
    if total1_sens > 0 and total2_sens > 0:
        contingency_sens = [
            [sens1, total1_sens - sens1],
            [sens2, total2_sens - sens2]
        ]
        chi2_sens, p_val_sens, dof_sens, expected_sens = chi2_contingency(contingency_sens)
        print(f"   χ²={chi2_sens:.4f}, p={p_val_sens:.4f}")
        results['sens_set1_vs_set2'] = {'chi2': chi2_sens, 'p_val': p_val_sens}

    return results

def add_significance_markers(ax, results, x_positions, y_max):
    y_offset = 0.15
    line_height = 0.02

    for i, (x_pos, set_name) in enumerate(zip(x_positions, ['set1', 'set2', 'set3'])):
        key = f'{set_name}_prod_vs_sens'
        if key in results and results[key]['p_val'] < 0.05:
            y_start = y_max + y_offset
            ax.plot([x_pos - 0.15, x_pos + 0.15], [y_start, y_start], 'k-', linewidth=1)
            ax.plot([x_pos - 0.15, x_pos - 0.15], [y_start - line_height, y_start], 'k-', linewidth=1)
            ax.plot([x_pos + 0.15, x_pos + 0.15], [y_start - line_height, y_start], 'k-', linewidth=1)
            p_val = results[key]['p_val']
            if p_val < 0.001:
                marker = '***'
            elif p_val < 0.01:
                marker = '**'
            else:
                marker = '*'
            ax.text(x_pos, y_start + 0.02, marker, ha='center', va='bottom', fontsize=12, fontweight='bold')

    if 'prod_set1_vs_set2' in results and results['prod_set1_vs_set2']['p_val'] < 0.05:
        y_start = y_max + y_offset + 4.0
        ax.plot([x_positions[0] - 0.15, x_positions[1] - 0.15], [y_start, y_start], 'k-', linewidth=1)
        ax.plot([x_positions[0] - 0.15, x_positions[0] - 0.15], [y_start - line_height, y_start], 'k-', linewidth=1)
        ax.plot([x_positions[1] - 0.15, x_positions[1] - 0.15], [y_start - line_height, y_start], 'k-', linewidth=1)
        p_val = results['prod_set1_vs_set2']['p_val']
        if p_val < 0.001:
            marker = '***'
        elif p_val < 0.01:
            marker = '**'
        else:
            marker = '*'
        ax.text((x_positions[0] + x_positions[1])/2 - 0.15, y_start + 0.02, marker, ha='center', va='bottom', fontsize=12, fontweight='bold')

    if 'sens_set1_vs_set2' in results and results['sens_set1_vs_set2']['p_val'] < 0.05:
        y_start = y_max + y_offset + 8.0
        ax.plot([x_positions[0] + 0.15, x_positions[1] + 0.15], [y_start, y_start], 'k-', linewidth=1)
        ax.plot([x_positions[0] + 0.15, x_positions[0] + 0.15], [y_start - line_height, y_start], 'k-', linewidth=1)
        ax.plot([x_positions[1] + 0.15, x_positions[1] + 0.15], [y_start - line_height, y_start], 'k-', linewidth=1)
        p_val = results['sens_set1_vs_set2']['p_val']
        if p_val < 0.001:
            marker = '***'
        elif p_val < 0.01:
            marker = '**'
        else:
            marker = '*'
        ax.text((x_positions[0] + x_positions[1])/2 + 0.15, y_start + 0.02, marker, ha='center', va='bottom', fontsize=12, fontweight='bold')

# Filter for opercular=0 and target regions
opercular_0_indices = [i for i, val in enumerate(opercular_list) if val == 0]
target_regions_set1 = ["vPrCG", "parsTr", "parsOp"]
target_regions_set2 = ["pSTG"]

set1_matches = [i for i in opercular_0_indices if region_list[i] in target_regions_set1]
set2_matches = [i for i in opercular_0_indices if region_list[i] in target_regions_set2]
set3_matches = [i for i in opercular_0_indices if region_list[i] not in target_regions_set1 and region_list[i] not in target_regions_set2]

print(f"Set1 matches (vPrCG/parsTr/parsOp): {set1_matches}")
print(f"Set2 matches (pSTG): {set2_matches}")
print(f"Set3 matches (other regions): {set3_matches}")

# Print all region names for opercular=0
print(f"\nAll regions for opercular=0:")
for i in opercular_0_indices:
    print(f"  Index {i}: {region_list[i]}")

print(f"\nSet1 regions (vPrCG/parsTr/parsOp):")
for i in set1_matches:
    print(f"  Index {i}: {region_list[i]}")

print(f"\nSet2 regions (pSTG):")
for i in set2_matches:
    print(f"  Index {i}: {region_list[i]}")

print(f"\nSet3 regions (other):")
for i in set3_matches:
    print(f"  Index {i}: {region_list[i]}")

# Calculate percentages and fractions for each set
def get_percentages_for_set(indices):
    # For prod: use insertions with prod data
    prod_indices = [i for i in indices if not np.isnan(prod_neuron_ct_list[i])]
    prod_total = sum(total_neuron_ct_list[i] for i in prod_indices)
    prod_count = sum(prod_neuron_ct_list[i] for i in prod_indices)
    prod_percentage = (prod_count / prod_total * 100) if prod_total > 0 else 0

    # For sens: use insertions with sens data
    sens_indices = [i for i in indices if not np.isnan(sens_neuron_ct_list[i])]
    sens_total = sum(total_neuron_ct_list[i] for i in sens_indices)
    sens_count = sum(sens_neuron_ct_list[i] for i in sens_indices)
    sens_percentage = (sens_count / sens_total * 100) if sens_total > 0 else 0

    return [prod_percentage, sens_percentage]

def get_fractions_for_set(indices):
    # For prod: use insertions with prod data
    prod_indices = [i for i in indices if not np.isnan(prod_neuron_ct_list[i])]
    prod_total = sum(total_neuron_ct_list[i] for i in prod_indices)
    prod_count = sum(prod_neuron_ct_list[i] for i in prod_indices)
    prod_fraction = f"{int(prod_count)}/{int(prod_total)}"

    # For sens: use insertions with sens data
    sens_indices = [i for i in indices if not np.isnan(sens_neuron_ct_list[i])]
    sens_total = sum(total_neuron_ct_list[i] for i in sens_indices)
    sens_count = sum(sens_neuron_ct_list[i] for i in sens_indices)
    sens_fraction = f"{int(sens_count)}/{int(sens_total)}"

    return [prod_fraction, sens_fraction]

percentages_set1 = get_percentages_for_set(set1_matches)
percentages_set2 = get_percentages_for_set(set2_matches)
percentages_set3 = get_percentages_for_set(set3_matches)

fractions_set1 = get_fractions_for_set(set1_matches)
fractions_set2 = get_fractions_for_set(set2_matches)
fractions_set3 = get_fractions_for_set(set3_matches)

print(f"Set 1 percentages (prod, sens): {percentages_set1}")
print(f"Set 2 percentages (prod, sens): {percentages_set2}")
print(f"Set 3 percentages (prod, sens): {percentages_set3}")
print(f"Set 1 fractions (prod, sens): {fractions_set1}")
print(f"Set 2 fractions (prod, sens): {fractions_set2}")
print(f"Set 3 fractions (prod, sens): {fractions_set3}")

# Perform significance tests for opercular=0
results_op0 = perform_significance_tests(set1_matches, set2_matches, set3_matches, 0)

# Create bar plot with regions on x-axis and color for prod/sens
regions = ['vPrCG/parsTr/parsOp', 'pSTG', 'Other Regions']
x = np.arange(len(regions))
width = 0.35  # width of the bars

fig, ax = plt.subplots(figsize=(12, 8))
ax.set_box_aspect(0.5)  # Set plot aspect ratio to 5

# Prod bars
prod_values = [percentages_set1[0], percentages_set2[0], percentages_set3[0]]
rects1 = ax.bar(x - width/2, prod_values, width, label='Prod', color='skyblue', alpha=0.8)

# Sens bars
sens_values = [percentages_set1[1], percentages_set2[1], percentages_set3[1]]
rects2 = ax.bar(x + width/2, sens_values, width, label='Sens', color='lightcoral', alpha=0.8)

# Add labels, title, legend
ax.set_ylabel('Percentage of Neurons (%)')
ax.set_title('Percentage of Responsive Neurons by Region and Type (opercular=0)')
ax.set_xticks(x)
ax.set_xticklabels(regions)
ax.legend()
ax.set_ylim(0, 100)  # Set y-axis range from -5% to 100%

# Annotate bars with fractions
def autolabel(rects, fractions, ax_handle):
    for i, rect in enumerate(rects):
        height = rect.get_height()
        ax_handle.annotate(fractions[i],
                    xy=(rect.get_x() + rect.get_width() / 2, height),
                    xytext=(0, 3),  # 3 points vertical offset
                    textcoords="offset points",
                    ha='center', va='bottom')

autolabel(rects1, [fractions_set1[0], fractions_set2[0], fractions_set3[0]], ax)
autolabel(rects2, [fractions_set1[1], fractions_set2[1], fractions_set3[1]], ax)

# Add individual lines for each insertion (without points)
def add_individual_fractions_with_lines(ax, indices, group_idx, bar_width=0.35):
    # For each insertion, plot a line from prod to sens if both are available, else plot a point for whichever is available
    for ins_idx in indices:
        total = total_neuron_ct_list[ins_idx]
        prod = prod_neuron_ct_list[ins_idx]
        sens = sens_neuron_ct_list[ins_idx]
        has_prod = not np.isnan(prod)
        has_sens = not np.isnan(sens)
        if total > 0:
            if has_prod and has_sens:
                prod_frac = prod / total * 100
                sens_frac = sens / total * 100
                x_prod = group_idx - bar_width/2
                x_sens = group_idx + bar_width/2
                ax.plot([x_prod, x_sens], [prod_frac, sens_frac], color='gray', alpha=0.6, linewidth=1.5, zorder=2)
            elif has_prod:
                prod_frac = prod / total * 100
                x_prod = group_idx - bar_width/2
                ax.plot(x_prod, prod_frac, marker='o', color='gray', alpha=0.6, zorder=2)
            elif has_sens:
                sens_frac = sens / total * 100
                x_sens = group_idx + bar_width/2
                ax.plot(x_sens, sens_frac, marker='o', color='gray', alpha=0.6, zorder=2)

# Add individual lines for each group
add_individual_fractions_with_lines(ax, set1_matches, 0, width)
add_individual_fractions_with_lines(ax, set2_matches, 1, width)
add_individual_fractions_with_lines(ax, set3_matches, 2, width)

# Add significance markers
y_max = max(max(prod_values), max(sens_values))
add_significance_markers(ax, results_op0, x, y_max)

plt.tight_layout()
fig.savefig("responsiveFxBar_opercular0.svg", format="svg", dpi=300, bbox_inches="tight")
plt.show()  # Show the plot

# Now create the same plot for opercular=1
print("\n" + "="*50)
print("OPERCULAR = 1")
print("="*50)

# Filter for opercular=1 and target regions
opercular_1_indices = [i for i, val in enumerate(opercular_list) if val == 1]

set1_matches_op1 = [i for i in opercular_1_indices if region_list[i] in target_regions_set1]
set2_matches_op1 = [i for i in opercular_1_indices if region_list[i] in target_regions_set2]
set3_matches_op1 = [i for i in opercular_1_indices if region_list[i] not in target_regions_set1 and region_list[i] not in target_regions_set2]

print(f"Set1 matches (vPrCG/parsTr/parsOp): {set1_matches_op1}")
print(f"Set2 matches (pSTG): {set2_matches_op1}")
print(f"Set3 matches (other regions): {set3_matches_op1}")

# Calculate percentages and fractions for each set (opercular=1)
percentages_set1_op1 = get_percentages_for_set(set1_matches_op1)
percentages_set2_op1 = get_percentages_for_set(set2_matches_op1)
percentages_set3_op1 = get_percentages_for_set(set3_matches_op1)

fractions_set1_op1 = get_fractions_for_set(set1_matches_op1)
fractions_set2_op1 = get_fractions_for_set(set2_matches_op1)
fractions_set3_op1 = get_fractions_for_set(set3_matches_op1)

print(f"Set 1 percentages (prod, sens): {percentages_set1_op1}")
print(f"Set 2 percentages (prod, sens): {percentages_set2_op1}")
print(f"Set 3 percentages (prod, sens): {percentages_set3_op1}")
print(f"Set 1 fractions (prod, sens): {fractions_set1_op1}")
print(f"Set 2 fractions (prod, sens): {fractions_set2_op1}")
print(f"Set 3 fractions (prod, sens): {fractions_set3_op1}")

# Perform significance tests for opercular=1
results_op1 = perform_significance_tests(set1_matches_op1, set2_matches_op1, set3_matches_op1, 1)

# Create second bar plot for opercular=1
fig2, ax2 = plt.subplots(figsize=(12, 8))
ax2.set_box_aspect(0.5)  # Set plot aspect ratio to 5

# Prod bars
prod_values_op1 = [percentages_set1_op1[0], percentages_set2_op1[0], percentages_set3_op1[0]]
rects1_op1 = ax2.bar(x - width/2, prod_values_op1, width, label='Prod', color='skyblue', alpha=0.8)

# Sens bars
sens_values_op1 = [percentages_set1_op1[1], percentages_set2_op1[1], percentages_set3_op1[1]]
rects2_op1 = ax2.bar(x + width/2, sens_values_op1, width, label='Sens', color='lightcoral', alpha=0.8)

# Add labels, title, legend
ax2.set_ylabel('Percentage of Neurons (%)')
ax2.set_title('Percentage of Responsive Neurons by Region and Type (opercular=1)')
ax2.set_xticks(x)
ax2.set_xticklabels(regions)
ax2.legend()
ax2.set_ylim(0, 100)  # Set y-axis range from -5% to 100%

# Annotate bars with fractions
autolabel(rects1_op1, [fractions_set1_op1[0], fractions_set2_op1[0], fractions_set3_op1[0]], ax2)
autolabel(rects2_op1, [fractions_set1_op1[1], fractions_set2_op1[1], fractions_set3_op1[1]], ax2)

# Add individual lines for each group (no points)
add_individual_fractions_with_lines(ax2, set1_matches_op1, 0, width)
add_individual_fractions_with_lines(ax2, set2_matches_op1, 1, width)
add_individual_fractions_with_lines(ax2, set3_matches_op1, 2, width)

# Add significance markers
y_max_op1 = max(max(prod_values_op1), max(sens_values_op1))
add_significance_markers(ax2, results_op1, x, y_max_op1)

plt.tight_layout()
fig2.savefig("responsiveFxBar_opercular1.svg", format="svg", dpi=300, bbox_inches="tight")
plt.show()  # Show the plot
```

```python
# BRESLOW-DAY TEST FOR INTERACTION
# Tests for homogeneity of odds ratios across region types
# This is the proper test for interaction using odds ratios

import numpy as np
from scipy.stats import chi2

print("="*60)
print("BRESLOW-DAY TEST FOR INTERACTION")
print("="*60)

# Data from opercular=0 analysis:
# prod region, prod task = 333/519
# prod region, sens task = 78/519
# sens region, prod task = 22/79
# sens region, sens task = 67/79

# Create 2x2 tables for each region
# Format: [prod task: resp, not resp], [sens task: resp, not resp]

prod_region_table = np.array([
    [333, 519 - 333],  # Prod task: responsive, not responsive
    [78, 519 - 78]      # Sens task: responsive, not responsive
])

sens_region_table = np.array([
    [22, 79 - 22],      # Prod task: responsive, not responsive
    [67, 79 - 67]       # Sens task: responsive, not responsive
])

# Calculate odds ratios for each region
def calc_or(a, b, c, d):
    """Calculate odds ratio from 2x2 table"""
    return (a * d) / (b * c) if (b > 0 and c > 0) else np.nan

a1, b1 = prod_region_table[0, 0], prod_region_table[0, 1]
c1, d1 = prod_region_table[1, 0], prod_region_table[1, 1]
or_prod = calc_or(a1, b1, c1, d1)

a2, b2 = sens_region_table[0, 0], sens_region_table[0, 1]
c2, d2 = sens_region_table[1, 0], sens_region_table[1, 1]
or_sens = calc_or(a2, b2, c2, d2)

print(f"\nProd region OR: {or_prod:.4f}")
print(f"Sens region OR: {or_sens:.4f}")
print(f"Ratio of ORs: {or_prod / or_sens:.4f}")

# Breslow-Day test: Test if log(OR1) = log(OR2)
# This tests for homogeneity of odds ratios (no interaction if equal)

log_or1 = np.log(or_prod)
log_or2 = np.log(or_sens)

# Variance of log OR: Var(log OR) = 1/a + 1/b + 1/c + 1/d
var_log_or1 = 1/a1 + 1/b1 + 1/c1 + 1/d1
var_log_or2 = 1/a2 + 1/b2 + 1/c2 + 1/d2

# Test statistic: Z = (log_or1 - log_or2) / sqrt(var1 + var2)
se_diff = np.sqrt(var_log_or1 + var_log_or2)
z_stat = (log_or1 - log_or2) / se_diff

# Chi-square = Z^2 (with 1 df)
# This is because: Z ~ N(0,1), so Z² ~ χ²(1)
# Both tests are equivalent - you can report either one
chi2_bd = z_stat**2

# Calculate p-value using higher precision methods
from scipy.stats import norm
from scipy.special import erfc
import numpy as np

# For very small p-values, use survival function (sf) instead of 1 - cdf
# This provides better numerical precision
p_z = 2 * norm.sf(abs(z_stat))  # Two-tailed z-test (sf = 1 - cdf, more precise)
p_chi2 = chi2.sf(chi2_bd, df=1)  # Chi-square test (sf = 1 - cdf, more precise)

# For extremely small values, also calculate using complementary error function
# erfc(x) = 2 * (1 - Φ(x√2)) for standard normal
# For Z-test: p = erfc(|z|/√2)
p_z_erfc = erfc(abs(z_stat) / np.sqrt(2))

# They should be identical (within rounding)
p_bd = p_chi2

print("\n" + "-"*60)
print("BRESLOW-DAY TEST RESULTS:")
print("-"*60)
print(f"Log OR (prod region): {log_or1:.4f}, variance: {var_log_or1:.6f}")
print(f"Log OR (sens region): {log_or2:.4f}, variance: {var_log_or2:.6f}")
print(f"\nDifference in log ORs: {log_or1 - log_or2:.4f}")
print(f"Standard error: {se_diff:.6f}")
print(f"\nZ-statistic: {z_stat:.4f}")
print(f"Chi-square statistic: {chi2_bd:.4f} (note: χ² = Z²)")
print(f"Degrees of freedom: 1")

# Display p-value in multiple formats with high precision
print(f"\np-value (from z-test, using sf): {p_z:.15e}")
print(f"p-value (from chi-square, using sf): {p_chi2:.15e}")
print(f"p-value (from erfc method): {p_z_erfc:.15e}")

# All three methods should give the same result (within numerical precision)
# Use the chi-square survival function as the primary value
p_bd = p_chi2

# Display in multiple formats
print(f"\np-value (decimal, 20 places): {p_bd:.20f}")
print(f"p-value (scientific, 15 places): {p_bd:.15e}")
print(f"p-value (scientific, 20 places): {p_bd:.20e}")

# For reporting
if p_bd < 1e-15:
    p_report = f"p < 0.0001"
    p_exact = f"p = {p_bd:.3e}"
elif p_bd < 1e-10:
    p_report = f"p < 0.0001"
    p_exact = f"p = {p_bd:.2e}"
elif p_bd < 0.001:
    p_report = f"p < 0.001"
    p_exact = f"p = {p_bd:.3e}"
else:
    p_report = f"p = {p_bd:.4f}"
    p_exact = p_report

if p_bd < 0.05:
    print(f"\n✓ Significant interaction detected")
    print(f"\n" + "="*60)
    print("FOR REPORTING (choose one format):")
    print("="*60)
    print(f"\nOption 1 (Z-test, with exact p-value):")
    print(f"  Breslow-Day test: Z = {z_stat:.4f}, {p_exact}")
    print(f"\nOption 2 (Z-test, with p < threshold):")
    print(f"  Breslow-Day test: Z = {z_stat:.4f}, {p_report}")
    print(f"\nOption 3 (Chi-square, with exact p-value):")
    print(f"  Breslow-Day test: χ²(1) = {chi2_bd:.4f}, {p_exact}")
    print(f"\nOption 4 (Chi-square, with p < threshold):")
    print(f"  Breslow-Day test: χ²(1) = {chi2_bd:.4f}, {p_report}")
    print(f"\nNote: These are equivalent since χ² = Z² and both have 1 degree of freedom")
    print(f"      The exact p-value is: {p_bd:.20e}")
else:
    print(f"\n✗ No significant interaction (p >= 0.05)")
    print(f"\nFOR REPORTING:")
    print(f"  Breslow-Day test: χ²(1) = {chi2_bd:.4f}, p = {p_bd:.4f}")

print("\n" + "="*60)
print("MATHEMATICAL EXPLANATION:")
print("="*60)
print("The relationship between Z and χ²:")
print("  - Z ~ N(0,1) (standard normal distribution)")
print("  - Z² ~ χ²(1) (chi-square with 1 degree of freedom)")
print("  - Therefore: χ² = Z² and both have the same p-value")
print("\nFor this test:")
print(f"  Z = {z_stat:.4f}")
print(f"  χ² = Z² = {z_stat:.4f}² = {chi2_bd:.4f}")
print(f"  Both tests give p = {p_bd:.3e}")

print("\n" + "="*60)
print("INTERPRETATION:")
print("="*60)
print("H0: OR_prod_region = OR_sens_region (no interaction)")
print("H1: OR_prod_region ≠ OR_sens_region (interaction exists)")
print("\nA significant result indicates that the effect of task type")
print("on responsiveness differs between region types (interaction).")

```
