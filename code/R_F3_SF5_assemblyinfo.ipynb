```python
# MAIN LOOP TO DEFINE NEURON AND ASSEMBLY INFORMATION CAPACITY
import fnmatch
import matplotlib as mpl
import matplotlib.pyplot as plt
import numpy as np
import re
import os
from scipy import stats
from sklearn.decomposition import PCA, FastICA
from scipy.linalg import norm
from scipy.stats import kstest

MIN_BEHAVIOR_MINUTES = 0

def _behavior_duration_seconds(task_times_obj):
    starts = np.asarray(task_times_obj.start, dtype=float).ravel()
    ends = np.asarray(task_times_obj.end, dtype=float).ravel()
    if starts.size != ends.size:
        return np.nan
    return float(np.sum(ends - starts))

def _ks_time_bounds_full_recording(spike_times_obj):
    """Min/max time for uniform KS (full recording)."""
    tmin, tmax = np.inf, -np.inf
    for u in range(len(spike_times_obj)):
        idx = spike_times_obj[u].as_series().index.values
        if len(idx):
            tmin = min(tmin, float(np.min(idx)))
            tmax = max(tmax, float(np.max(idx)))
    if np.isfinite(tmin) and np.isfinite(tmax):
        return tmin, tmax
    return np.nan, np.nan

def imec_key_sorter(key):
    """Extract the number after 'imec', default to a large number if not found."""
    match = re.search(r'imec(\d+)', key)
    return int(match.group(1)) if match else float('inf')

def calculate_mutual_information(x, y, bins=20):
    """
    Calculate mutual information between two variables using histogram-based approach.
    """
    valid_mask = ~(np.isnan(x) | np.isnan(y))
    if np.sum(valid_mask) < 2:
        return np.nan

    x_valid = x[valid_mask]
    y_valid = y[valid_mask]

    hist_2d, _, _ = np.histogram2d(x_valid, y_valid, bins=bins)
    hist_x = np.sum(hist_2d, axis=1)
    hist_y = np.sum(hist_2d, axis=0)

    p_xy = hist_2d / np.sum(hist_2d)
    p_x = hist_x / np.sum(hist_x)
    p_y = hist_y / np.sum(hist_y)

    mi = 0.0
    for i in range(len(p_x)):
        for j in range(len(p_y)):
            if p_xy[i, j] > 0 and p_x[i] > 0 and p_y[j] > 0:
                mi += p_xy[i, j] * np.log2(p_xy[i, j] / (p_x[i] * p_y[j]))
    return mi

def calculate_information_capacity(data, time_bins, min_samples=10):
    """
    Calculate information storage capacity using multiple metrics.
    """
    valid_mask = ~np.isnan(data)
    if np.sum(valid_mask) < min_samples:
        return {
            'entropy': np.nan,
            'mutual_info_temporal': np.nan,
            'capacity': np.nan,
            'n_valid_samples': np.sum(valid_mask),
            'data_range': np.nan,
            'data_std': np.nan
        }

    valid_data = data[valid_mask]

    n_bins = min(20, len(valid_data) // 5)
    if n_bins < 5:
        n_bins = 5

    hist, _ = np.histogram(valid_data, bins=n_bins)
    prob = hist / np.sum(hist)
    prob = prob[prob > 0]

    if len(prob) < 2:
        entropy_val = np.nan
    else:
        entropy_val = -np.sum(prob * np.log2(prob))

    if len(valid_data) > 1:
        try:
            mi_temporal = calculate_mutual_information(
                valid_data[:-1],
                valid_data[1:],
                bins=min(10, len(valid_data) // 10)
            )
        except Exception:
            mi_temporal = np.nan
    else:
        mi_temporal = np.nan

    if not np.isnan(entropy_val) and not np.isnan(mi_temporal):
        capacity = entropy_val - mi_temporal
    elif not np.isnan(entropy_val):
        capacity = entropy_val
    else:
        capacity = np.nan

    data_range = np.max(valid_data) - np.min(valid_data) if len(valid_data) > 0 else np.nan
    data_std = np.std(valid_data) if len(valid_data) > 0 else np.nan

    return {
        'entropy': entropy_val,
        'mutual_info_temporal': mi_temporal,
        'capacity': capacity,
        'n_valid_samples': np.sum(valid_mask),
        'data_range': data_range,
        'data_std': data_std
    }

    # Correct expression strength calculation using outer product method
def calculate_expression_strength(firingRateMatrix, ap_norm_matrix):
    """
    Calculate expression strength using the method from the paper:
    E(b) = R(b)^T * Oi * R(b)
    where E(b) is the expression of assembly i in time bin b,
    R(b) is the firing rate matrix column for time bin b,
    and Oi is the outer product of the weight vector of assembly i.
    
    Parameters:
    - firingRateMatrix: shape (time_bins, neurons) - z-scored firing rates
    - ap_norm_matrix: shape (n_assemblies, neurons) - normalized assembly patterns
    
    Returns:
    - expression_strength: shape (time_bins, n_assemblies) - expression strength for each assembly at each time bin
    """
    
    n_time_bins, n_neurons = firingRateMatrix.shape
    n_assemblies = ap_norm_matrix.shape[0]
    
    expression_strength = np.zeros((n_time_bins, n_assemblies))
    
    for assembly_idx in range(n_assemblies):
        # Get the weight vector for this assembly
        weight_vector = ap_norm_matrix[assembly_idx, :]  # shape: (n_neurons,)
        
        # Calculate outer product Oi = weight_vector * weight_vector^T
        outer_product = np.outer(weight_vector, weight_vector)  # shape: (n_neurons, n_neurons)
        
        # Calculate expression strength for each time bin
        for time_bin in range(n_time_bins):
            # R(b) is the firing rate vector for time bin b
            firing_rate_vector = firingRateMatrix[time_bin, :]  # shape: (n_neurons,)
            
            # E(b) = R(b)^T * Oi * R(b)
            # This is equivalent to: firing_rate_vector^T @ outer_product @ firing_rate_vector
            expression_strength[time_bin, assembly_idx] = firing_rate_vector.T @ outer_product @ firing_rate_vector
    
    return expression_strength

# define variables to save out from loop
ap_norm_matrix_list = []
ap_norm_member_list = []
sparsity_list_list = []

# Information capacity variables
assembly_info_capacity_all_list = []
neuron_info_capacity_all_list = []
insertion = 0

for i in range(len(nwb_paths)):
    data = nap.load_file(nwb_paths[i])
    keys = data.keys()

    template = "*imec*"
    keys = [key for key in keys if fnmatch.fnmatch(key, template)]
    ks_keys = [key for key in keys if "KS4" in key]
    if ks_keys:
        keys = ks_keys
    th8_keys = [key for key in keys if "Th=8" in key]
    if th8_keys:
        keys = th8_keys
    else:
        th_keys = [key for key in keys if "Th=" in key]
        if th_keys:
            keys = th_keys
    template_sentgen = "*sentgen*"
    template_auto = "*_auto*"
    keys = [key for key in keys if not fnmatch.fnmatch(key, template_sentgen) and not fnmatch.fnmatch(key, template_auto)]
    # For NP137, remove 'imec1'
    if 'NP137' in str(nwb_paths[i]) or 'NP139_B2' in str(nwb_paths[i]):
        keys = [key for key in keys if 'imec1' not in key]
    keys = sorted(keys, key=imec_key_sorter)

    for s in range(len(keys)):
        print(keys[s])
        print(f"Region = {region_list[insertion]}")

        spike_times = data[keys[s]]

        firingRates_all = spike_times.metadata["rate"]
        if "TaskTimes" in data.keys():
            task_times = data["TaskTimes"]
        else:
            task_times = data["task_times"]

        start_time = task_times.start
        end_time = task_times.end
        beh_epochs = nap.IntervalSet(start=task_times.start, end=task_times.end)

        beh_sec = _behavior_duration_seconds(task_times)
        beh_min = beh_sec / 60.0 if np.isfinite(beh_sec) else np.nan
        use_full_recording = (not np.isfinite(beh_min)) or (beh_min < MIN_BEHAVIOR_MINUTES)

        if use_full_recording:
            spike_times_beh = spike_times
            firingRates_beh = firingRates_all
            min_time, max_time = _ks_time_bounds_full_recording(spike_times)
            if not (np.isfinite(min_time) and np.isfinite(max_time)) or (max_time <= min_time):
                print(f"  Short behavior ({beh_min:.2f} min) but could not get recording bounds; skipping shank")
                insertion += 1
                continue
            print(
                f"  Behavioral duration {beh_min:.2f} min (< {MIN_BEHAVIOR_MINUTES} min) — using full recording for filters; KS window [{min_time:.3f}, {max_time:.3f}] s"
            )
        else:
            spike_times_beh = spike_times.restrict(beh_epochs)
            firingRates_beh = spike_times_beh.metadata["rate"]
            min_time = start_time[0]
            max_time = end_time[-1]
            print(
                f"  Behavioral duration {beh_min:.2f} min (>= {MIN_BEHAVIOR_MINUTES} min) — using TaskTimes restrict; KS window [{min_time:.3f}, {max_time:.3f}] s"
            )

        ks_stats = np.zeros(len(spike_times))
        ks_pvals = np.zeros(len(spike_times))
        for u in range(len(spike_times)):
            test = spike_times[u].as_series().index.values
            if len(test) > 1:
                normalized_spike_times = (test - min_time) / (max_time - min_time)
                ks_result = kstest(normalized_spike_times, 'uniform')
                ks_stats[u] = ks_result.statistic
                ks_pvals[u] = ks_result.pvalue
            else:
                ks_stats[u] = np.nan
                ks_pvals[u] = np.nan

        violationThreshold = 3 / 1000
        violationPct = np.zeros(len(spike_times))
        for u in range(len(spike_times)):
            unit = spike_times[u]
            unit = unit.as_series().index
            if len(unit) < 100:
                violationPct[u] = 1
            else:
                isi = unit.diff()[1:]
                violations = np.where(isi < violationThreshold)
                violations = np.array(violations)
                violationPct[u] = violations.size / len(isi)

        if "KSLabel" in spike_times.metadata:
            KSLabels = spike_times.metadata["KSLabel"]
        else:
            KSLabels = spike_times.metadata["quality"]
        firingRates = firingRates_beh
        mask1 = violationPct < 3 / 100
        mask2 = firingRates > 0.5
        mask3 = KSLabels != "noise"
        mask4 = ks_stats < 0.3
        mask = mask1 & mask2 & mask3 & mask4
        indicesFinal = firingRates.index[mask]
        # now exclude neurons manually rejected
        indicesFinal = np.setdiff1d(indicesFinal, manual_exclude_lists[insertion])
        insertion+=1
        spike_times_good = spike_times[indicesFinal]
        print(f"Subject: {i}, insertion: {s}, Number of good neurons: {len(spike_times_good)}")

        timescale = 25 / 1000
        spikeCountMatrix = spike_times_good.count(bin_size=timescale)
        bin_edges = spikeCountMatrix.index.values
        bin_centers = bin_edges[:-1] + np.diff(bin_edges) / 2
        if len(bin_centers) < len(bin_edges):
            last_width = bin_edges[-1] - bin_edges[-2] if len(bin_edges) > 1 else 0
            last_center = bin_edges[-1] - last_width / 2
            bin_centers = np.append(bin_centers, last_center)
        spikeCountMatrix = spikeCountMatrix.values
        firingRateMatrix = spikeCountMatrix / timescale
        firingRateMatrix = stats.zscore(firingRateMatrix, axis=0)

        pca = PCA()
        firingRateMatrix_pca = pca.fit_transform(firingRateMatrix)
        eigenvalues = pca.explained_variance_
        upperbound = (1 + np.sqrt(firingRateMatrix.shape[1] / firingRateMatrix.shape[0])) ** 2
        assemblyIndices = np.where(eigenvalues > upperbound)
        assemblyEigenvalues = eigenvalues[assemblyIndices]
        print(f"Number of assembly patterns: {len(assemblyEigenvalues)}")

        n_pcs = len(assemblyEigenvalues)
        pc_vectors = pca.components_[:n_pcs, :]

        projections = firingRateMatrix @ pc_vectors.T

        if n_pcs > 0:
            fastica = FastICA(
                n_components=n_pcs,
                algorithm='parallel',
                whiten='unit-variance',
                max_iter=500,
                tol=1e-7,
                random_state=1
            )
            ica_components = fastica.fit_transform(projections)
            mixing_matrix = fastica.mixing_
            unmixing_matrix = fastica.components_
            ica_assembly_patterns = unmixing_matrix @ pc_vectors

            ap_norm_matrix = ica_assembly_patterns
            ap_norm_member = np.zeros(ica_assembly_patterns.shape)
            for ii in range(n_pcs):
                ap = ica_assembly_patterns[ii, :]
                ap_norm = ap / norm(ap)
                maxW = np.max(ap_norm)
                minW = np.min(ap_norm)
                if abs(minW) > maxW:
                    ap_norm = ap_norm * -1
                ap_norm_matrix[ii, :] = ap_norm
                threshold = np.mean(ap_norm) + np.std(ap_norm) * 2
                for c in range(len(ap)):
                    if ap_norm[c] > threshold:
                        ap_norm_member[ii, c] = 1

            sparsity_list = []
            for ap in range(len(ap_norm_matrix)):
                w = np.array(ap_norm_matrix[ap, :])
                n = w.size
                numerator = np.sqrt(n) - np.sum(np.abs(w))
                denominator = np.sqrt(n) - 1
                sparsity = 1 - numerator / denominator if denominator != 0 else np.nan
                sparsity_list.append(sparsity)

            expression_strength = calculate_expression_strength(firingRateMatrix, ap_norm_matrix) # this calculates for all assemblies

            # --- INFORMATION CAPACITY ANALYSIS ---
            assembly_info_capacity_all = {}
            neuron_info_capacity_all = {}

            if n_pcs > 0:
                for assembly_idx in range(n_pcs):
                    assembly_expression = ica_components[:, assembly_idx]
                    #assembly_expression = expression_strength[:,assembly_idx] # use expression strength rather than ica projections
                    all_time_data = assembly_expression
                    if len(all_time_data) > 0:
                        assembly_info_capacity_all[f'assembly_{assembly_idx}'] = calculate_information_capacity(
                            all_time_data, bin_centers
                        )
                    else:
                        assembly_info_capacity_all[f'assembly_{assembly_idx}'] = {
                            'entropy': np.nan, 'mutual_info_temporal': np.nan, 'capacity': np.nan,
                            'n_valid_samples': 0, 'data_range': np.nan, 'data_std': np.nan
                        }

                for neuron_idx in range(firingRateMatrix.shape[1]):
                    neuron_activity = firingRateMatrix[:, neuron_idx]
                    all_time_data = neuron_activity
                    if len(all_time_data) > 0:
                        neuron_info_capacity_all[f'neuron_{neuron_idx}'] = calculate_information_capacity(
                            all_time_data, bin_centers
                        )
                    else:
                        neuron_info_capacity_all[f'neuron_{neuron_idx}'] = {
                            'entropy': np.nan, 'mutual_info_temporal': np.nan, 'capacity': np.nan,
                            'n_valid_samples': 0, 'data_range': np.nan, 'data_std': np.nan
                        }

            ap_norm_matrix_list.append(ap_norm_matrix)
            ap_norm_member_list.append(ap_norm_member)
            sparsity_list_list.append(sparsity_list)
            assembly_info_capacity_all_list.append(assembly_info_capacity_all)
            neuron_info_capacity_all_list.append(neuron_info_capacity_all)
        else:
            ap_norm_matrix_list.append(np.array([]))
            ap_norm_member_list.append(np.array([]))
            sparsity_list_list.append(np.array([]))
            assembly_info_capacity_all_list.append({})
            neuron_info_capacity_all_list.append({})
```

```python
# GLOBAL MUTUAL INFORMATION BASED YIELD CORRECTION
import numpy as np
from sklearn.feature_selection import mutual_info_regression
import matplotlib.pyplot as plt
from scipy.stats import pearsonr

plt.rcParams['svg.fonttype'] = 'none'

def calculate_global_mutual_information_with_yield(capacity_values, yield_values, bins=20, min_samples=10):
    """
    Calculate mutual information between capacity and yield globally across all assemblies

    Parameters:
    -----------
    capacity_values : list
        List of capacity values from all assemblies
    yield_values : list  
        List of yield values from all assemblies
    bins : int
        Number of bins for discretization
    min_samples : int
        Minimum samples required for reliable MI calculation

    Returns:
    --------
    float
        Mutual information between capacity and yield
    """
    if len(capacity_values) < min_samples:
        return 0.0

    # Convert to numpy arrays and ensure numeric types
    try:
        capacity_array = np.array(capacity_values, dtype=float)
        yield_array = np.array(yield_values, dtype=float)
    except (ValueError, TypeError):
        # If conversion fails, filter out non-numeric values
        capacity_array = []
        yield_array = []
        for cap, yld in zip(capacity_values, yield_values):
            try:
                cap_float = float(cap)
                yld_float = float(yld)
                capacity_array.append(cap_float)
                yield_array.append(yld_float)
            except (ValueError, TypeError):
                continue
        capacity_array = np.array(capacity_array)
        yield_array = np.array(yield_array)

    # Remove any NaN values
    valid_mask = ~(np.isnan(capacity_array) | np.isnan(yield_array))
    capacity_clean = capacity_array[valid_mask]
    yield_clean = yield_array[valid_mask]

    if len(capacity_clean) < min_samples:
        return 0.0

    # Calculate mutual information using scikit-learn
    yield_reshaped = yield_clean.reshape(-1, 1)
    mi = mutual_info_regression(yield_reshaped, capacity_clean, discrete_features=False, random_state=42)
    return mi[0]  # Return the MI value

# Calculate global Method 4 (MI with yield) across all assemblies
print("Calculating Global Method 4: Mutual Information with Yield")
print("=" * 60)

# Debug: Check the data structure
print("Debugging data structure:")
print(f"assembly_info_capacity_all_list type: {type(assembly_info_capacity_all_list)}")
print(f"assembly_info_capacity_all_list length: {len(assembly_info_capacity_all_list)}")
print(f"yield_list type: {type(yield_list)}")
print(f"yield_list length: {len(yield_list)}")

# Check first few items
for i in range(min(3, len(assembly_info_capacity_all_list))):
    print(f"Insertion {i}: {type(assembly_info_capacity_all_list[i])}")
    if isinstance(assembly_info_capacity_all_list[i], dict):
        print(f"  Dict with {len(assembly_info_capacity_all_list[i])} assemblies")
        if len(assembly_info_capacity_all_list[i]) > 0:
            first_key = list(assembly_info_capacity_all_list[i].keys())[0]
            first_value = assembly_info_capacity_all_list[i][first_key]
            print(f"  First assembly value: {first_value} (type: {type(first_value)})")
    else:
        print(f"  Not a dict: {assembly_info_capacity_all_list[i]}")

# Collect all capacity and yield data across all insertions
all_capacity_values = []
all_yield_values = []
all_assembly_info = []

for insertion_idx, insertion_data in enumerate(assembly_info_capacity_all_list):
    if isinstance(insertion_data, dict) and len(insertion_data) > 0:
        insertion_yield = yield_list[insertion_idx]
        for assembly_key, assembly_data in insertion_data.items():
            if isinstance(assembly_data, dict) and 'capacity' in assembly_data:
                capacity_value = assembly_data['capacity']
                try:
                    capacity_float = float(capacity_value)
                    if not np.isnan(capacity_float):
                        all_capacity_values.append(capacity_float)
                        all_yield_values.append(insertion_yield)
                        all_assembly_info.append({
                            'insertion_idx': insertion_idx,
                            'assembly_key': assembly_key,
                            'capacity': capacity_float,
                            'yield': insertion_yield,
                            'opercular': opercular_list[insertion_idx],
                            'pathology': path_list[insertion_idx],
                            'grade': grade_list[insertion_idx]
                        })
                except (ValueError, TypeError):
                    continue

print(f"Total assemblies across all insertions: {len(all_capacity_values)}")
print(f"Total insertions with data: {len(set([info['insertion_idx'] for info in all_assembly_info]))}")

# Calculate global mutual information
global_mi = calculate_global_mutual_information_with_yield(all_capacity_values, all_yield_values)
print(f"Global Mutual Information (Capacity vs Yield): {global_mi:.6f}")

# Calculate global Method 4 values (capacity - MI component)
global_method4_values = []
for info in all_assembly_info:
    adjusted_capacity = info['capacity'] - (global_mi * info['yield'] / np.mean(all_yield_values))
    global_method4_values.append(adjusted_capacity)

print(f"Global Method 4 values calculated: {len(global_method4_values)}")
print(f"Mean Global Method 4: {np.mean(global_method4_values):.6f}")
print(f"Std Global Method 4: {np.std(global_method4_values):.6f}")

# --- Significance calculations for correlations ---
corr_orig, pval_orig = pearsonr(all_yield_values, all_capacity_values)
corr_method4, pval_method4 = pearsonr(all_yield_values, global_method4_values)

# Create visualization comparing original capacity vs global Method 4
fig, axes = plt.subplots(2, 2, figsize=(15, 12))
fig.suptitle('Global Method 4: Mutual Information with Yield', fontsize=16, fontweight='bold')

# Plot 1: Original capacity vs yield (add line of best fit)
ax1 = axes[0, 0]
scatter = ax1.scatter(all_yield_values, all_capacity_values, alpha=0.6, c='blue')
ax1.set_xlabel('Neuron Yield')
ax1.set_ylabel('Original Assembly Capacity')
ax1.set_title('Original Capacity vs Yield')
ax1.grid(True, alpha=0.3)
# Add line of best fit
if len(all_yield_values) > 1:
    coeffs1 = np.polyfit(all_yield_values, all_capacity_values, 1)
    fit1_y = np.polyval(coeffs1, all_yield_values)
    # Sort for plotting line
    sorted_idx = np.argsort(all_yield_values)
    ax1.plot(np.array(all_yield_values)[sorted_idx], fit1_y[sorted_idx], color='black', lw=2, zorder=10, label="Best Fit")
# Add correlation info with significance
signif_str_orig = f"Correlation: {corr_orig:.4f}\n(p = {pval_orig:.2e})"
ax1.text(0.05, 0.95, signif_str_orig, transform=ax1.transAxes, 
         bbox=dict(boxstyle="round,pad=0.3", facecolor="white", alpha=0.8), va='top')

# Plot 2: Global Method 4 vs yield (add line of best fit)
ax2 = axes[0, 1]
scatter = ax2.scatter(all_yield_values, global_method4_values, alpha=0.6, c='red')
ax2.set_xlabel('Neuron Yield')
ax2.set_ylabel('Global Method 4 (Yield-Adjusted)')
ax2.set_title('Global Method 4 vs Yield')
ax2.grid(True, alpha=0.3)
# Add line of best fit
if len(all_yield_values) > 1:
    coeffs2 = np.polyfit(all_yield_values, global_method4_values, 1)
    fit2_y = np.polyval(coeffs2, all_yield_values)
    sorted_idx2 = np.argsort(all_yield_values)
    ax2.plot(np.array(all_yield_values)[sorted_idx2], fit2_y[sorted_idx2], color='black', lw=2, zorder=10, label="Best Fit")
# Add correlation info with significance
signif_str_method4 = f"Correlation: {corr_method4:.4f}\n(p = {pval_method4:.2e})"
ax2.text(0.05, 0.95, signif_str_method4, transform=ax2.transAxes,
         bbox=dict(boxstyle="round,pad=0.3", facecolor="white", alpha=0.8), va='top')

# Plot 3: Distribution comparison
ax3 = axes[1, 0]
ax3.hist(all_capacity_values, bins=30, alpha=0.6, label='Original Capacity', color='blue', density=True)
ax3.hist(global_method4_values, bins=30, alpha=0.6, label='Global Method 4', color='red', density=True)
ax3.set_xlabel('Assembly Capacity')
ax3.set_ylabel('Density')
ax3.set_title('Distribution Comparison')
ax3.legend()
ax3.grid(True, alpha=0.3)

# Plot 4: Correlation reduction
ax4 = axes[1, 1]
correlations = [corr_orig, corr_method4]
labels = ['Original\nCapacity', 'Global\nMethod 4']
colors = ['blue', 'red']
bars = ax4.bar(labels, correlations, color=colors, alpha=0.7)
ax4.set_ylabel('Correlation with Yield')
ax4.set_title('Correlation Reduction')
ax4.set_ylim(0, max(correlations) * 1.1)
ax4.grid(True, alpha=0.3)

# Add value labels on bars with significance
for bar, corr, pval in zip(bars, correlations, [pval_orig, pval_method4]):
    height = bar.get_height()
    ax4.text(bar.get_x() + bar.get_width()/2., height + 0.01,
             f'{corr:.4f}\np={pval:.2e}', ha='center', va='bottom')

plt.tight_layout()
fig.savefig("yield_info_assembly_correlation.svg", dpi=300, bbox_inches='tight')
plt.show()

# Print summary statistics
print(f"\nSummary Statistics:")
print(f"Original Capacity - Mean: {np.mean(all_capacity_values):.6f}, Std: {np.std(all_capacity_values):.6f}")
print(f"Global Method 4 - Mean: {np.mean(global_method4_values):.6f}, Std: {np.std(global_method4_values):.6f}")
print(f"Correlation with Yield - Original: {corr_orig:.6f} (p={pval_orig:.2e}), Global Method 4: {corr_method4:.6f} (p={pval_method4:.2e})")
print(f"Correlation Reduction: {((corr_orig - corr_method4) / corr_orig * 100):.2f}%")

# Store global Method 4 values for further analysis
global_method4_by_insertion = {}

# Group the global Method 4 values by insertion for comparison with other methods
for info, method4_value in zip(all_assembly_info, global_method4_values):
    insertion_idx = info['insertion_idx']
    assembly_key = info['assembly_key']

    if insertion_idx not in global_method4_by_insertion:
        global_method4_by_insertion[insertion_idx] = {}

    global_method4_by_insertion[insertion_idx][assembly_key] = method4_value

print(f"\nGlobal Method 4 values organized by insertion: {len(global_method4_by_insertion)} insertions")
print("Ready for comparison with other methods!")
```

```python
# REGRESSION-BASED YIELD CORRECTION (Proper Method)
# This replaces the MI-based correction with a statistically sound regression approach

import numpy as np
from scipy.stats import pearsonr
import matplotlib.pyplot as plt

print("="*80)
print("REGRESSION-BASED YIELD CORRECTION")
print("="*80)

# Ensure we have the data (assuming all_capacity_values and all_yield_values exist)
# If not, collect them the same way as before
if 'all_capacity_values' not in locals() or 'all_yield_values' not in locals():
    print("Collecting data from assembly_info_capacity_all_list...")
    all_capacity_values = []
    all_yield_values = []
    all_assembly_info = []
    
    for insertion_idx, insertion_data in enumerate(assembly_info_capacity_all_list):
        if isinstance(insertion_data, dict) and len(insertion_data) > 0:
            insertion_yield = yield_list[insertion_idx]
            for assembly_key, assembly_data in insertion_data.items():
                if isinstance(assembly_data, dict) and 'capacity' in assembly_data:
                    capacity_value = assembly_data['capacity']
                    try:
                        capacity_float = float(capacity_value)
                        if not np.isnan(capacity_float):
                            all_capacity_values.append(capacity_float)
                            all_yield_values.append(insertion_yield)
                            all_assembly_info.append({
                                'insertion_idx': insertion_idx,
                                'assembly_key': assembly_key,
                                'capacity': capacity_float,
                                'yield': insertion_yield,
                                'opercular': opercular_list[insertion_idx],
                                'pathology': path_list[insertion_idx],
                                'grade': grade_list[insertion_idx]
                            })
                    except (ValueError, TypeError):
                        continue

# Convert to numpy arrays
all_capacity_values = np.array(all_capacity_values)
all_yield_values = np.array(all_yield_values)
mean_yield = np.mean(all_yield_values)

print(f"\nData Summary:")
print(f"  Total assemblies: {len(all_capacity_values)}")
print(f"  Mean yield: {mean_yield:.2f}")
print(f"  Mean capacity: {np.mean(all_capacity_values):.4f} bits")
print(f"  Capacity range: [{np.min(all_capacity_values):.4f}, {np.max(all_capacity_values):.4f}] bits")

# 1. Compute regression slope
slope, intercept = np.polyfit(all_yield_values, all_capacity_values, 1)
print(f"\n{'='*80}")
print("REGRESSION ANALYSIS")
print(f"{'='*80}")
print(f"Linear fit: capacity = {slope:.6f} * yield + {intercept:.6f}")
print(f"  Slope: {slope:.6f} bits/neuron")
print(f"  Intercept: {intercept:.6f} bits")

# 2. Compare with MI-based approach
if 'global_mi' in locals():
    print(f"\n{'='*80}")
    print("COMPARISON: Regression vs MI-based")
    print(f"{'='*80}")
    print(f"MI-based correction at mean yield: {global_mi:.6f} bits")
    print(f"Regression-based correction at mean yield: {slope * mean_yield:.6f} bits")
    print(f"  Difference: {abs(global_mi - slope * mean_yield):.6f} bits")
    print(f"  Ratio (MI/Regression): {global_mi / (slope * mean_yield):.4f}")
    
    if abs(global_mi - slope * mean_yield) < 0.01:
        print(f"  ✅ Values are very close - MI approach may be acceptable")
    else:
        print(f"  ⚠️  Values differ significantly - regression approach is more accurate")

# 3. Compute regression-based corrected capacity
regression_corrected_capacity = []
for info in all_assembly_info:
    # Method 1: Subtract only the yield-dependent component
    adjusted = info['capacity'] - (slope * info['yield'])
    # Alternative: Subtract full prediction (uncomment if preferred)
    # adjusted = info['capacity'] - (slope * info['yield'] + intercept)
    regression_corrected_capacity.append(adjusted)

regression_corrected_capacity = np.array(regression_corrected_capacity)

# 4. Compare correlations
corr_orig, pval_orig = pearsonr(all_yield_values, all_capacity_values)
corr_regression, pval_regression = pearsonr(all_yield_values, regression_corrected_capacity)

if 'global_method4_values' in locals():
    corr_mi, pval_mi = pearsonr(all_yield_values, global_method4_values)
    
    print(f"\n{'='*80}")
    print("CORRELATION COMPARISON")
    print(f"{'='*80}")
    print(f"Original capacity vs yield:        r = {corr_orig:.6f}, p = {pval_orig:.2e}")
    print(f"MI-based correction vs yield:       r = {corr_mi:.6f}, p = {pval_mi:.2e}")
    print(f"Regression-based correction vs yield: r = {corr_regression:.6f}, p = {pval_regression:.2e}")
    print(f"\nCorrelation reduction:")
    print(f"  MI-based:       {((corr_orig - corr_mi) / corr_orig * 100):.2f}%")
    print(f"  Regression-based: {((corr_orig - corr_regression) / corr_orig * 100):.2f}%")
else:
    print(f"\n{'='*80}")
    print("CORRELATION ANALYSIS")
    print(f"{'='*80}")
    print(f"Original capacity vs yield:        r = {corr_orig:.6f}, p = {pval_orig:.2e}")
    print(f"Regression-corrected vs yield:     r = {corr_regression:.6f}, p = {pval_regression:.2e}")
    print(f"Correlation reduction: {((corr_orig - corr_regression) / corr_orig * 100):.2f}%")

# 5. Store corrected values for downstream analysis
regression_corrected_by_insertion = {}
for i, info in enumerate(all_assembly_info):
    insertion_idx = info['insertion_idx']
    assembly_key = info['assembly_key']
    
    if insertion_idx not in regression_corrected_by_insertion:
        regression_corrected_by_insertion[insertion_idx] = {}
    
    regression_corrected_by_insertion[insertion_idx][assembly_key] = regression_corrected_capacity[i]

print(f"\n{'='*80}")
print("SUMMARY")
print(f"{'='*80}")
print(f"✅ Regression-based correction computed")
print(f"✅ Correlation with yield: {corr_regression:.6f} (target: 0.000)")
print(f"✅ Values stored in: regression_corrected_by_insertion")
print(f"✅ Use 'regression_corrected_capacity' for analysis")
print(f"{'='*80}")

# 6. Visualization comparing both methods (if MI-based exists)
if 'global_method4_values' in locals():
    fig, axes = plt.subplots(2, 2, figsize=(15, 12))
    fig.suptitle('Regression vs MI-based Yield Correction Comparison', fontsize=16, fontweight='bold')
    
    # Plot 1: Original capacity vs yield
    ax1 = axes[0, 0]
    ax1.scatter(all_yield_values, all_capacity_values, alpha=0.6, c='blue', label='Original')
    slope_plot, intercept_plot = np.polyfit(all_yield_values, all_capacity_values, 1)
    x_fit = np.linspace(all_yield_values.min(), all_yield_values.max(), 100)
    ax1.plot(x_fit, slope_plot * x_fit + intercept_plot, 'k--', lw=2, label=f'Fit: y={slope_plot:.4f}x+{intercept_plot:.2f}')
    ax1.set_xlabel('Neuron Yield')
    ax1.set_ylabel('Assembly Capacity (bits)')
    ax1.set_title(f'Original Capacity vs Yield\nr={corr_orig:.4f}, p={pval_orig:.2e}')
    ax1.legend()
    ax1.grid(True, alpha=0.3)
    
    # Plot 2: MI-based correction vs yield
    ax2 = axes[0, 1]
    ax2.scatter(all_yield_values, global_method4_values, alpha=0.6, c='red', label='MI-corrected')
    ax2.axhline(0, color='gray', linestyle='--', alpha=0.5)
    # Add regression line for MI-based correction
    slope_mi, intercept_mi = np.polyfit(all_yield_values, global_method4_values, 1)
    ax2.plot(x_fit, slope_mi * x_fit + intercept_mi, 'k--', lw=2, label=f'Fit: y={slope_mi:.4f}x+{intercept_mi:.2f}')
    ax2.set_xlabel('Neuron Yield')
    ax2.set_ylabel('MI-Corrected Capacity')
    ax2.set_title(f'MI-based Correction vs Yield\nr={corr_mi:.4f}, p={pval_mi:.2e}')
    ax2.legend()
    ax2.grid(True, alpha=0.3)
    
    # Plot 3: Regression-based correction vs yield
    ax3 = axes[1, 0]
    ax3.scatter(all_yield_values, regression_corrected_capacity, alpha=0.6, c='green', label='Regression-corrected')
    ax3.axhline(0, color='gray', linestyle='--', alpha=0.5)
    # Add regression line for regression-corrected
    slope_reg, intercept_reg = np.polyfit(all_yield_values, regression_corrected_capacity, 1)
    ax3.plot(x_fit, slope_reg * x_fit + intercept_reg, 'k--', lw=2, label=f'Fit: y={slope_reg:.4f}x+{intercept_reg:.2f}')
    ax3.set_xlabel('Neuron Yield')
    ax3.set_ylabel('Regression-Corrected Capacity')
    ax3.set_title(f'Regression-based Correction vs Yield\nr={corr_regression:.4f}, p={pval_regression:.2e}')
    ax3.legend()
    ax3.grid(True, alpha=0.3)
    
    # Plot 4: Comparison of both corrections
    ax4 = axes[1, 1]
    ax4.scatter(global_method4_values, regression_corrected_capacity, alpha=0.6, c='purple')
    # Add diagonal line
    min_val = min(np.min(global_method4_values), np.min(regression_corrected_capacity))
    max_val = max(np.max(global_method4_values), np.max(regression_corrected_capacity))
    ax4.plot([min_val, max_val], [min_val, max_val], 'k--', lw=2, alpha=0.5, label='y=x')
    # Add regression line to scatter MI vs regression-corrected
    slope_methods, intercept_methods = np.polyfit(global_method4_values, regression_corrected_capacity, 1)
    x_diag = np.linspace(min_val, max_val, 100)
    ax4.plot(x_diag, slope_methods * x_diag + intercept_methods, 'r-', lw=2, alpha=0.5, label=f'Fit: y={slope_methods:.4f}x+{intercept_methods:.2f}')
    ax4.set_xlabel('MI-based Correction')
    ax4.set_ylabel('Regression-based Correction')
    ax4.set_title('Comparison of Correction Methods')
    ax4.legend()
    ax4.grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.savefig("regression_vs_mi_correction_comparison.svg", dpi=300, bbox_inches='tight')
    plt.show()
else:
    # Just plot regression-based if MI doesn't exist
    fig, axes = plt.subplots(1, 2, figsize=(12, 5))
    fig.suptitle('Regression-based Yield Correction', fontsize=16, fontweight='bold')
    
    # Plot 1: Original
    ax1 = axes[0]
    ax1.scatter(all_yield_values, all_capacity_values, alpha=0.6, c='blue')
    slope_plot, intercept_plot = np.polyfit(all_yield_values, all_capacity_values, 1)
    x_fit = np.linspace(all_yield_values.min(), all_yield_values.max(), 100)
    ax1.plot(x_fit, slope_plot * x_fit + intercept_plot, 'k--', lw=2, label=f'Fit: y={slope_plot:.4f}x+{intercept_plot:.2f}')
    ax1.set_xlabel('Neuron Yield')
    ax1.set_ylabel('Assembly Capacity (bits)')
    ax1.set_title(f'Original: r={corr_orig:.4f}')
    ax1.legend()
    ax1.grid(True, alpha=0.3)
    
    # Plot 2: Corrected
    ax2 = axes[1]
    ax2.scatter(all_yield_values, regression_corrected_capacity, alpha=0.6, c='green')
    ax2.axhline(0, color='gray', linestyle='--', alpha=0.5)
    # Add regression line for regression-corrected capacity
    slope_reg, intercept_reg = np.polyfit(all_yield_values, regression_corrected_capacity, 1)
    ax2.plot(x_fit, slope_reg * x_fit + intercept_reg, 'k--', lw=2, label=f'Fit: y={slope_reg:.4f}x+{intercept_reg:.2f}')
    ax2.set_xlabel('Neuron Yield')
    ax2.set_ylabel('Corrected Capacity')
    ax2.set_title(f'Corrected: r={corr_regression:.4f}')
    ax2.legend()
    ax2.grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.savefig("regression_correction.svg", dpi=300, bbox_inches='tight')
    plt.show()

print("\n✅ Analysis complete! Use 'regression_corrected_capacity' for downstream analyses.")
```

```python
# SUMMARY PLOT - ASSEMBLY
import matplotlib.pyplot as plt
import numpy as np
from scipy import stats

plt.rcParams['svg.fonttype'] = 'none' 

def compare_global_method4_groups_opercular(stat_method='parametric'):
    """
    Compare global Method 4 (yield-independent assembly capacity) across tumor grades and pathology groups
    Show both opercular=0 and opercular=1 side by side for direct comparison
    Grade 0 analysis removed.
    Allows choice of parametric (t-test) or non-parametric (Mann-Whitney U) statistics.
    All tests are one-sided (greater: non-opercular > opercular, or mutant > wildtype).
    
    Parameters
    ----------
    stat_method : str
        'parametric' for t-test, 'nonparametric' for Mann-Whitney U test.
    """
    
    print("Comparing Global Method 4 across Tumor Grades and Pathology Groups")
    print("=" * 70)
    print("Comparing: Non-opercular (opercular=0) vs Opercular (opercular=1) regions")
    print(f"Statistical method: {stat_method} ({'t-test' if stat_method=='parametric' else 'Mann-Whitney U'})")
    
    # Extract values for each tumor grade group (both opercular types)
    grade_2_3_non_opercular = []  # Grade 2+3, Non-opercular
    grade_2_3_opercular = []      # Grade 2+3, Opercular
    grade_4_non_opercular = []    # Grade 4, Non-opercular
    grade_4_opercular = []        # Grade 4, Opercular
    
    # Extract values for each pathology group (both opercular types)
    vas_non_opercular = []        # Vascular, Non-opercular
    vas_opercular = []            # Vascular, Opercular
    idh_mut_non_opercular = []    # IDH-mutant, Non-opercular
    idh_mut_opercular = []        # IDH-mutant, Opercular
    idh_wt_non_opercular = []     # IDH-wildtype, Non-opercular
    idh_wt_opercular = []         # IDH-wildtype, Opercular
    
    # Process global Method 4 data (both opercular types)
    for info in all_assembly_info:
        # Get the corresponding global Method 4 value
        insertion_idx = info['insertion_idx']
        assembly_key = info['assembly_key']
        
        if insertion_idx in global_method4_by_insertion and assembly_key in global_method4_by_insertion[insertion_idx]:
            method4_value = global_method4_by_insertion[insertion_idx][assembly_key]
            opercular_type = info['opercular']
            
            # Group by tumor grade and opercular type (grade 0 removed)
            grade = info['grade']
            if grade in [2, 3]:
                if opercular_type == 0:
                    grade_2_3_non_opercular.append(method4_value)
                elif opercular_type == 1:
                    grade_2_3_opercular.append(method4_value)
            elif grade == 4:
                if opercular_type == 0:
                    grade_4_non_opercular.append(method4_value)
                elif opercular_type == 1:
                    grade_4_opercular.append(method4_value)
            
            # Group by pathology and opercular type
            pathology = info['pathology']
            if pathology in ['ast', 'oli']:
                if opercular_type == 0:
                    idh_mut_non_opercular.append(method4_value)
                elif opercular_type == 1:
                    idh_mut_opercular.append(method4_value)
            elif pathology == 'gbm':
                if opercular_type == 0:
                    idh_wt_non_opercular.append(method4_value)
                elif opercular_type == 1:
                    idh_wt_opercular.append(method4_value)
    
    # Print summary statistics
    print(f"\nTumor Grade Groups:")
    print(f"Grade 2+3 - Non-opercular: n={len(grade_2_3_non_opercular)}, mean={np.mean(grade_2_3_non_opercular):.4f}, std={np.std(grade_2_3_non_opercular):.4f}")
    print(f"Grade 2+3 - Opercular: n={len(grade_2_3_opercular)}, mean={np.mean(grade_2_3_opercular):.4f}, std={np.std(grade_2_3_opercular):.4f}")
    print(f"Grade 4 - Non-opercular: n={len(grade_4_non_opercular)}, mean={np.mean(grade_4_non_opercular):.4f}, std={np.std(grade_4_non_opercular):.4f}")
    print(f"Grade 4 - Opercular: n={len(grade_4_opercular)}, mean={np.mean(grade_4_opercular):.4f}, std={np.std(grade_4_opercular):.4f}")
    
    print(f"\nPathology Groups:")
    print(f"IDH-mutant - Non-opercular: n={len(idh_mut_non_opercular)}, mean={np.mean(idh_mut_non_opercular):.4f}, std={np.std(idh_mut_non_opercular):.4f}")
    print(f"IDH-mutant - Opercular: n={len(idh_mut_opercular)}, mean={np.mean(idh_mut_opercular):.4f}, std={np.std(idh_mut_opercular):.4f}")
    print(f"IDH-wildtype - Non-opercular: n={len(idh_wt_non_opercular)}, mean={np.mean(idh_wt_non_opercular):.4f}, std={np.std(idh_wt_non_opercular):.4f}")
    print(f"IDH-wildtype - Opercular: n={len(idh_wt_opercular)}, mean={np.mean(idh_wt_opercular):.4f}, std={np.std(idh_wt_opercular):.4f}")
    
    # Helper function for one-sided stats
    def one_sided_test(a, b, alternative='greater'):
        if stat_method == 'parametric':
            # t-test, one-sided
            t_stat, p_val = stats.ttest_ind(a, b, alternative=alternative)
            return t_stat, p_val
        elif stat_method == 'nonparametric':
            # Mann-Whitney U, one-sided
            # For alternative='greater', test if median(a) > median(b)
            u_stat, p_val = stats.mannwhitneyu(a, b, alternative=alternative)
            return u_stat, p_val
        else:
            raise ValueError("stat_method must be 'parametric' or 'nonparametric'")
    
    # Statistical tests for regional differences within each grade (one-sided: non-opercular > opercular)
    print(f"\nStatistical Tests - Regional Differences within Grades (one-sided, non-opercular > opercular):")
    
    # Grade 2+3 regional comparison
    grade_2_3_p_value = None
    if len(grade_2_3_non_opercular) > 0 and len(grade_2_3_opercular) > 0:
        stat, p_val = one_sided_test(grade_2_3_non_opercular, grade_2_3_opercular, alternative='greater')
        grade_2_3_p_value = p_val
        print(f"Grade 2+3: Non-opercular vs Opercular: stat={stat:.4f}, p={p_val:.4f}")
    
    # Grade 4 regional comparison
    grade_4_p_value = None
    if len(grade_4_non_opercular) > 0 and len(grade_4_opercular) > 0:
        stat, p_val = one_sided_test(grade_4_non_opercular, grade_4_opercular, alternative='greater')
        grade_4_p_value = p_val
        print(f"Grade 4: Non-opercular vs Opercular: stat={stat:.4f}, p={p_val:.4f}")
    
    # Cross-grade comparisons (one-sided: 2+3 > 4)
    print(f"\nStatistical Tests - Cross-Grade Comparisons (one-sided, 2+3 > 4):")
    
    # Grade 2+3 vs Grade 4 (Tumor regions)
    grade_2_3_vs_4_tumor_p_value = None
    if len(grade_2_3_non_opercular) > 0 and len(grade_4_non_opercular) > 0:
        stat, p_val = one_sided_test(grade_2_3_non_opercular, grade_4_non_opercular, alternative='greater')
        grade_2_3_vs_4_tumor_p_value = p_val
        print(f"Grade 2+3 vs Grade 4 (Tumor): stat={stat:.4f}, p={p_val:.4f}")
    
    # Grade 2+3 vs Grade 4 (Opercular regions)
    grade_2_3_vs_4_opercular_p_value = None
    if len(grade_2_3_opercular) > 0 and len(grade_4_opercular) > 0:
        stat, p_val = one_sided_test(grade_2_3_opercular, grade_4_opercular, alternative='greater')
        grade_2_3_vs_4_opercular_p_value = p_val
        print(f"Grade 2+3 vs Grade 4 (Opercular): stat={stat:.4f}, p={p_val:.4f}")
    
    # Statistical tests for regional differences within each pathology (one-sided: non-opercular > opercular)
    print(f"\nStatistical Tests - Regional Differences within Pathology (one-sided, non-opercular > opercular):")
    
    # Vascular regional comparison
    if len(vas_non_opercular) > 0 and len(vas_opercular) > 0:
        stat, p_val = one_sided_test(vas_non_opercular, vas_opercular, alternative='greater')
        print(f"Vascular: Non-opercular vs Opercular: stat={stat:.4f}, p={p_val:.4f}")
    
    # IDH-mutant regional comparison
    idh_mut_p_value = None
    if len(idh_mut_non_opercular) > 0 and len(idh_mut_opercular) > 0:
        stat, p_val = one_sided_test(idh_mut_non_opercular, idh_mut_opercular, alternative='greater')
        idh_mut_p_value = p_val
        print(f"IDH-mutant: Non-opercular vs Opercular: stat={stat:.4f}, p={p_val:.4f}")
    
    # IDH-wildtype regional comparison
    idh_wt_p_value = None
    if len(idh_wt_non_opercular) > 0 and len(idh_wt_opercular) > 0:
        stat, p_val = one_sided_test(idh_wt_non_opercular, idh_wt_opercular, alternative='greater')
        idh_wt_p_value = p_val
        print(f"IDH-wildtype: Non-opercular vs Opercular: stat={stat:.4f}, p={p_val:.4f}")
    
    # Cross-pathology comparisons (one-sided: mutant > wildtype)
    print(f"\nStatistical Tests - Cross-Pathology Comparisons (one-sided, mutant > wildtype):")
    
    # IDH-mutant vs IDH-wildtype (Tumor regions)
    idh_mut_vs_wt_tumor_p_value = None
    if len(idh_mut_non_opercular) > 0 and len(idh_wt_non_opercular) > 0:
        stat, p_val = one_sided_test(idh_mut_non_opercular, idh_wt_non_opercular, alternative='greater')
        idh_mut_vs_wt_tumor_p_value = p_val
        print(f"IDH-mutant vs IDH-wildtype (Tumor): stat={stat:.4f}, p={p_val:.4f}")
    
    # IDH-mutant vs IDH-wildtype (Opercular regions)
    idh_mut_vs_wt_opercular_p_value = None
    if len(idh_mut_opercular) > 0 and len(idh_wt_opercular) > 0:
        stat, p_val = one_sided_test(idh_mut_opercular, idh_wt_opercular, alternative='greater')
        idh_mut_vs_wt_opercular_p_value = p_val
        print(f"IDH-mutant vs IDH-wildtype (Opercular): stat={stat:.4f}, p={p_val:.4f}")
    
    # Create visualization
    fig, axes = plt.subplots(2, 2, figsize=(18, 12))
    fig.suptitle('Global Method 4: Yield-Independent Assembly Capacity - Regional Comparison', fontsize=16, fontweight='bold')
    
    # Plot 1: Tumor Grade Groups (side by side with closer spacing)
    ax1 = axes[0, 0]
    
    # Prepare data for side-by-side box plots with closer spacing
    grade_data = []
    grade_labels = []
    grade_colors = []
    grade_positions = []
    
    # Grade 2+3 (positions 1, 2)
    if len(grade_2_3_non_opercular) > 0:
        grade_data.append(grade_2_3_non_opercular)
        grade_labels.append(f'Grade 2+3\nTumor\n(n={len(grade_2_3_non_opercular)})')
        grade_colors.append('#ff7f0e')  # Orange
        grade_positions.append(1)
    
    if len(grade_2_3_opercular) > 0:
        grade_data.append(grade_2_3_opercular)
        grade_labels.append(f'Grade 2+3\nOpercular\n(n={len(grade_2_3_opercular)})')
        grade_colors.append('#FFB366')  # Light orange
        grade_positions.append(2)
    
    # Grade 4 (positions 4, 5)
    if len(grade_4_non_opercular) > 0:
        grade_data.append(grade_4_non_opercular)
        grade_labels.append(f'Grade 4\nTumor\n(n={len(grade_4_non_opercular)})')
        grade_colors.append('#d62728')  # Red
        grade_positions.append(4)
    
    if len(grade_4_opercular) > 0:
        grade_data.append(grade_4_opercular)
        grade_labels.append(f'Grade 4\nOpercular\n(n={len(grade_4_opercular)})')
        grade_colors.append('#FF6B6B')  # Light red
        grade_positions.append(5)
    
    if grade_data:
        bp1 = ax1.boxplot(grade_data, positions=grade_positions, patch_artist=True)
        for patch, color in zip(bp1['boxes'], grade_colors):
            patch.set_facecolor(color)
            patch.set_alpha(0.7)

        # ADD INDIVIDUAL DATA POINTS
        for i, (data, pos) in enumerate(zip(grade_data, grade_positions)):
        # Add small random jitter to x-position to avoid overlap
            jitter = np.random.normal(0, 0.05, len(data))
            ax1.scatter(pos + jitter, data, s=8, c='black', alpha=0.6, zorder=10)
        
        ax1.set_ylabel('Global Method 4 Capacity', fontsize=12)
        ax1.set_title('Tumor Grade Groups - Regional Comparison', fontsize=12)
        ax1.grid(True, alpha=0.3)
        
        # Set x-axis labels and ticks
        ax1.set_xticks([1.5, 4.5])
        ax1.set_xticklabels(['Grade 2+3', 'Grade 4'])
        
        # Add legend
        from matplotlib.patches import Patch
        legend_elements = [Patch(facecolor='#ff7f0e', alpha=0.7, label='Tumor'),
                          Patch(facecolor='#FFB366', alpha=0.7, label='Opercular')]
        ax1.legend(handles=legend_elements, loc='upper right')
        
        # Add statistical annotations with significance lines
        def get_significance_symbol(p_value):
            if p_value is None:
                return ""
            elif p_value < 0.001:
                return "***"
            elif p_value < 0.01:
                return "**"
            elif p_value < 0.05:
                return "*"
            else:
                return "ns"
        
        def add_significance_line(ax, x1, x2, y, p_value, label=""):
            """Add significance line with symbol above groups"""
            if p_value is None:
                return
            
            sig_symbol = get_significance_symbol(p_value)
            y_max = ax.get_ylim()[1]
            y_line = y_max * y
            
            # Draw horizontal line
            ax.plot([x1, x2], [y_line, y_line], 'k-', linewidth=1)
            
            # Draw vertical lines at ends
            ax.plot([x1, x1], [y_line - y_max*0.01, y_line], 'k-', linewidth=1)
            ax.plot([x2, x2], [y_line - y_max*0.01, y_line], 'k-', linewidth=1)
            
            # Add significance symbol
            ax.text((x1 + x2) / 2, y_line + y_max*0.02, sig_symbol, 
                   ha='center', va='bottom', fontsize=12, fontweight='bold')
            
            # Add label if provided
            if label:
                ax.text((x1 + x2) / 2, y_line - y_max*0.05, label, 
                       ha='center', va='top', fontsize=8, style='italic')
        
        # Regional comparisons (within each grade)
        if grade_2_3_p_value is not None:
            add_significance_line(ax1, 1, 2, 0.85, grade_2_3_p_value, "Regional")
        
        if grade_4_p_value is not None:
            add_significance_line(ax1, 4, 5, 0.85, grade_4_p_value, "Regional")
        
        # Cross-grade comparisons
        if grade_2_3_vs_4_tumor_p_value is not None:
            add_significance_line(ax1, 1, 4, 0.95, grade_2_3_vs_4_tumor_p_value, "Grade")
        
        if grade_2_3_vs_4_opercular_p_value is not None:
            add_significance_line(ax1, 2, 5, 0.95, grade_2_3_vs_4_opercular_p_value, "Grade")
    
    # Plot 2: Pathology Groups (side by side with closer spacing)
    ax2 = axes[0, 1]
    
    # Prepare data for side-by-side box plots with closer spacing
    path_data = []
    path_labels = []
    path_colors = []
    path_positions = []
    
    # IDH-mutant (positions 1, 2)
    if len(idh_mut_non_opercular) > 0:
        path_data.append(idh_mut_non_opercular)
        path_labels.append(f'IDH-mutant\nTumor\n(n={len(idh_mut_non_opercular)})')
        path_colors.append('#1f77b4')  # Blue
        path_positions.append(1)
    
    if len(idh_mut_opercular) > 0:
        path_data.append(idh_mut_opercular)
        path_labels.append(f'IDH-mutant\nOpercular\n(n={len(idh_mut_opercular)})')
        path_colors.append('#87CEEB')  # Light blue
        path_positions.append(2)
    
    # IDH-wildtype (positions 4, 5)
    if len(idh_wt_non_opercular) > 0:
        path_data.append(idh_wt_non_opercular)
        path_labels.append(f'IDH-wildtype\nTumor\n(n={len(idh_wt_non_opercular)})')
        path_colors.append('#d62728')  # Red
        path_positions.append(4)
    
    if len(idh_wt_opercular) > 0:
        path_data.append(idh_wt_opercular)
        path_labels.append(f'IDH-wildtype\nOpercular\n(n={len(idh_wt_opercular)})')
        path_colors.append('#FF6B6B')  # Light red
        path_positions.append(5)
    
    if path_data:
        bp2 = ax2.boxplot(path_data, positions=path_positions, patch_artist=True)
        for patch, color in zip(bp2['boxes'], path_colors):
            patch.set_facecolor(color)
            patch.set_alpha(0.7)

            # ADD INDIVIDUAL DATA POINTS
        for i, (data, pos) in enumerate(zip(path_data, path_positions)):
            # Add small random jitter to x-position to avoid overlap
            jitter = np.random.normal(0, 0.05, len(data))
            ax2.scatter(pos + jitter, data, s=8, c='black', alpha=0.6, zorder=10)
        
        ax2.set_ylabel('Global Method 4 Capacity', fontsize=12)
        ax2.set_title('Pathology Groups - Regional Comparison', fontsize=12)
        ax2.grid(True, alpha=0.3)
        
        # Set x-axis labels and ticks
        ax2.set_xticks([1.5, 4.5])
        ax2.set_xticklabels(['IDH-mutant', 'IDH-wildtype'])
        
        # Add legend
        legend_elements = [Patch(facecolor='#1f77b4', alpha=0.7, label='Tumor'),
                          Patch(facecolor='#87CEEB', alpha=0.7, label='Opercular')]
        ax2.legend(handles=legend_elements, loc='upper right')
        
        # Add statistical annotations
        # IDH-mutant significance
        if idh_mut_p_value is not None:
            add_significance_line(ax2, 1, 2, 0.85, idh_mut_p_value, "Regional")
        
        # IDH-wildtype significance
        if idh_wt_p_value is not None:
            add_significance_line(ax2, 4, 5, 0.85, idh_wt_p_value, "Regional")
        
        # Cross-pathology comparisons
        if idh_mut_vs_wt_tumor_p_value is not None:
            add_significance_line(ax2, 1, 4, 0.95, idh_mut_vs_wt_tumor_p_value, "Pathology")
        
        if idh_mut_vs_wt_opercular_p_value is not None:
            add_significance_line(ax2, 2, 5, 0.95, idh_mut_vs_wt_opercular_p_value, "Pathology")
    
    # Plot 3: Grade Distribution Histograms (Non-opercular only)
    ax3 = axes[1, 0]
    
    # Only non-opercular grades (grade 0 removed)
    if len(grade_2_3_non_opercular) > 0:
        ax3.hist(grade_2_3_non_opercular, bins=15, alpha=0.6, label='Grade 2+3 Tumor', color='#ff7f0e', density=True)
    if len(grade_4_non_opercular) > 0:
        ax3.hist(grade_4_non_opercular, bins=15, alpha=0.6, label='Grade 4 Tumor', color='#d62728', density=True)
    
    ax3.set_xlabel('Global Method 4 Capacity', fontsize=12)
    ax3.set_ylabel('Density', fontsize=12)
    ax3.set_title('Grade Distribution Histograms (Tumor Only)', fontsize=12)
    ax3.legend()
    ax3.grid(True, alpha=0.3)
    
    # Plot 4: Pathology Distribution Histograms (Non-opercular only)
    ax4 = axes[1, 1]
    
    # Only non-opercular pathology

    if len(idh_mut_non_opercular) > 0:
        ax4.hist(idh_mut_non_opercular, bins=15, alpha=0.6, label='IDH-mutant Tumor', color='#1f77b4', density=True)
    if len(idh_wt_non_opercular) > 0:
        ax4.hist(idh_wt_non_opercular, bins=15, alpha=0.6, label='IDH-wildtype Tumor', color='#d62728', density=True)
    
    ax4.set_xlabel('Global Method 4 Capacity', fontsize=12)
    ax4.set_ylabel('Density', fontsize=12)
    ax4.set_title('Pathology Distribution Histograms (Tumor Only)', fontsize=12)
    ax4.legend()
    ax4.grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.savefig("assembly_information_summary.svg", format="svg", dpi=300, bbox_inches='tight')
    plt.show()
    
    return {
        'grade_groups': {
            'grade_2_3_non_opercular': grade_2_3_non_opercular,
            'grade_2_3_opercular': grade_2_3_opercular,
            'grade_4_non_opercular': grade_4_non_opercular,
            'grade_4_opercular': grade_4_opercular
        },
        'pathology_groups': {

            'idh_mutant_non_opercular': idh_mut_non_opercular,
            'idh_mutant_opercular': idh_mut_opercular,
            'idh_wildtype_non_opercular': idh_wt_non_opercular,
            'idh_wildtype_opercular': idh_wt_opercular
        }
    }

# Run the comparison for opercular regions
# Example usage:
# results_global_method4_opercular = compare_global_method4_groups_opercular(stat_method='parametric')
# results_global_method4_opercular = compare_global_method4_groups_opercular(stat_method='nonparametric')
results_global_method4_opercular = compare_global_method4_groups_opercular(stat_method='parametric')
```

```python
# SUMMARY PLOT - NEURON
import matplotlib.pyplot as plt
import numpy as np
from scipy import stats

plt.rcParams['svg.fonttype'] = 'none' 

def compare_single_neuron_capacity_groups(stat_method='parametric'):
    """
    Compare single neuron capacity across tumor grades and pathology groups
    Show both opercular=0 and opercular=1 side by side for direct comparison
    Only analyze tumor grades 2+3 and 4 (removes grade 0 analysis).
    All statistical tests are one-sided (greater).
    
    Parameters
    ----------
    stat_method : str
        'parametric' for t-test, 'nonparametric' for Mann-Whitney U test.
    """
    
    print("Comparing Single Neuron Capacity across Tumor Grades and Pathology Groups")
    print("=" * 70)
    print("Comparing: Non-opercular (opercular=0) vs Opercular (opercular=1) regions")
    print(f"Statistical method: {stat_method} ({'t-test' if stat_method=='parametric' else 'Mann-Whitney U'})")
    
    # Extract values for each tumor grade group (both opercular types)
    grade_2_3_non_opercular = []  # Grade 2+3, Non-opercular
    grade_2_3_opercular = []      # Grade 2+3, Opercular
    grade_4_non_opercular = []    # Grade 4, Non-opercular
    grade_4_opercular = []        # Grade 4, Opercular
    
    # Extract values for each pathology group (both opercular types)
    idh_mut_non_opercular = []    # IDH-mutant, Non-opercular
    idh_mut_opercular = []        # IDH-mutant, Opercular
    idh_wt_non_opercular = []     # IDH-wildtype, Non-opercular
    idh_wt_opercular = []         # IDH-wildtype, Opercular
    
    # Debug: Check the neuron capacity data structure
    print("Debugging neuron capacity data structure:")
    print(f"neuron_info_capacity_all_list type: {type(neuron_info_capacity_all_list)}")
    print(f"neuron_info_capacity_all_list length: {len(neuron_info_capacity_all_list)}")
    
    # Check first few items
    for i in range(min(3, len(neuron_info_capacity_all_list))):
        print(f"Insertion {i}: {type(neuron_info_capacity_all_list[i])}")
        if isinstance(neuron_info_capacity_all_list[i], dict):
            print(f"  Dict with {len(neuron_info_capacity_all_list[i])} entries")
            if len(neuron_info_capacity_all_list[i]) > 0:
                first_key = list(neuron_info_capacity_all_list[i].keys())[0]
                first_value = neuron_info_capacity_all_list[i][first_key]
                print(f"  First entry key: {first_key}")
                print(f"  First entry value: {first_value} (type: {type(first_value)})")
                if isinstance(first_value, dict):
                    print(f"    Dict keys: {list(first_value.keys())}")
                    if 'capacity' in first_value:
                        print(f"    Capacity value: {first_value['capacity']} (type: {type(first_value['capacity'])})")
        else:
            print(f"  Not a dict: {neuron_info_capacity_all_list[i]}")
    
    print("\n" + "="*50)
    
    # Process single neuron capacity data (both opercular types)
    valid_count = 0
    total_count = 0
    
    # Instead of matching assembly keys, collect all neuron capacities per insertion
    for insertion_idx in range(len(neuron_info_capacity_all_list)):
        insertion_data = neuron_info_capacity_all_list[insertion_idx]
        
        if isinstance(insertion_data, dict) and len(insertion_data) > 0:
            # Get metadata for this insertion
            if insertion_idx < len(grade_list):
                grade = grade_list[insertion_idx]
            else:
                continue
                
            if insertion_idx < len(path_list):
                pathology = path_list[insertion_idx]
            else:
                continue
                
            if insertion_idx < len(opercular_list):
                opercular_type = opercular_list[insertion_idx]
            else:
                continue
            
            # Collect all neuron capacities for this insertion
            neuron_capacities = []
            for neuron_key, neuron_data in insertion_data.items():
                if isinstance(neuron_data, dict) and 'capacity' in neuron_data:
                    try:
                        capacity_float = float(neuron_data['capacity'])
                        if not np.isnan(capacity_float):
                            neuron_capacities.append(capacity_float)
                            valid_count += 1
                    except (ValueError, TypeError):
                        continue
            
            # If we have valid neuron capacities, add them to the appropriate groups
            if len(neuron_capacities) > 0:
                total_count += len(neuron_capacities)
                
                # Group by tumor grade and opercular type (only grades 2+3 and 4)
                if grade in [2, 3]:
                    if opercular_type == 0:
                        grade_2_3_non_opercular.extend(neuron_capacities)
                    elif opercular_type == 1:
                        grade_2_3_opercular.extend(neuron_capacities)
                elif grade == 4:
                    if opercular_type == 0:
                        grade_4_non_opercular.extend(neuron_capacities)
                    elif opercular_type == 1:
                        grade_4_opercular.extend(neuron_capacities)
                
                # Group by pathology and opercular type
                if pathology in ['ast', 'oli']:
                    if opercular_type == 0:
                        idh_mut_non_opercular.extend(neuron_capacities)
                    elif opercular_type == 1:
                        idh_mut_opercular.extend(neuron_capacities)
                elif pathology == 'gbm':
                    if opercular_type == 0:
                        idh_wt_non_opercular.extend(neuron_capacities)
                    elif opercular_type == 1:
                        idh_wt_opercular.extend(neuron_capacities)
    
    print(f"Data processing summary:")
    print(f"Total neuron capacity values processed: {total_count}")
    print(f"Valid capacity values found: {valid_count}")
    print(f"Valid data percentage: {valid_count/total_count*100:.1f}%")
    
    # Print summary statistics
    print(f"\nTumor Grade Groups:")
    print(f"Grade 2+3 - Non-opercular: n={len(grade_2_3_non_opercular)}, mean={np.mean(grade_2_3_non_opercular):.4f}, std={np.std(grade_2_3_non_opercular):.4f}")
    print(f"Grade 2+3 - Opercular: n={len(grade_2_3_opercular)}, mean={np.mean(grade_2_3_opercular):.4f}, std={np.std(grade_2_3_opercular):.4f}")
    print(f"Grade 4 - Non-opercular: n={len(grade_4_non_opercular)}, mean={np.mean(grade_4_non_opercular):.4f}, std={np.std(grade_4_non_opercular):.4f}")
    print(f"Grade 4 - Opercular: n={len(grade_4_opercular)}, mean={np.mean(grade_4_opercular):.4f}, std={np.std(grade_4_opercular):.4f}")
    
    print(f"\nPathology Groups:")
    print(f"IDH-mutant - Non-opercular: n={len(idh_mut_non_opercular)}, mean={np.mean(idh_mut_non_opercular):.4f}, std={np.std(idh_mut_non_opercular):.4f}")
    print(f"IDH-mutant - Opercular: n={len(idh_mut_opercular)}, mean={np.mean(idh_mut_opercular):.4f}, std={np.std(idh_mut_opercular):.4f}")
    print(f"IDH-wildtype - Non-opercular: n={len(idh_wt_non_opercular)}, mean={np.mean(idh_wt_non_opercular):.4f}, std={np.std(idh_wt_non_opercular):.4f}")
    print(f"IDH-wildtype - Opercular: n={len(idh_wt_opercular)}, mean={np.mean(idh_wt_opercular):.4f}, std={np.std(idh_wt_opercular):.4f}")
    
    # Helper function for one-sided stats
    def one_sided_test(a, b, stat_method='parametric'):
        if stat_method == 'parametric':
            # t-test, one-sided (greater)
            t_stat, p_val = stats.ttest_ind(a, b, alternative='greater')
            return t_stat, p_val
        else:
            # Mann-Whitney U, one-sided (greater)
            # Use alternative='greater' (tests if a > b)
            u_stat, p_val = stats.mannwhitneyu(a, b, alternative='greater')
            return u_stat, p_val

    # Statistical tests for regional differences within each grade (one-sided, greater)
    print(f"\nStatistical Tests - Regional Differences within Grades (one-sided, greater):")
    
    # Grade 2+3 regional comparison
    grade_2_3_p_value = None
    if len(grade_2_3_non_opercular) > 0 and len(grade_2_3_opercular) > 0:
        stat, p_val = one_sided_test(grade_2_3_non_opercular, grade_2_3_opercular, stat_method)
        grade_2_3_p_value = p_val
        stat_name = "t" if stat_method == "parametric" else "U"
        print(f"Grade 2+3: Non-opercular > Opercular: {stat_name}={stat:.4f}, p(one-sided)={p_val:.4f}")
    
    # Grade 4 regional comparison
    grade_4_p_value = None
    if len(grade_4_non_opercular) > 0 and len(grade_4_opercular) > 0:
        stat, p_val = one_sided_test(grade_4_non_opercular, grade_4_opercular, stat_method)
        grade_4_p_value = p_val
        stat_name = "t" if stat_method == "parametric" else "U"
        print(f"Grade 4: Non-opercular > Opercular: {stat_name}={stat:.4f}, p(one-sided)={p_val:.4f}")
    
    # Cross-grade comparisons (one-sided, greater)
    print(f"\nStatistical Tests - Cross-Grade Comparisons (one-sided, greater):")
    
    # Grade 2+3 vs Grade 4 (Tumor regions)
    grade_2_3_vs_4_tumor_p_value = None
    if len(grade_2_3_non_opercular) > 0 and len(grade_4_non_opercular) > 0:
        stat, p_val = one_sided_test(grade_2_3_non_opercular, grade_4_non_opercular, stat_method)
        grade_2_3_vs_4_tumor_p_value = p_val
        stat_name = "t" if stat_method == "parametric" else "U"
        print(f"Grade 2+3 > Grade 4 (Tumor): {stat_name}={stat:.4f}, p(one-sided)={p_val:.4f}")
    
    # Grade 2+3 vs Grade 4 (Opercular regions)
    grade_2_3_vs_4_opercular_p_value = None
    if len(grade_2_3_opercular) > 0 and len(grade_4_opercular) > 0:
        stat, p_val = one_sided_test(grade_2_3_opercular, grade_4_opercular, stat_method)
        grade_2_3_vs_4_opercular_p_value = p_val
        stat_name = "t" if stat_method == "parametric" else "U"
        print(f"Grade 2+3 > Grade 4 (Opercular): {stat_name}={stat:.4f}, p(one-sided)={p_val:.4f}")
    
    # Statistical tests for regional differences within each pathology (one-sided, greater)
    print(f"\nStatistical Tests - Regional Differences within Pathology (one-sided, greater):")
    
    # IDH-mutant regional comparison
    idh_mut_p_value = None
    if len(idh_mut_non_opercular) > 0 and len(idh_mut_opercular) > 0:
        stat, p_val = one_sided_test(idh_mut_non_opercular, idh_mut_opercular, stat_method)
        idh_mut_p_value = p_val
        stat_name = "t" if stat_method == "parametric" else "U"
        print(f"IDH-mutant: Non-opercular > Opercular: {stat_name}={stat:.4f}, p(one-sided)={p_val:.4f}")
    
    # IDH-wildtype regional comparison
    idh_wt_p_value = None
    if len(idh_wt_non_opercular) > 0 and len(idh_wt_opercular) > 0:
        stat, p_val = one_sided_test(idh_wt_non_opercular, idh_wt_opercular, stat_method)
        idh_wt_p_value = p_val
        stat_name = "t" if stat_method == "parametric" else "U"
        print(f"IDH-wildtype: Non-opercular > Opercular: {stat_name}={stat:.4f}, p(one-sided)={p_val:.4f}")
    
    # Cross-pathology comparisons (one-sided, greater)
    print(f"\nStatistical Tests - Cross-Pathology Comparisons (one-sided, greater):")
    
    # IDH-mutant vs IDH-wildtype (Tumor regions)
    idh_mut_vs_wt_tumor_p_value = None
    if len(idh_mut_non_opercular) > 0 and len(idh_wt_non_opercular) > 0:
        stat, p_val = one_sided_test(idh_mut_non_opercular, idh_wt_non_opercular, stat_method)
        idh_mut_vs_wt_tumor_p_value = p_val
        stat_name = "t" if stat_method == "parametric" else "U"
        print(f"IDH-mutant > IDH-wildtype (Tumor): {stat_name}={stat:.4f}, p(one-sided)={p_val:.4f}")
    
    # IDH-mutant vs IDH-wildtype (Opercular regions)
    idh_mut_vs_wt_opercular_p_value = None
    if len(idh_mut_opercular) > 0 and len(idh_wt_opercular) > 0:
        stat, p_val = one_sided_test(idh_mut_opercular, idh_wt_opercular, stat_method)
        idh_mut_vs_wt_opercular_p_value = p_val
        stat_name = "t" if stat_method == "parametric" else "U"
        print(f"IDH-mutant > IDH-wildtype (Opercular): {stat_name}={stat:.4f}, p(one-sided)={p_val:.4f}")
    
    # Create visualization
    fig, axes = plt.subplots(2, 2, figsize=(18, 12))
    fig.suptitle('Single Neuron Capacity - Regional Comparison', fontsize=16, fontweight='bold')
    
    # Plot 1: Tumor Grade Groups (side by side with closer spacing)
    ax1 = axes[0, 0]
    
    # Prepare data for side-by-side box plots with closer spacing
    grade_data = []
    grade_labels = []
    grade_colors = []
    grade_positions = []
    
    # Grade 2+3 (positions 1, 2)
    if len(grade_2_3_non_opercular) > 0:
        grade_data.append(grade_2_3_non_opercular)
        grade_labels.append(f'Grade 2+3\nTumor\n(n={len(grade_2_3_non_opercular)})')
        grade_colors.append('#ff7f0e')  # Orange
        grade_positions.append(1)
    
    if len(grade_2_3_opercular) > 0:
        grade_data.append(grade_2_3_opercular)
        grade_labels.append(f'Grade 2+3\nOpercular\n(n={len(grade_2_3_opercular)})')
        grade_colors.append('#FFB366')  # Light orange
        grade_positions.append(2)
    
    # Grade 4 (positions 4, 5)
    if len(grade_4_non_opercular) > 0:
        grade_data.append(grade_4_non_opercular)
        grade_labels.append(f'Grade 4\nTumor\n(n={len(grade_4_non_opercular)})')
        grade_colors.append('#d62728')  # Red
        grade_positions.append(4)
    
    if len(grade_4_opercular) > 0:
        grade_data.append(grade_4_opercular)
        grade_labels.append(f'Grade 4\nOpercular\n(n={len(grade_4_opercular)})')
        grade_colors.append('#FF6B6B')  # Light red
        grade_positions.append(5)
    
    if grade_data:
        bp1 = ax1.boxplot(grade_data, positions=grade_positions, patch_artist=True)
        for patch, color in zip(bp1['boxes'], grade_colors):
            patch.set_facecolor(color)
            patch.set_alpha(0.7)

        # ADD INDIVIDUAL DATA POINTS
        for i, (data, pos) in enumerate(zip(grade_data, grade_positions)):
            # Add small random jitter to x-position to avoid overlap
            jitter = np.random.normal(0, 0.05, len(data))
            ax1.scatter(pos + jitter, data, s=8, c='black', alpha=0.6, zorder=10)
        
        ax1.set_ylabel('Single Neuron Capacity', fontsize=12)
        ax1.set_title('Tumor Grade Groups - Regional Comparison', fontsize=12)
        ax1.grid(True, alpha=0.3)
        
        # Set x-axis labels and ticks
        ax1.set_xticks([1.5, 4.5])
        ax1.set_xticklabels(['Grade 2+3', 'Grade 4'])
        
        # Add legend
        from matplotlib.patches import Patch
        legend_elements = [Patch(facecolor='#ff7f0e', alpha=0.7, label='Tumor'),
                          Patch(facecolor='#FFB366', alpha=0.7, label='Opercular')]
        ax1.legend(handles=legend_elements, loc='upper right')
        
        # Add statistical annotations with significance lines
        def get_significance_symbol(p_value):
            if p_value is None:
                return ""
            elif p_value < 0.001:
                return "***"
            elif p_value < 0.01:
                return "**"
            elif p_value < 0.05:
                return "*"
            else:
                return "ns"
        
        def add_significance_line(ax, x1, x2, y, p_value, label=""):
            """Add significance line with symbol above groups"""
            if p_value is None:
                return
            
            sig_symbol = get_significance_symbol(p_value)
            y_max = ax.get_ylim()[1]
            y_line = y_max * y
            
            # Draw horizontal line
            ax.plot([x1, x2], [y_line, y_line], 'k-', linewidth=1)
            
            # Draw vertical lines at ends
            ax.plot([x1, x1], [y_line - y_max*0.01, y_line], 'k-', linewidth=1)
            ax.plot([x2, x2], [y_line - y_max*0.01, y_line], 'k-', linewidth=1)
            
            # Add significance symbol
            ax.text((x1 + x2) / 2, y_line + y_max*0.02, sig_symbol, 
                   ha='center', va='bottom', fontsize=12, fontweight='bold')
            
            # Add label if provided
            if label:
                ax.text((x1 + x2) / 2, y_line - y_max*0.05, label, 
                       ha='center', va='top', fontsize=8, style='italic')
        
        # Regional comparisons (within each grade)
        if grade_2_3_p_value is not None:
            add_significance_line(ax1, 1, 2, 0.85, grade_2_3_p_value, "Regional")
        
        if grade_4_p_value is not None:
            add_significance_line(ax1, 4, 5, 0.85, grade_4_p_value, "Regional")
        
        # Cross-grade comparisons
        if grade_2_3_vs_4_tumor_p_value is not None:
            add_significance_line(ax1, 1, 4, 0.95, grade_2_3_vs_4_tumor_p_value, "Grade")
        
        if grade_2_3_vs_4_opercular_p_value is not None:
            add_significance_line(ax1, 2, 5, 0.95, grade_2_3_vs_4_opercular_p_value, "Grade")
    
    # Plot 2: Pathology Groups (side by side with closer spacing)
    ax2 = axes[0, 1]
    
    # Prepare data for side-by-side box plots with closer spacing
    path_data = []
    path_labels = []
    path_colors = []
    path_positions = []
    
    # IDH-mutant (positions 1, 2)
    if len(idh_mut_non_opercular) > 0:
        path_data.append(idh_mut_non_opercular)
        path_labels.append(f'IDH-mutant\nTumor\n(n={len(idh_mut_non_opercular)})')
        path_colors.append('#1f77b4')  # Blue
        path_positions.append(1)
    
    if len(idh_mut_opercular) > 0:
        path_data.append(idh_mut_opercular)
        path_labels.append(f'IDH-mutant\nOpercular\n(n={len(idh_mut_opercular)})')
        path_colors.append('#87CEEB')  # Light blue
        path_positions.append(2)
    
    # IDH-wildtype (positions 4, 5)
    if len(idh_wt_non_opercular) > 0:
        path_data.append(idh_wt_non_opercular)
        path_labels.append(f'IDH-wildtype\nTumor\n(n={len(idh_wt_non_opercular)})')
        path_colors.append('#d62728')  # Red
        path_positions.append(4)
    
    if len(idh_wt_opercular) > 0:
        path_data.append(idh_wt_opercular)
        path_labels.append(f'IDH-wildtype\nOpercular\n(n={len(idh_wt_opercular)})')
        path_colors.append('#FF6B6B')  # Light red
        path_positions.append(5)
    
    if path_data:
        bp2 = ax2.boxplot(path_data, positions=path_positions, patch_artist=True)
        for patch, color in zip(bp2['boxes'], path_colors):
            patch.set_facecolor(color)
            patch.set_alpha(0.7)

        # ADD INDIVIDUAL DATA POINTS
        for i, (data, pos) in enumerate(zip(path_data, path_positions)):
            # Add small random jitter to x-position to avoid overlap
            jitter = np.random.normal(0, 0.05, len(data))
            ax2.scatter(pos + jitter, data, s=8, c='black', alpha=0.6, zorder=10)
        
        ax2.set_ylabel('Single Neuron Capacity', fontsize=12)
        ax2.set_title('Pathology Groups - Regional Comparison', fontsize=12)
        ax2.grid(True, alpha=0.3)
        
        # Set x-axis labels and ticks
        ax2.set_xticks([1.5, 4.5])
        ax2.set_xticklabels(['IDH-mutant', 'IDH-wildtype'])
        
        # Add legend
        legend_elements = [Patch(facecolor='#1f77b4', alpha=0.7, label='Tumor'),
                          Patch(facecolor='#87CEEB', alpha=0.7, label='Opercular')]
        ax2.legend(handles=legend_elements, loc='upper right')
        
        # Add statistical annotations
        # IDH-mutant significance
        if idh_mut_p_value is not None:
            add_significance_line(ax2, 1, 2, 0.85, idh_mut_p_value, "Regional")
        
        # IDH-wildtype significance
        if idh_wt_p_value is not None:
            add_significance_line(ax2, 4, 5, 0.85, idh_wt_p_value, "Regional")
        
        # Cross-pathology comparisons
        if idh_mut_vs_wt_tumor_p_value is not None:
            add_significance_line(ax2, 1, 4, 0.95, idh_mut_vs_wt_tumor_p_value, "Pathology")
        
        if idh_mut_vs_wt_opercular_p_value is not None:
            add_significance_line(ax2, 2, 5, 0.95, idh_mut_vs_wt_opercular_p_value, "Pathology")
    
    # Plot 3: Grade Distribution Histograms (Non-opercular only)
    ax3 = axes[1, 0]
    
    # Only non-opercular grades (no grade 0)
    if len(grade_2_3_non_opercular) > 0:
        ax3.hist(grade_2_3_non_opercular, bins=15, alpha=0.6, label='Grade 2+3 Tumor', color='#ff7f0e', density=True)
    if len(grade_4_non_opercular) > 0:
        ax3.hist(grade_4_non_opercular, bins=15, alpha=0.6, label='Grade 4 Tumor', color='#d62728', density=True)
    
    ax3.set_xlabel('Single Neuron Capacity', fontsize=12)
    ax3.set_ylabel('Density', fontsize=12)
    ax3.set_title('Grade Distribution Histograms (Tumor Only)', fontsize=12)
    ax3.legend()
    ax3.grid(True, alpha=0.3)
    
    # Plot 4: Pathology Distribution Histograms (Non-opercular only)
    ax4 = axes[1, 1]
    
    # Only non-opercular pathology
    if len(idh_mut_non_opercular) > 0:
        ax4.hist(idh_mut_non_opercular, bins=15, alpha=0.6, label='IDH-mutant Tumor', color='#1f77b4', density=True)
    if len(idh_wt_non_opercular) > 0:
        ax4.hist(idh_wt_non_opercular, bins=15, alpha=0.6, label='IDH-wildtype Tumor', color='#d62728', density=True)
    
    ax4.set_xlabel('Single Neuron Capacity', fontsize=12)
    ax4.set_ylabel('Density', fontsize=12)
    ax4.set_title('Pathology Distribution Histograms (Tumor Only)', fontsize=12)
    ax4.legend()
    ax4.grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.savefig("neuron_information_summary.svg", format="svg", dpi=300, bbox_inches='tight')
    plt.show()
    
    return {
        'grade_groups': {
            'grade_2_3_non_opercular': grade_2_3_non_opercular,
            'grade_2_3_opercular': grade_2_3_opercular,
            'grade_4_non_opercular': grade_4_non_opercular,
            'grade_4_opercular': grade_4_opercular
        },
        'pathology_groups': {

            'idh_mutant_non_opercular': idh_mut_non_opercular,
            'idh_mutant_opercular': idh_mut_opercular,
            'idh_wildtype_non_opercular': idh_wt_non_opercular,
            'idh_wildtype_opercular': idh_wt_opercular
        }
    }

# Run the comparison for single neuron capacity
# Example usage:
# results_single_neuron = compare_single_neuron_capacity_groups(stat_method='parametric')
# results_single_neuron = compare_single_neuron_capacity_groups(stat_method='nonparametric')
results_single_neuron = compare_single_neuron_capacity_groups(stat_method='parametric')
```

```python
# SHUFFLE CONTROL
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

plt.rcParams['svg.fonttype'] = 'none' 

# Scenarios supported (dataset_type in {'neuron','method4'})
SCENARIOS = {
    # Single neuron
    'neuron_grade23_tumor_vs_opercular': {
        'dataset': 'neuron', 'base': {'grade_in': [2, 3]}, 'groupA': {'opercular': 0}, 'groupB': {'opercular': 1},
        'label': 'Single Neuron • Grade 2+3 • Tumor vs Opercular', 'y_label': 'Single Neuron Capacity'
    },
    'neuron_grade4_tumor_vs_opercular': {
        'dataset': 'neuron', 'base': {'grade_in': [4]}, 'groupA': {'opercular': 0}, 'groupB': {'opercular': 1},
        'label': 'Single Neuron • Grade 4 • Tumor vs Opercular', 'y_label': 'Single Neuron Capacity'
    },
    'neuron_idhmut_tumor_vs_opercular': {
        'dataset': 'neuron', 'base': {'path_in': ['ast', 'oli']}, 'groupA': {'opercular': 0}, 'groupB': {'opercular': 1},
        'label': 'Single Neuron • IDH-mut • Tumor vs Opercular', 'y_label': 'Single Neuron Capacity'
    },
    'neuron_idhwt_tumor_vs_opercular': {
        'dataset': 'neuron', 'base': {'path_in': ['gbm']}, 'groupA': {'opercular': 0}, 'groupB': {'opercular': 1},
        'label': 'Single Neuron • IDH-wt • Tumor vs Opercular', 'y_label': 'Single Neuron Capacity'
    },
    'neuron_grade23_opercular_vs_grade4_opercular': {
        'dataset': 'neuron', 'base': {'opercular': 1}, 'groupA': {'grade_in': [2, 3]}, 'groupB': {'grade_in': [4]},
        'label': 'Single Neuron • Opercular • Grade 2+3 vs Grade 4', 'y_label': 'Single Neuron Capacity'
    },
    'neuron_idhmut_opercular_vs_idhwt_opercular': {
        'dataset': 'neuron', 'base': {'opercular': 1}, 'groupA': {'path_in': ['ast', 'oli']}, 'groupB': {'path_in': ['gbm']},
        'label': 'Single Neuron • Opercular • IDH-mut vs IDH-wt', 'y_label': 'Single Neuron Capacity'
    },
    # Global Method 4
    'method4_grade23_tumor_vs_opercular': {
        'dataset': 'method4', 'base': {'grade_in': [2, 3]}, 'groupA': {'opercular': 0}, 'groupB': {'opercular': 1},
        'label': 'Global Method 4 • Grade 2+3 • Tumor vs Opercular', 'y_label': 'Global Method 4 Capacity'
    },
    'method4_grade4_tumor_vs_opercular': {
        'dataset': 'method4', 'base': {'grade_in': [4]}, 'groupA': {'opercular': 0}, 'groupB': {'opercular': 1},
        'label': 'Global Method 4 • Grade 4 • Tumor vs Opercular', 'y_label': 'Global Method 4 Capacity'
    },
    'method4_idhmut_tumor_vs_opercular': {
        'dataset': 'method4', 'base': {'path_in': ['ast', 'oli']}, 'groupA': {'opercular': 0}, 'groupB': {'opercular': 1},
        'label': 'Global Method 4 • IDH-mut • Tumor vs Opercular', 'y_label': 'Global Method 4 Capacity'
    },
    'method4_idhwt_tumor_vs_opercular': {
        'dataset': 'method4', 'base': {'path_in': ['gbm']}, 'groupA': {'opercular': 0}, 'groupB': {'opercular': 1},
        'label': 'Global Method 4 • IDH-wt • Tumor vs Opercular', 'y_label': 'Global Method 4 Capacity'
    },
    'method4_grade23_opercular_vs_grade4_opercular': {
        'dataset': 'method4', 'base': {'opercular': 1}, 'groupA': {'grade_in': [2, 3]}, 'groupB': {'grade_in': [4]},
        'label': 'Global Method 4 • Opercular • Grade 2+3 vs Grade 4', 'y_label': 'Global Method 4 Capacity'
    },
    'method4_idhmut_opercular_vs_idhwt_opercular': {
        'dataset': 'method4', 'base': {'opercular': 1}, 'groupA': {'path_in': ['ast', 'oli']}, 'groupB': {'path_in': ['gbm']},
        'label': 'Global Method 4 • Opercular • IDH-mut vs IDH-wt', 'y_label': 'Global Method 4 Capacity'
    },
    # Single neuron - tumor only comparisons
    'neuron_tumor_grade23_vs_grade4': {
        'dataset': 'neuron', 'base': {'opercular': 0}, 'groupA': {'grade_in': [2, 3]}, 'groupB': {'grade_in': [4]},
        'label': 'Single Neuron • Tumor • Grade 2+3 vs Grade 4', 'y_label': 'Single Neuron Capacity'
    },
    'neuron_tumor_idhmut_vs_idhwt': {
        'dataset': 'neuron', 'base': {'opercular': 0}, 'groupA': {'path_in': ['ast', 'oli']}, 'groupB': {'path_in': ['gbm']},
        'label': 'Single Neuron • Tumor • IDH-mut vs IDH-wt', 'y_label': 'Single Neuron Capacity'
    },
    # Global Method 4 - tumor only comparisons
    'method4_tumor_grade23_vs_grade4': {
        'dataset': 'method4', 'base': {'opercular': 0}, 'groupA': {'grade_in': [2, 3]}, 'groupB': {'grade_in': [4]},
        'label': 'Global Method 4 • Tumor • Grade 2+3 vs Grade 4', 'y_label': 'Global Method 4 Capacity'
    },
    'method4_tumor_idhmut_vs_idhwt': {
        'dataset': 'method4', 'base': {'opercular': 0}, 'groupA': {'path_in': ['ast', 'oli']}, 'groupB': {'path_in': ['gbm']},
        'label': 'Global Method 4 • Tumor • IDH-mut vs IDH-wt', 'y_label': 'Global Method 4 Capacity'
    },
}

def _extract_dataset(dataset):
    """Return arrays: capacity, grade, pathology, opercular for requested dataset."""
    capacities = []
    grades = []
    paths = []
    operculars = []
    if dataset == 'neuron':
        for insertion_idx in range(len(neuron_info_capacity_all_list)):
            insertion_data = neuron_info_capacity_all_list[insertion_idx]
            if not isinstance(insertion_data, dict) or len(insertion_data) == 0:
                continue
            grade = grade_list[insertion_idx]
            path = path_list[insertion_idx]
            opercular_type = opercular_list[insertion_idx]
            for _, neuron_data in insertion_data.items():
                if isinstance(neuron_data, dict) and 'capacity' in neuron_data:
                    try:
                        cap = float(neuron_data['capacity'])
                        if not np.isnan(cap):
                            capacities.append(cap)
                            grades.append(grade)
                            paths.append(path)
                            operculars.append(opercular_type)
                    except (ValueError, TypeError):
                        continue
    elif dataset == 'method4':
        for info in all_assembly_info:
            insertion_idx = info['insertion_idx']
            assembly_key = info['assembly_key']
            if insertion_idx in global_method4_by_insertion and assembly_key in global_method4_by_insertion[insertion_idx]:
                cap = global_method4_by_insertion[insertion_idx][assembly_key]
                grade = info['grade']
                path = info['pathology']
                opercular_type = info['opercular']
                if not np.isnan(cap):
                    capacities.append(float(cap))
                    grades.append(grade)
                    paths.append(path)
                    operculars.append(opercular_type)
    return np.array(capacities), np.array(grades), np.array(paths), np.array(operculars)

def _mask_from_criteria(grades, paths, operculars, criteria):
    """Build boolean mask from criteria dict like {'grade_in': [...], 'path_in': [...], 'opercular': 0/1}"""
    if criteria is None:
        return np.ones_like(grades, dtype=bool)
    m = np.ones_like(grades, dtype=bool)
    if 'grade_in' in criteria:
        m &= np.isin(grades, criteria['grade_in'])
    if 'path_in' in criteria:
        m &= np.isin(paths, criteria['path_in'])
    if 'opercular' in criteria:
        m &= (operculars == criteria['opercular'])
    return m

def run_flexible_shuffle(scenario_key, n_shuffles=10000, seed=42):
    if scenario_key not in SCENARIOS:
        raise ValueError(f"Unknown scenario_key. Choose one of: {list(SCENARIOS.keys())}")
    cfg = SCENARIOS[scenario_key]
    dataset = cfg['dataset']
    label = cfg['label']
    y_label = cfg['y_label']

    np.random.seed(seed)
    rng = np.random.default_rng(seed)

    # Load dataset
    cap, grades, paths, operculars = _extract_dataset(dataset)

    # Base subset mask
    base_mask = _mask_from_criteria(grades, paths, operculars, cfg.get('base'))

    # Group membership within base subset
    groupA_mask = base_mask & _mask_from_criteria(grades, paths, operculars, cfg.get('groupA'))
    groupB_mask = base_mask & _mask_from_criteria(grades, paths, operculars, cfg.get('groupB'))

    # Sanity check sizes
    nA, nB = int(groupA_mask.sum()), int(groupB_mask.sum())
    print(label)
    print('-' * 80)
    print(f"nA={nA}, nB={nB}")
    if nA == 0 or nB == 0:
        raise ValueError('One of the groups is empty after filtering. Adjust scenario or check data.')

    # Real t-stat (one-sided A > B)
    t_real, _ = stats.ttest_ind(cap[groupA_mask], cap[groupB_mask], alternative='greater')
    print(f"Real t = {t_real:.4f}")

    # Build base indices and group labels (1 for A, 0 for B) for permutation within base
    base_idx = np.where(base_mask)[0]
    base_group_labels = np.zeros(base_idx.shape[0], dtype=int)
    # Map base idx to A/B
    base_group_labels[np.isin(base_idx, np.where(groupA_mask)[0])] = 1
    base_group_labels[np.isin(base_idx, np.where(groupB_mask)[0])] = 0

    # Ensure counts preserved
    nA_base = int(base_group_labels.sum())
    nB_base = base_group_labels.size - nA_base

    # Shuffle
    t_shuff = np.empty(n_shuffles)
    base_cap = cap[base_idx]
    for i in range(n_shuffles):
        perm = rng.permutation(base_group_labels)
        # Compose groups by permuted labels
        A_perm = base_cap[perm == 1]
        B_perm = base_cap[perm == 0]
        # Guard in case rare all-equal by chance (should not happen with preserved counts)
        if A_perm.size == 0 or B_perm.size == 0:
            t_shuff[i] = np.nan
        else:
            t_shuff[i], _ = stats.ttest_ind(A_perm, B_perm, alternative='greater')

    t_shuff = t_shuff[~np.isnan(t_shuff)]
    p_shuffle = np.mean(t_shuff >= t_real)
    print(f"Shuffle p (one-sided A>B) = {p_shuffle:.4f}")

    # Plot
    fig, axes = plt.subplots(2, 1, figsize=(6, 9))

    ax = axes[0]
    ax.hist(t_shuff, bins=60, color='lightblue', density=True)
    ax.axvline(t_real, color='red', lw=2, label=f"real t = {t_real:.3f}")
    ax.set_title(f"Shuffle t distribution\np = {p_shuffle:.4f}")
    ax.set_xlabel('t-statistic')
    ax.set_ylabel('Density')
    ax.legend()

    ax = axes[1]
    dataA = cap[groupA_mask]
    dataB = cap[groupB_mask]
    labels2 = ['Group A', 'Group B']
    bp = ax.boxplot([dataA, dataB], tick_labels=labels2, patch_artist=True)
    bp['boxes'][0].set_facecolor('#ff7f0e')
    bp['boxes'][1].set_facecolor('#87CEEB')
    ax.set_ylabel(y_label)
    ax.set_title(label)
    ax.grid(True, alpha=0.3)

    plt.tight_layout()
    plt.savefig(f"{runtype}_information_summary.svg", format="svg", dpi=300, bbox_inches='tight')
    plt.show()

    return {
        't_real': t_real,
        't_shuff': t_shuff,
        'p_shuffle': p_shuffle,
        'nA': nA,
        'nB': nB,
        'label': label,
        'dataset': dataset
    }

# USAGE EXAMPLES (uncomment one to run)
### runtype = 'neuron_grade23_tumor_vs_opercular'
### runtype = 'neuron_grade4_tumor_vs_opercular'
### runtype = 'neuron_idhmut_tumor_vs_opercular'
### runtype = 'neuron_idhwt_tumor_vs_opercular'
### runtype = 'method4_grade23_tumor_vs_opercular'
### runtype = 'method4_grade4_tumor_vs_opercular'
### runtype = 'method4_idhmut_tumor_vs_opercular'
### runtype = 'method4_idhwt_tumor_vs_opercular'

runtype = 'method4_tumor_grade23_vs_grade4'  
#runtype = 'neuron_tumor_grade23_vs_grade4'  

#runtype = 'method4_tumor_idhmut_vs_idhwt'       
#runtype = 'neuron_tumor_idhmut_vs_idhwt' 

#runtype = 'method4_grade23_opercular_vs_grade4_opercular'
#runtype = 'neuron_grade23_opercular_vs_grade4_opercular'

#runtype = 'method4_idhmut_opercular_vs_idhwt_opercular'
#runtype = 'neuron_idhmut_opercular_vs_idhwt_opercular'       
   

run_flexible_shuffle(runtype)

```
