```python
import fnmatch
import re
import numpy as np
import pandas as pd

def imec_key_sorter(key):
    match = re.search(r"imec(\d+)", key)
    return int(match.group(1)) if match else float("inf")

def get_imec_keys(data, nwb_path):
    """Match insertion-selection logic from the neural analysis loop."""
    keys = list(data.keys())
    keys = [k for k in keys if fnmatch.fnmatch(k, "*imec*")]
    ks_keys = [k for k in keys if "KS4" in k]
    if ks_keys:
        keys = ks_keys
    th8_keys = [k for k in keys if "Th=8" in k]
    if th8_keys:
        keys = th8_keys
    else:
        th_keys = [k for k in keys if "Th=" in k]
        if th_keys:
            keys = th_keys
    keys = [
        k for k in keys
        if not fnmatch.fnmatch(k, "*sentgen*") and not fnmatch.fnmatch(k, "*_auto*")
    ]
    if "NP137" in str(nwb_path) or "NP139_B2" in str(nwb_path):
        keys = [k for k in keys if "imec1" not in k]
    return sorted(keys, key=imec_key_sorter)

# Collect behavioral trial info (start, end, duration) across all NWB files.
prod_trials = []
sens_trials = []
insertion = 0

for i in range(len(nwb_paths)):
    data = nap.load_file(nwb_paths[i])
    patient_id = nwb_paths[i].stem
    imec_keys = get_imec_keys(data, nwb_paths[i])
    insertion_ids = list(range(insertion, insertion + len(imec_keys)))
    insertion += len(imec_keys)
    print(f"{patient_id}  insertions {insertion_ids}  ({', '.join(imec_keys)})")

    def _append_trials(trial_list, onsets, offsets, trial_type):
        for s_t, e_t in zip(onsets, offsets):
            trial_list.append({
                "patient_id": patient_id,
                "nwb_index": i,
                "insertion_id": insertion_ids[0] if insertion_ids else None,
                "insertion_ids": insertion_ids,
                "trial_type": trial_type,
                "start": float(s_t),
                "end": float(e_t),
                "duration": float(e_t) - float(s_t),
            })

    if "ProdSpeechWords" in data:
        speech_prod_times = data["ProdSpeechWords"]
        prod_onsets = np.asarray(speech_prod_times.start, dtype=float).ravel()
        prod_offsets = np.asarray(speech_prod_times.end, dtype=float).ravel()
        _append_trials(prod_trials, prod_onsets, prod_offsets, "speech_production")
        print(f"  prod: {len(prod_onsets)} trials  (insertion_ids={insertion_ids})")
    else:
        print("  prod: no ProdSpeechWords, skipping")

    if ("StimSpeechWords" in data) or ("mfa_stim_words" in data):
        speech_sens_times = data["StimSpeechWords"] if "StimSpeechWords" in data else data["mfa_stim_words"]
        sens_onsets = np.asarray(speech_sens_times.start, dtype=float).ravel()
        sens_offsets = np.asarray(speech_sens_times.end, dtype=float).ravel()
        _append_trials(sens_trials, sens_onsets, sens_offsets, "speech_reception")
        print(f"  sens: {len(sens_onsets)} trials  (insertion_ids={insertion_ids})")
    else:
        print("  sens: no StimSpeechWords/mfa_stim_words, skipping")

prod_trials = pd.DataFrame(prod_trials)
sens_trials = pd.DataFrame(sens_trials)
print(f"\nTotal prod trial rows: {len(prod_trials)}")
print(f"Total sens trial rows: {len(sens_trials)}")
```

```python
def patient_duration_summary(trials_df, label):
    """One row per patient: trial count, duration range, median (seconds)."""
    if trials_df.empty:
        return pd.DataFrame(columns=["patient_id", "n_trials", "duration_min", "duration_max", "duration_median"])

    per_patient = (
        trials_df.groupby("patient_id", as_index=False)
        .agg(
            insertion_id=("insertion_id", "first"),
            insertion_ids=("insertion_ids", "first"),
            n_trials=("duration", "count"),
            duration_min=("duration", "min"),
            duration_max=("duration", "max"),
            duration_median=("duration", "median"),
        )
        .set_index("patient_id")
    )
    per_patient.index.name = label
    return per_patient

def trial_counts_per_insertion(trials_df, n_insertions):
    """Trial count per insertion; insertions with no events count as 0."""
    counts = pd.Series(0, index=range(n_insertions), dtype=int)
    if trials_df.empty:
        return counts
    per_insertion = (
        trials_df.drop(columns=["insertion_id"])
        .explode("insertion_ids")
        .rename(columns={"insertion_ids": "insertion_id"})
        .groupby("insertion_id")
        .size()
    )
    counts.loc[per_insertion.index] = per_insertion
    return counts

prod_by_patient = patient_duration_summary(prod_trials, "speech_production")
sens_by_patient = patient_duration_summary(sens_trials, "speech_reception")

print("=== Speech production (per patient) ===")
display(prod_by_patient)

print("\n=== Speech reception (per patient) ===")
display(sens_by_patient)

n_insertions = len(manual_exclude_lists)
prod_per_insertion = trial_counts_per_insertion(prod_trials, n_insertions)
sens_per_insertion = trial_counts_per_insertion(sens_trials, n_insertions)

summary = pd.DataFrame({
    "n_events": [
        len(prod_trials),
        len(sens_trials),
    ],
    "mean_duration": [
        prod_trials["duration"].mean() if len(prod_trials) else np.nan,
        sens_trials["duration"].mean() if len(sens_trials) else np.nan,
    ],
    "std_duration": [
        prod_trials["duration"].std() if len(prod_trials) else np.nan,
        sens_trials["duration"].std() if len(sens_trials) else np.nan,
    ],
    "n_trials_min_across_insertions": [
        prod_per_insertion.min(),
        sens_per_insertion.min(),
    ],
    "n_trials_max_across_insertions": [
        prod_per_insertion.max(),
        sens_per_insertion.max(),
    ],
    "n_trials_median_across_insertions": [
        prod_per_insertion.median(),
        sens_per_insertion.median(),
    ],
}, index=["speech_production", "speech_reception"])

print("\n=== Summary ===")
display(summary)
```
