```python
# MAIN LOOP TO GET ASSEMBLY YIELD
import fnmatch
import matplotlib as mpl
import matplotlib.pyplot as plt
import numpy as np
import re
import os
from scipy import stats
from sklearn.decomposition import PCA
from sklearn.decomposition import FastICA
from scipy.linalg import norm
from scipy.stats import kstest

def imec_key_sorter(key):
    # Extract the number after 'imec', default to a large number if not found
    match = re.search(r'imec(\d+)', key)
    return int(match.group(1)) if match else float('inf')

# define variables to save out from loop
assembly_yield_list = []
insertion = 0

# loop through all nwb files
for i in range(len(nwb_paths)):
    data = nap.load_file(nwb_paths[i])
    keys = data.keys()

    # find the spike TsGroup objects, number various based on recording apparatus
    template = "*imec*"
    keys = [key for key in keys if fnmatch.fnmatch(key, template)]
    # Preferentially select keys containing 'KS4'; if none, use all keys
    ks_keys = [key for key in keys if "KS4" in key]
    if ks_keys:
        keys = ks_keys
    # Preferentially select keys containing 'Th='; if none, use all keys
    th8_keys = [key for key in keys if "Th=8" in key]
    if th8_keys:
        keys = th8_keys
    else:
        th_keys = [key for key in keys if "Th=" in key]
        if th_keys:
            keys = th_keys
    # Exclude keys that include 'sentgen' or '_auto'
    template_sentgen = "*sentgen*"
    template_auto = "*_auto*"
    keys = [key for key in keys if not fnmatch.fnmatch(key, template_sentgen) and not fnmatch.fnmatch(key, template_auto)]
    # For NP137, remove 'imec1'
    if 'NP137' in str(nwb_paths[i]):
        keys = [key for key in keys if 'imec1' not in key]
    # Sort keys so that imec0 is first, then imec1, then imec2, etc.
    keys = sorted(keys, key=imec_key_sorter)
    
    # loop through all individual probes
    for s in range(len(keys)):
        print(keys[s])
        spike_times = data[keys[s]]
        
        # 1. filter to good neurons
        firingRates_all = spike_times.metadata["rate"]
        task_times = data["TaskTimes"]
        beh_epochs = nap.IntervalSet(start=task_times.start, end=task_times.end)
        spike_times_beh = spike_times.restrict(beh_epochs)
        firingRates_beh = spike_times_beh.metadata["rate"]

        # calculate ks stats
        start_time = task_times.start
        end_time = task_times.end
        ks_stats = np.zeros(len(spike_times))
        ks_pvals = np.zeros(len(spike_times))
        for u in range(len(spike_times)):
            # get spike times for this unit
            test = spike_times[u].as_series().index.values

            # The null hypothesis is that the spike times are uniformly distributed between start_time and end_time
            if len(test) > 1:
                min_time = start_time[0]
                max_time = end_time[-1]
                # Normalize spike times to [0, 1] for KS test against uniform
                normalized_spike_times = (test - min_time) / (max_time - min_time)
                ks_result = kstest(normalized_spike_times, 'uniform')
                ks_stats[u] = ks_result.statistic
                ks_pvals[u] = ks_result.pvalue
            else:
                ks_stats[u] = np.nan
                ks_pvals[u] = np.nan

        # now do ISI calculation
        violationThreshold = 3/1000 #for 3 ms isi refractory period, what used before, middle of range that kilosort cites
        violationPct = np.zeros(len(spike_times))
        for u in range(len(spike_times)):
            unit = spike_times[u]
            unit = unit.as_series().index
            if len(unit) < 100:
                violationPct[u] = 1 
            else:
                isi = unit.diff()[1:len(unit)]
                violations = np.where(isi<violationThreshold)
                violations = np.array(violations)
                violationPct[u] = violations.size/len(isi)

        # now apply masks
        if "KSLabel" in spike_times.metadata:
            KSLabels = spike_times.metadata["KSLabel"]
        else:
            KSLabels = spike_times.metadata["quality"]
        firingRates = firingRates_beh
        mask1 = violationPct<3/100 # throwing out neurons with > 3% isi violations with isi = 3 ms; lose quite a bit with 5 ms esp when including cdf stats
        mask2 = firingRates>0.5 # firing rate > 0.5 ms during all behavior epochs
        mask3 = KSLabels!="noise" # not noise per auto label
        mask4 = ks_stats < 0.3 # KS stat < 0.3 (stability across recording)
        mask = mask1 & mask2 & mask3 & mask4
        indicesFinal = firingRates.index[mask]
        
        # now exclude neurons manually rejected
        indicesFinal = np.setdiff1d(indicesFinal, manual_exclude_lists[insertion])
        insertion+=1
        spike_times_good = spike_times[indicesFinal]
        print(f"Subject: {i}, insertion: {s}, Number of good neurons: {len(spike_times_good)}")



        # 2.  make spike rate matrix
        # make spike count matrix
        timescale = 25/1000;
        spikeCountMatrix = spike_times_good.count(bin_size=timescale)
        spikeCountMatrix = spikeCountMatrix.values # matrix in spike counts
        firingRateMatrix = spikeCountMatrix/timescale # matrix in Hz
        firingRateMatrix = stats.zscore(firingRateMatrix, axis=0) # zscore across time (within neuron)



        # 3. run PCA to get first pass assembly patterns
        pca = PCA()

        # fit and transform data
        # note don't have to center bc each row is zscores so center of matrix is 0
        firingRateMatrix_pca = pca.fit_transform(firingRateMatrix)
        eigenvalues = pca.explained_variance_ 

        # calculate the upperbound of the Marchenko-Pastur law, eigenvalues above the 
        # upper bound are patterns that will be considered "assemblies"
        upperbound = (1+np.sqrt(firingRateMatrix.shape[1]/firingRateMatrix.shape[0]))**2

        # classic definition
        assemblyIndices = np.where(eigenvalues>upperbound)

        # more stringent definition
        # assemblyIndices = np.where((eigenvalues > upperbound) & (eigenvalues > np.mean(eigenvalues) + np.std(eigenvalues) * 2))

        # save assembly eigenvalues
        assemblyEigenvalues = eigenvalues[assemblyIndices]



        # 4. orthogonalize the assembly patterns with ICA
        # only get the assembly pca patterns
        n_pcs = len(assemblyEigenvalues) # only want the above threshold patterns
        pc_vectors = pca.components_[:n_pcs, :]
        print(f"Number of assembly patterns: {n_pcs}")

        # project the assembly pca patterns onto the firing rate matrix
        projections = firingRateMatrix @ pc_vectors.T

        # define and run the ICA
        if n_pcs > 0:
            fastica = FastICA(
                n_components=n_pcs,
                algorithm='parallel',
                whiten='unit-variance',
                max_iter=500,
                tol=1e-7,
                random_state=1
            )
            ica_components = fastica.fit_transform(projections)

            # get mixing and unmixing matrices
            mixing_matrix = fastica.mixing_
            unmixing_matrix = fastica.components_

            # 5. project the assembly patterns back to the original feature space (neurons)
            ica_assembly_patterns = unmixing_matrix @ pc_vectors  # Shape: (n_pcs, n_features)

            # define members, save out unit vectors and bool in matrices
            ap_norm_matrix = ica_assembly_patterns
            ap_norm_member = np.zeros(ica_assembly_patterns.shape)
            for ii in range(n_pcs):
                ap = ica_assembly_patterns[ii,:]
                ap_norm = ap/norm(ap)
                maxW = max(ap_norm);
                minW = min(ap_norm);
                if abs(minW) > maxW:
                    ap_norm = ap_norm*-1
                ap_norm_matrix[ii,:] = ap_norm
                threshold = np.mean(ap_norm) + np.std(ap_norm) * 2
                for c in range(len(ap)):
                    if ap_norm[c] > threshold:
                        ap_norm_member[ii,c] = 1

        # 7. save out count
        assembly_yield_list.append(len(assemblyEigenvalues))
```

```python
# DEFINE NEURON YIELD INDEPENDANT ASSEMBLY YIELD VARIABLE - PT. 1
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.metrics import r2_score
import pandas as pd

# Convert to numpy arrays for easier manipulation
yield_arr = np.array(yield_list)
assembly_yield_arr = np.array(assembly_yield_list)

# Remove invalid entries (NaN, zero yield)
valid_mask = (
    ~np.isnan(yield_arr) & 
    ~np.isnan(assembly_yield_arr) & 
    (yield_arr > 0)
)

yield_valid = yield_arr[valid_mask]
assembly_yield_valid = assembly_yield_arr[valid_mask]

print(f"Working with {len(yield_valid)} valid recordings")
print(f"Yield range: {yield_valid.min():.0f} - {yield_valid.max():.0f}")
print(f"Assembly yield range: {assembly_yield_valid.min():.0f} - {assembly_yield_valid.max():.0f}")

# Method 1: Linear Regression Residuals
# Fit linear regression: assembly_yield ~ yield
X_linear = yield_valid.reshape(-1, 1)
y_linear = assembly_yield_valid

linear_model = LinearRegression()
linear_model.fit(X_linear, y_linear)
y_pred_linear = linear_model.predict(X_linear)
linear_residuals = y_linear - y_pred_linear

# Method 2: Polynomial Regression Residuals (quadratic)
# Fit polynomial regression: assembly_yield ~ yield + yield^2
poly_features = PolynomialFeatures(degree=2, include_bias=False)
X_poly = poly_features.fit_transform(yield_valid.reshape(-1, 1))

poly_model = LinearRegression()
poly_model.fit(X_poly, y_linear)
y_pred_poly = poly_model.predict(X_poly)
poly_residuals = y_linear - y_pred_poly

# Method 3: Log-transformed Regression
# Fit regression on log-transformed data: log(assembly_yield) ~ log(yield)
log_yield = np.log(yield_valid)
log_assembly_yield = np.log(assembly_yield_valid + 1)  # +1 to avoid log(0)

X_log = log_yield.reshape(-1, 1)
log_model = LinearRegression()
log_model.fit(X_log, log_assembly_yield)
y_pred_log = log_model.predict(X_log)
log_residuals = log_assembly_yield - y_pred_log

# Method 4: Z-score normalization within yield bins
# Create yield bins and normalize assembly_yield within each bin
n_bins = 5
yield_bins = np.percentile(yield_valid, np.linspace(0, 100, n_bins + 1))
bin_indices = np.digitize(yield_valid, yield_bins) - 1
bin_indices = np.clip(bin_indices, 0, n_bins - 1)

zscore_normalized = np.zeros_like(assembly_yield_valid)
for bin_idx in range(n_bins):
    mask = bin_indices == bin_idx
    if np.sum(mask) > 1:  # Need at least 2 points for z-score
        bin_values = assembly_yield_valid[mask]
        zscore_normalized[mask] = (bin_values - np.mean(bin_values)) / np.std(bin_values)

# Method 5: Percentile ranking within yield bins
percentile_ranked = np.zeros_like(assembly_yield_valid)
for bin_idx in range(n_bins):
    mask = bin_indices == bin_idx
    if np.sum(mask) > 1:
        bin_values = assembly_yield_valid[mask]
        percentile_ranked[mask] = stats.rankdata(bin_values) / len(bin_values)

# Create comprehensive results dataframe
results_df = pd.DataFrame({
    'yield': yield_valid,
    'assembly_yield': assembly_yield_valid,
    'assembly_yield_per_yield': assembly_yield_valid / yield_valid,
    'linear_residuals': linear_residuals,
    'poly_residuals': poly_residuals,
    'log_residuals': log_residuals,
    'zscore_normalized': zscore_normalized,
    'percentile_ranked': percentile_ranked,
    'yield_bin': bin_indices
})

print("\n" + "="*80)
print("ASSEMBLY YIELD INDEPENDENCE METHODS SUMMARY")
print("="*80)
print(f"Linear regression R²: {r2_score(y_linear, y_pred_linear):.4f}")
print(f"Polynomial regression R²: {r2_score(y_linear, y_pred_poly):.4f}")
print(f"Log regression R²: {r2_score(log_assembly_yield, y_pred_log):.4f}")

# Calculate correlations with yield for each method
methods = ['assembly_yield_per_yield', 'linear_residuals', 'poly_residuals', 
           'log_residuals', 'zscore_normalized', 'percentile_ranked']

print(f"\n{'Method':<25} {'Correlation with yield':<20} {'Interpretation'}")
print("-" * 70)
for method in methods:
    corr, pval = stats.pearsonr(yield_valid, results_df[method])
    if abs(corr) < 0.1:
        interpretation = "✅ Independent"
    elif abs(corr) < 0.3:
        interpretation = "⚠️  Weakly dependent"
    else:
        interpretation = "❌ Strongly dependent"
    print(f"{method:<25} {corr:>6.3f} (p={pval:.3f}){'':>8} {interpretation}")

print("="*80)
```

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import ttest_ind, f_oneway, mannwhitneyu
import itertools

plt.rcParams['svg.fonttype'] = 'none'  # Ensures text is editable in Illustrator

# --- Adjustable aspect ratio parameter ---
pbaspect = (0.5, 1)  # (width, height) ratio; change as desired, e.g., (2,1), (1,1), etc.

# Filter data for valid entries and apply yield threshold
yield_thresh = 1
valid_for_analysis = (
    ~np.isnan(independent_assembly_yield) & 
    (np.array(yield_list) > yield_thresh)
)

independent_values = independent_assembly_yield[valid_for_analysis]
grade_values = np.array(grade_list)[valid_for_analysis]

# Get unique grades and sort them
unique_grades = sorted(np.unique(grade_values))
print(f"Grades present in data: {unique_grades}")

# Group data by grade
grade_data = []
means = []
stds = []
counts = []

for grade in unique_grades:
    grade_mask = grade_values == grade
    grade_subset = independent_values[grade_mask]
    grade_data.append(grade_subset)
    means.append(np.mean(grade_subset))
    stds.append(np.std(grade_subset))
    counts.append(len(grade_subset))

# --- ADDITIONAL PLOT: Only opercular = 0 or 2 ---
# Group grade 2 and 3 together as "2/3"

# Filter for opercular = 0 or 2
opercular_arr = np.array(opercular_list)[valid_for_analysis]
mask_opercular_0_2 = np.isin(opercular_arr, [0, 2])

independent_values_02 = independent_values[mask_opercular_0_2]
grade_values_02 = grade_values[mask_opercular_0_2]
opercular_values_02 = opercular_arr[mask_opercular_0_2]

# Map grades: group 2 and 3 together as 2
grade_values_02_grouped = np.copy(grade_values_02)
grade_values_02_grouped[(grade_values_02 == 3)] = 2

# Get unique grades present in this subset (after grouping)
unique_grades_02 = sorted(np.unique(grade_values_02_grouped))
# For display, use '2/3' instead of 2 if both 2 and 3 were present in original data
display_grades_02 = []
if 2 in unique_grades_02:
    # Check if both 2 and 3 were present in original data
    if (2 in grade_values_02) and (3 in grade_values_02):
        display_grades_02 = [g if g != 2 else '2/3' for g in unique_grades_02]
    else:
        display_grades_02 = [str(g) for g in unique_grades_02]
else:
    display_grades_02 = [str(g) for g in unique_grades_02]

print(f"\nGrades present in data (opercular=0 or 2, grouped 2/3): {display_grades_02}")

# Group data by grade for opercular=0 or 2 (with 2/3 grouped)
grade_data_02 = []
means_02 = []
stds_02 = []
counts_02 = []

for grade in unique_grades_02:
    grade_mask = grade_values_02_grouped == grade
    grade_subset = independent_values_02[grade_mask]
    grade_data_02.append(grade_subset)
    means_02.append(np.mean(grade_subset))
    stds_02.append(np.std(grade_subset))
    counts_02.append(len(grade_subset))

# --- Overall significance: One-way ANOVA across groups ---
fig2 = plt.figure(figsize=(10, 6))
ax2 = plt.gca()

if len(grade_data_02) > 1 and all(len(vals) > 0 for vals in grade_data_02):
    anova_stat_02, anova_pval_02 = f_oneway(*grade_data_02)
    print(f"\nOne-way ANOVA across grades (opercular=0 or 2, grouped 2/3): F = {anova_stat_02:.3f}, p = {anova_pval_02:.4g}")
    ax2.text(0.5, 1.05, f"One-way ANOVA: p = {anova_pval_02:.3g}", 
            ha='center', va='bottom', transform=ax2.transAxes, fontsize=13, fontweight='bold')
else:
    print("\nNot enough groups or data for ANOVA (opercular=0 or 2, grouped 2/3).")

# Create bar plot
bar_positions_02 = np.arange(len(unique_grades_02))
bar_width_02 = 0.6

bars_02 = ax2.bar(bar_positions_02, means_02, yerr=stds_02, capsize=7, 
                  color='lightgreen', edgecolor='k', width=bar_width_02, 
                  alpha=0.8, label='Mean ± SD')

# Add individual data points with jitter
for i, (grade, vals) in enumerate(zip(unique_grades_02, grade_data_02)):
    x_jitter = bar_positions_02[i] + np.random.uniform(-0.15, 0.15, size=len(vals))
    ax2.scatter(x_jitter, vals, color='k', alpha=0.7, s=50, zorder=10)

# Customize the plot
ax2.set_xticks(bar_positions_02)
ax2.set_xticklabels([f'Grade {g}' for g in display_grades_02])
ax2.set_xlabel('Grade', fontsize=14, fontweight='bold')
ax2.set_ylabel('Independent Assembly Yield', fontsize=14, fontweight='bold')
ax2.set_title(f'Independent Assembly Yield by Grade (opercular=0 or 2, 2/3 grouped)\n(yield > {yield_thresh})', 
          fontsize=16, fontweight='bold')

# Add sample size annotations on top of bars
for i, (bar, count) in enumerate(zip(bars_02, counts_02)):
    height = bar.get_height()
    ax2.text(bar.get_x() + bar.get_width()/2., height + stds_02[i] + 0.01,
             f'n={count}', ha='center', va='bottom', fontweight='bold')

# Add statistical comparisons
comparisons_02 = list(itertools.combinations(range(len(unique_grades_02)), 2))
y_max_02 = max([np.max(vals) if len(vals) > 0 else 0 for vals in grade_data_02])
y_start_02 = y_max_02 + 0.2 * y_max_02 if y_max_02 > 0 else 1
h_02 = 0.05 * y_max_02 if y_max_02 > 0 else 0.5

print(f"\nStatistical comparisons (Welch's t-test, one-sided t-tests, & Mann-Whitney U, opercular=0 or 2, 2/3 grouped):")
for idx, (i, j) in enumerate(comparisons_02):
    vals1 = grade_data_02[i]
    vals2 = grade_data_02[j]
    if len(vals1) > 0 and len(vals2) > 0:
        label1 = display_grades_02[i]
        label2 = display_grades_02[j]

        # Welch's t-test (two-sided)
        stat_t, pval_t = ttest_ind(vals1, vals2, equal_var=False)
        print(f"  Grade {label1} vs Grade {label2}: t = {stat_t:.3f}, p = {pval_t:.4g} (Welch's t-test, two-sided)")

        # Welch's t-test (one-sided: vals1 > vals2)
        try:
            stat_t_greater, pval_t_greater = ttest_ind(vals1, vals2, equal_var=False, alternative='greater')
            print(f"  Grade {label1} vs Grade {label2} (one-sided, {label1}>{label2}): p = {pval_t_greater:.4g} (t-test, one-sided, greater)")
        except TypeError:
            # For older scipy: fallback to manual one-sided adjustment
            if stat_t > 0:
                pval_t_greater = pval_t / 2
            else:
                pval_t_greater = 1 - pval_t / 2
            print(f"  Grade {label1} vs Grade {label2} (one-sided, {label1}>{label2}): p = {pval_t_greater:.4g} (manual approx)")

        # Welch's t-test (one-sided: vals1 < vals2)
        try:
            stat_t_less, pval_t_less = ttest_ind(vals1, vals2, equal_var=False, alternative='less')
            print(f"  Grade {label1} vs Grade {label2} (one-sided, {label1}<{label2}): p = {pval_t_less:.4g} (t-test, one-sided, less)")
        except TypeError:
            if stat_t < 0:
                pval_t_less = pval_t / 2
            else:
                pval_t_less = 1 - pval_t / 2
            print(f"  Grade {label1} vs Grade {label2} (one-sided, {label1}<{label2}): p = {pval_t_less:.4g} (manual approx)")

        # Mann-Whitney U test
        try:
            stat_u, pval_u = mannwhitneyu(vals1, vals2, alternative='two-sided')
            print(f"  Grade {label1} vs Grade {label2}: U = {stat_u:.3f}, p = {pval_u:.4g} (Mann-Whitney U)")
        except Exception as e:
            print(f"  Mann-Whitney U test failed for Grade {label1} vs Grade {label2}: {e}")
        
        # Add significance lines (for two-sided Welch's t-test only, keep as in original)
        y = y_start_02 + idx * h_02
        x1, x2 = bar_positions_02[i], bar_positions_02[j]
        ax2.plot([x1, x1, x2, x2], [y, y+h_02/4, y+h_02/4, y], lw=1.5, c='k')
        
        # Add significance stars according to two-sided t-test p-value
        if pval_t < 0.001:
            stars = '***'
        elif pval_t < 0.01:
            stars = '**'
        elif pval_t < 0.05:
            stars = '*'
        else:
            stars = 'n.s.'
        ax2.text((x1 + x2) / 2, y + h_02/6, stars, ha='center', va='bottom', 
                color='k', fontsize=12, fontweight='bold')

# Add grid for better readability
ax2.grid(True, alpha=0.3, axis='y')
ax2.set_axisbelow(True)

# Set adjustable aspect ratio
ax2.set_box_aspect(pbaspect[1]/pbaspect[0])  # height/width

# Adjust layout
plt.tight_layout()

# Save the figure as SVG
plt.savefig("independent_assembly_yield_by_grade_opercular_0_2_grouped_2_3.svg", format="svg", dpi=300, bbox_inches='tight')

# Display the plot in notebook output
plt.show()

# Print summary statistics
print(f"\n" + "="*60)
print("SUMMARY STATISTICS (opercular=0 or 2, 2/3 grouped)")
print("="*60)
for i, grade in enumerate(unique_grades_02):
    vals = grade_data_02[i]
    label = display_grades_02[i]
    print(f"Grade {label}: n={len(vals):2d} | mean={means_02[i]:.4f} | std={stds_02[i]:.4f} | range=[{np.min(vals):.4f}, {np.max(vals):.4f}]")
print("="*60)
print("="*60)
print("="*60)
```
