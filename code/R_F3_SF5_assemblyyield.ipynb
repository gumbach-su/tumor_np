```python
# MAIN LOOP TO GET ASSEMBLY YIELD
import fnmatch
import matplotlib as mpl
import matplotlib.pyplot as plt
import numpy as np
import re
import os
from scipy import stats
from sklearn.decomposition import PCA
from sklearn.decomposition import FastICA
from scipy.linalg import norm
from scipy.stats import kstest

MIN_BEHAVIOR_MINUTES = 0

def _behavior_duration_seconds(task_times_obj):
    starts = np.asarray(task_times_obj.start, dtype=float).ravel()
    ends = np.asarray(task_times_obj.end, dtype=float).ravel()
    if starts.size != ends.size:
        return np.nan
    return float(np.sum(ends - starts))

def _ks_time_bounds_full_recording(spike_times_obj):
    """Min/max time for uniform KS (full recording)."""
    tmin, tmax = np.inf, -np.inf
    for u in range(len(spike_times_obj)):
        idx = spike_times_obj[u].as_series().index.values
        if len(idx):
            tmin = min(tmin, float(np.min(idx)))
            tmax = max(tmax, float(np.max(idx)))
    if np.isfinite(tmin) and np.isfinite(tmax):
        return tmin, tmax
    return np.nan, np.nan

def imec_key_sorter(key):
    # Extract the number after 'imec', default to a large number if not found
    match = re.search(r'imec(\d+)', key)
    return int(match.group(1)) if match else float('inf')

# define variables to save out from loop
assembly_yield_list = []
insertion = 0

# loop through all nwb files
for i in range(len(nwb_paths)):
    data = nap.load_file(nwb_paths[i])
    keys = data.keys()

    # find the spike TsGroup objects, number various based on recording apparatus
    template = "*imec*"
    keys = [key for key in keys if fnmatch.fnmatch(key, template)]
    # Preferentially select keys containing 'KS4'; if none, use all keys
    ks_keys = [key for key in keys if "KS4" in key]
    if ks_keys:
        keys = ks_keys
    # Preferentially select keys containing 'Th='; if none, use all keys
    th8_keys = [key for key in keys if "Th=8" in key]
    if th8_keys:
        keys = th8_keys
    else:
        th_keys = [key for key in keys if "Th=" in key]
        if th_keys:
            keys = th_keys
    # Exclude keys that include 'sentgen' or '_auto'
    template_sentgen = "*sentgen*"
    template_auto = "*_auto*"
    keys = [key for key in keys if not fnmatch.fnmatch(key, template_sentgen) and not fnmatch.fnmatch(key, template_auto)]
    # For NP137, remove 'imec1'
    if 'NP137' in str(nwb_paths[i]) or 'NP139_B2' in str(nwb_paths[i]):
        keys = [key for key in keys if 'imec1' not in key]
    # Sort keys so that imec0 is first, then imec1, then imec2, etc.
    keys = sorted(keys, key=imec_key_sorter)
    
    # loop through all individual probes
    for s in range(len(keys)):
        print(keys[s])
        spike_times = data[keys[s]]
        
        # 1. filter to good neurons
        firingRates_all = spike_times.metadata["rate"]
        if "TaskTimes" in data.keys():
            task_times = data["TaskTimes"]
        else:
            task_times = data["task_times"]

        start_time = task_times.start
        end_time = task_times.end
        beh_epochs = nap.IntervalSet(start=task_times.start, end=task_times.end)

        beh_sec = _behavior_duration_seconds(task_times)
        beh_min = beh_sec / 60.0 if np.isfinite(beh_sec) else np.nan
        use_full_recording = (not np.isfinite(beh_min)) or (beh_min < MIN_BEHAVIOR_MINUTES)

        if use_full_recording:
            spike_times_beh = spike_times
            firingRates_beh = firingRates_all
            min_time, max_time = _ks_time_bounds_full_recording(spike_times)
            if not (np.isfinite(min_time) and np.isfinite(max_time)) or (max_time <= min_time):
                print(f"  Short behavior ({beh_min:.2f} min) but could not get recording bounds; skipping shank")
                insertion += 1
                continue
            print(
                f"  Behavioral duration {beh_min:.2f} min (< {MIN_BEHAVIOR_MINUTES} min) — using full recording for filters; KS window [{min_time:.3f}, {max_time:.3f}] s"
            )
        else:
            spike_times_beh = spike_times.restrict(beh_epochs)
            firingRates_beh = spike_times_beh.metadata["rate"]
            min_time = start_time[0]
            max_time = end_time[-1]
            print(
                f"  Behavioral duration {beh_min:.2f} min (>= {MIN_BEHAVIOR_MINUTES} min) — using TaskTimes restrict; KS window [{min_time:.3f}, {max_time:.3f}] s"
            )

        # calculate ks stats
        ks_stats = np.zeros(len(spike_times))
        ks_pvals = np.zeros(len(spike_times))
        for u in range(len(spike_times)):
            # get spike times for this unit
            test = spike_times[u].as_series().index.values

            # The null hypothesis is that the spike times are uniformly distributed between min_time and max_time
            if len(test) > 1:
                # Normalize spike times to [0, 1] for KS test against uniform
                normalized_spike_times = (test - min_time) / (max_time - min_time)
                ks_result = kstest(normalized_spike_times, 'uniform')
                ks_stats[u] = ks_result.statistic
                ks_pvals[u] = ks_result.pvalue
            else:
                ks_stats[u] = np.nan
                ks_pvals[u] = np.nan

        # now do ISI calculation
        violationThreshold = 3/1000 #for 3 ms isi refractory period, what used before, middle of range that kilosort cites
        violationPct = np.zeros(len(spike_times))
        for u in range(len(spike_times)):
            unit = spike_times[u]
            unit = unit.as_series().index
            if len(unit) < 100:
                violationPct[u] = 1 
            else:
                isi = unit.diff()[1:len(unit)]
                violations = np.where(isi<violationThreshold)
                violations = np.array(violations)
                violationPct[u] = violations.size/len(isi)

        # now apply masks
        if "KSLabel" in spike_times.metadata:
            KSLabels = spike_times.metadata["KSLabel"]
        else:
            KSLabels = spike_times.metadata["quality"]
        firingRates = firingRates_beh
        mask1 = violationPct<3/100 # throwing out neurons with > 3% isi violations with isi = 3 ms; lose quite a bit with 5 ms esp when including cdf stats
        mask2 = firingRates>0.5 # firing rate > 0.5 ms during all behavior epochs
        mask3 = KSLabels!="noise" # not noise per auto label
        mask4 = ks_stats < 0.3 # KS stat < 0.3 (stability across recording)
        mask = mask1 & mask2 & mask3 & mask4
        indicesFinal = firingRates.index[mask]
        
        # now exclude neurons manually rejected
        indicesFinal = np.setdiff1d(indicesFinal, manual_exclude_lists[insertion])
        insertion+=1
        spike_times_good = spike_times[indicesFinal]
        print(f"Subject: {i}, insertion: {s}, Number of good neurons: {len(spike_times_good)}")



        # 2.  make spike rate matrix
        # make spike count matrix
        timescale = 25/1000;
        spikeCountMatrix = spike_times_good.count(bin_size=timescale)
        spikeCountMatrix = spikeCountMatrix.values # matrix in spike counts
        firingRateMatrix = spikeCountMatrix/timescale # matrix in Hz
        firingRateMatrix = stats.zscore(firingRateMatrix, axis=0) # zscore across time (within neuron)



        # 3. run PCA to get first pass assembly patterns
        pca = PCA()

        # fit and transform data
        # note don't have to center bc each row is zscores so center of matrix is 0
        firingRateMatrix_pca = pca.fit_transform(firingRateMatrix)
        eigenvalues = pca.explained_variance_ 

        # calculate the upperbound of the Marchenko-Pastur law, eigenvalues above the 
        # upper bound are patterns that will be considered "assemblies"
        upperbound = (1+np.sqrt(firingRateMatrix.shape[1]/firingRateMatrix.shape[0]))**2

        # classic definition
        assemblyIndices = np.where(eigenvalues>upperbound)

        # more stringent definition
        # assemblyIndices = np.where((eigenvalues > upperbound) & (eigenvalues > np.mean(eigenvalues) + np.std(eigenvalues) * 2))

        # save assembly eigenvalues
        assemblyEigenvalues = eigenvalues[assemblyIndices]



        # 4. orthogonalize the assembly patterns with ICA
        # only get the assembly pca patterns
        n_pcs = len(assemblyEigenvalues) # only want the above threshold patterns
        pc_vectors = pca.components_[:n_pcs, :]
        print(f"Number of assembly patterns: {n_pcs}")

        # project the assembly pca patterns onto the firing rate matrix
        projections = firingRateMatrix @ pc_vectors.T

        # define and run the ICA
        if n_pcs > 0:
            fastica = FastICA(
                n_components=n_pcs,
                algorithm='parallel',
                whiten='unit-variance',
                max_iter=500,
                tol=1e-7,
                random_state=1
            )
            ica_components = fastica.fit_transform(projections)

            # get mixing and unmixing matrices
            mixing_matrix = fastica.mixing_
            unmixing_matrix = fastica.components_

            # 5. project the assembly patterns back to the original feature space (neurons)
            ica_assembly_patterns = unmixing_matrix @ pc_vectors  # Shape: (n_pcs, n_features)

            # define members, save out unit vectors and bool in matrices
            ap_norm_matrix = ica_assembly_patterns
            ap_norm_member = np.zeros(ica_assembly_patterns.shape)
            for ii in range(n_pcs):
                ap = ica_assembly_patterns[ii,:]
                ap_norm = ap/norm(ap)
                maxW = max(ap_norm);
                minW = min(ap_norm);
                if abs(minW) > maxW:
                    ap_norm = ap_norm*-1
                ap_norm_matrix[ii,:] = ap_norm
                threshold = np.mean(ap_norm) + np.std(ap_norm) * 2
                for c in range(len(ap)):
                    if ap_norm[c] > threshold:
                        ap_norm_member[ii,c] = 1

        # 7. save out count
        assembly_yield_list.append(len(assemblyEigenvalues))
```

```python
# DEFINE NEURON YIELD INDEPENDANT ASSEMBLY YIELD VARIABLE - PT. 1
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.metrics import r2_score
import pandas as pd

# Convert to numpy arrays for easier manipulation
yield_arr = np.array(yield_list)
assembly_yield_arr = np.array(assembly_yield_list)

# Remove invalid entries (NaN, zero yield)
valid_mask = (
    ~np.isnan(yield_arr) & 
    ~np.isnan(assembly_yield_arr) & 
    (yield_arr > 0)
)

yield_valid = yield_arr[valid_mask]
assembly_yield_valid = assembly_yield_arr[valid_mask]

print(f"Working with {len(yield_valid)} valid recordings")
print(f"Yield range: {yield_valid.min():.0f} - {yield_valid.max():.0f}")
print(f"Assembly yield range: {assembly_yield_valid.min():.0f} - {assembly_yield_valid.max():.0f}")

# Method 1: Linear Regression Residuals
# Fit linear regression: assembly_yield ~ yield
X_linear = yield_valid.reshape(-1, 1)
y_linear = assembly_yield_valid

linear_model = LinearRegression()
linear_model.fit(X_linear, y_linear)
y_pred_linear = linear_model.predict(X_linear)
linear_residuals = y_linear - y_pred_linear

# Method 2: Polynomial Regression Residuals (quadratic)
# Fit polynomial regression: assembly_yield ~ yield + yield^2
poly_features = PolynomialFeatures(degree=2, include_bias=False)
X_poly = poly_features.fit_transform(yield_valid.reshape(-1, 1))

poly_model = LinearRegression()
poly_model.fit(X_poly, y_linear)
y_pred_poly = poly_model.predict(X_poly)
poly_residuals = y_linear - y_pred_poly

# Method 3: Log-transformed Regression
# Fit regression on log-transformed data: log(assembly_yield) ~ log(yield)
log_yield = np.log(yield_valid)
log_assembly_yield = np.log(assembly_yield_valid + 1)  # +1 to avoid log(0)

X_log = log_yield.reshape(-1, 1)
log_model = LinearRegression()
log_model.fit(X_log, log_assembly_yield)
y_pred_log = log_model.predict(X_log)
log_residuals = log_assembly_yield - y_pred_log

# Method 4: Z-score normalization within yield bins
# Create yield bins and normalize assembly_yield within each bin
n_bins = 5
yield_bins = np.percentile(yield_valid, np.linspace(0, 100, n_bins + 1))
bin_indices = np.digitize(yield_valid, yield_bins) - 1
bin_indices = np.clip(bin_indices, 0, n_bins - 1)

zscore_normalized = np.zeros_like(assembly_yield_valid)
for bin_idx in range(n_bins):
    mask = bin_indices == bin_idx
    if np.sum(mask) > 1:  # Need at least 2 points for z-score
        bin_values = assembly_yield_valid[mask]
        zscore_normalized[mask] = (bin_values - np.mean(bin_values)) / np.std(bin_values)

# Method 5: Percentile ranking within yield bins
percentile_ranked = np.zeros_like(assembly_yield_valid)
for bin_idx in range(n_bins):
    mask = bin_indices == bin_idx
    if np.sum(mask) > 1:
        bin_values = assembly_yield_valid[mask]
        percentile_ranked[mask] = stats.rankdata(bin_values) / len(bin_values)

# Create comprehensive results dataframe
results_df = pd.DataFrame({
    'yield': yield_valid,
    'assembly_yield': assembly_yield_valid,
    'assembly_yield_per_yield': assembly_yield_valid / yield_valid,
    'linear_residuals': linear_residuals,
    'poly_residuals': poly_residuals,
    'log_residuals': log_residuals,
    'zscore_normalized': zscore_normalized,
    'percentile_ranked': percentile_ranked,
    'yield_bin': bin_indices
})

print("\n" + "="*80)
print("ASSEMBLY YIELD INDEPENDENCE METHODS SUMMARY")
print("="*80)
print(f"Linear regression R²: {r2_score(y_linear, y_pred_linear):.4f}")
print(f"Polynomial regression R²: {r2_score(y_linear, y_pred_poly):.4f}")
print(f"Log regression R²: {r2_score(log_assembly_yield, y_pred_log):.4f}")

# Calculate correlations with yield for each method
methods = ['assembly_yield_per_yield', 'linear_residuals', 'poly_residuals', 
           'log_residuals', 'zscore_normalized', 'percentile_ranked']

print(f"\n{'Method':<25} {'Correlation with yield':<20} {'Interpretation'}")
print("-" * 70)
for method in methods:
    corr, pval = stats.pearsonr(yield_valid, results_df[method])
    if abs(corr) < 0.1:
        interpretation = "✅ Independent"
    elif abs(corr) < 0.3:
        interpretation = "⚠️  Weakly dependent"
    else:
        interpretation = "❌ Strongly dependent"
    print(f"{method:<25} {corr:>6.3f} (p={pval:.3f}){'':>8} {interpretation}")

print("="*80)
```

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import ttest_ind, f_oneway
import itertools

plt.rcParams['svg.fonttype'] = 'none'  # Ensures text is editable in Illustrator

# Adjustable plot aspect ratio parameter (width, height)
pbaspect = (0.5, 1)  # This controls the aspect ratio (height/width) of the axes
figsize = (10, 6)  # This controls the initial figure size in inches

# Create bar plot of independent assembly yield by opercular condition
fig = plt.figure(figsize=figsize)

# Filter data for valid entries and apply yield threshold
yield_thresh = 1
valid_for_analysis = (
    ~np.isnan(independent_assembly_yield) & 
    (np.array(yield_list) > yield_thresh)
)

independent_values = independent_assembly_yield[valid_for_analysis]
opercular_values = np.array(opercular_list)[valid_for_analysis]

# Get unique opercular values and sort them
unique_opercular = sorted(np.unique(opercular_values))
print(f"Opercular conditions present in data: {unique_opercular}")

# Group data by opercular condition
opercular_data = []
means = []
stds = []
counts = []

for opercular in unique_opercular:
    opercular_mask = opercular_values == opercular
    opercular_subset = independent_values[opercular_mask]
    opercular_data.append(opercular_subset)
    means.append(np.mean(opercular_subset))
    stds.append(np.std(opercular_subset))
    counts.append(len(opercular_subset))

# --- One-way ANOVA across all opercular groups ---
if all(len(vals) > 0 for vals in opercular_data) and len(opercular_data) > 1:
    anova_stat, anova_p = f_oneway(*opercular_data)
    print(f"\nOne-way ANOVA across opercular groups: F = {anova_stat:.3f}, p = {anova_p:.4g}")
else:
    anova_stat, anova_p = np.nan, np.nan
    print("\nOne-way ANOVA could not be performed (not enough data in all groups).")

# Create bar plot
bar_positions = np.arange(len(unique_opercular))
bar_width = 0.6

# Plot bars with error bars
bars = plt.bar(bar_positions, means, yerr=stds, capsize=7, 
               color='lightcoral', edgecolor='k', width=bar_width, 
               alpha=0.8, label='Mean ± SD')

# Add individual data points with jitter
for i, (opercular, vals) in enumerate(zip(unique_opercular, opercular_data)):
    # Add small random jitter to x-coordinates to avoid overlap
    x_jitter = bar_positions[i] + np.random.uniform(-0.15, 0.15, size=len(vals))
    plt.scatter(x_jitter, vals, color='k', alpha=0.7, s=50, zorder=10)

# Customize the plot
plt.xticks(bar_positions, [f'Opercular {o}' for o in unique_opercular])
plt.xlabel('Opercular Condition', fontsize=14, fontweight='bold')
plt.ylabel('Independent Assembly Yield', fontsize=14, fontweight='bold')
plt.title(f'Independent Assembly Yield by Opercular Condition\n(yield > {yield_thresh})', 
          fontsize=16, fontweight='bold')

# Add sample size annotations on top of bars
for i, (bar, count) in enumerate(zip(bars, counts)):
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2., height + stds[i] + 0.01,
             f'n={count}', ha='center', va='bottom', fontweight='bold')

# Add statistical comparisons
comparisons = list(itertools.combinations(range(len(unique_opercular)), 2))
y_max = max([np.max(vals) if len(vals) > 0 else 0 for vals in opercular_data])
y_start = y_max + 0.2 * y_max if y_max > 0 else 1
h = 0.05 * y_max if y_max > 0 else 0.5

print(f"\nStatistical comparisons (Welch's t-test):")
for idx, (i, j) in enumerate(comparisons):
    vals1 = opercular_data[i]
    vals2 = opercular_data[j]
    if len(vals1) > 0 and len(vals2) > 0:
        stat, pval = ttest_ind(vals1, vals2, equal_var=False)
        print(f"  Opercular {unique_opercular[i]} vs Opercular {unique_opercular[j]}: t = {stat:.3f}, p = {pval:.4g}")
        
        # Add significance lines
        y = y_start + idx * h
        x1, x2 = bar_positions[i], bar_positions[j]
        plt.plot([x1, x1, x2, x2], [y, y+h/4, y+h/4, y], lw=1.5, c='k')
        
        # Add significance stars
        if pval < 0.001:
            stars = '***'
        elif pval < 0.01:
            stars = '**'
        elif pval < 0.05:
            stars = '*'
        else:
            stars = 'n.s.'
        plt.text((x1 + x2) / 2, y + h/6, stars, ha='center', va='bottom', 
                color='k', fontsize=12, fontweight='bold')

# Add ANOVA result annotation to the plot
if not np.isnan(anova_p):
    # Place ANOVA result above all bars
    y_annot = y_start + len(comparisons) * h + 0.05 * y_max
    anova_text = f"One-way ANOVA: F = {anova_stat:.2f}, p = {anova_p:.3g}"
    plt.text(np.mean(bar_positions), y_annot, anova_text, ha='center', va='bottom', 
             color='navy', fontsize=13, fontweight='bold', bbox=dict(facecolor='white', alpha=0.7, edgecolor='navy'))

# Add grid for better readability
plt.grid(True, alpha=0.3, axis='y')
ax = plt.gca()
ax.set_axisbelow(True)

# Set adjustable aspect ratio
ax.set_box_aspect(pbaspect[1]/pbaspect[0])  # height/width

# Adjust layout
plt.tight_layout()

# Save the figure as SVG
plt.savefig("independent_assembly_yield_by_opercular.svg", format="svg", dpi=300, bbox_inches='tight')

# Display the plot in notebook output
plt.show()

# Print summary statistics
print(f"\n" + "="*60)
print("SUMMARY STATISTICS")
print("="*60)
for i, opercular in enumerate(unique_opercular):
    vals = opercular_data[i]
    print(f"Opercular {opercular}: n={len(vals):2d} | mean={means[i]:.4f} | std={stds[i]:.4f} | range=[{np.min(vals):.4f}, {np.max(vals):.4f}]")
print("="*60)
```

```python
# now compare variances
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import levene, bartlett, f_oneway
import itertools

plt.rcParams['svg.fonttype'] = 'none'  # Ensures text is editable in Illustrator

# --- Adjustable aspect ratio parameter ---
pbaspect = (0.5, 1)  # (width, height) ratio; change as desired, e.g., (2,1), (1,1), etc.

# Filter data for valid entries and apply yield threshold
yield_thresh = 1
valid_for_analysis = (
    ~np.isnan(independent_assembly_yield) & 
    (np.array(yield_list) > yield_thresh)
)

independent_values = independent_assembly_yield[valid_for_analysis]
opercular_values = np.array(opercular_list)[valid_for_analysis]

# Group data by FLAIR status (opercular: 0 = FLAIR+, 1 = FLAIR-)
flair_pos_mask = opercular_values == 0  # FLAIR+
flair_neg_mask = opercular_values == 1  # FLAIR-

flair_pos_yield = independent_values[flair_pos_mask]
flair_neg_yield = independent_values[flair_neg_mask]

# Calculate statistics
flair_pos_mean = np.mean(flair_pos_yield)
flair_pos_std = np.std(flair_pos_yield)
flair_pos_var = np.var(flair_pos_yield, ddof=1)  # Sample variance
flair_pos_n = len(flair_pos_yield)

flair_neg_mean = np.mean(flair_neg_yield)
flair_neg_std = np.std(flair_neg_yield)
flair_neg_var = np.var(flair_neg_yield, ddof=1)  # Sample variance
flair_neg_n = len(flair_neg_yield)

# Perform variance tests
# Levene's test (robust to non-normality)
levene_stat, levene_pval = levene(flair_pos_yield, flair_neg_yield)

# Bartlett's test (assumes normality)
bartlett_stat, bartlett_pval = bartlett(flair_pos_yield, flair_neg_yield)

# F-test for variance ratio (classical test)
f_ratio = flair_pos_var / flair_neg_var if flair_neg_var > 0 else np.nan
from scipy.stats import f
f_stat = f_ratio
f_pval = 2 * min(f.cdf(f_ratio, flair_pos_n - 1, flair_neg_n - 1), 
                 1 - f.cdf(f_ratio, flair_pos_n - 1, flair_neg_n - 1))

# Print statistical test results
print("="*60)
print("VARIANCE COMPARISON: FLAIR+ vs FLAIR-")
print("="*60)
print(f"\nFLAIR+ (opercular=0): n={flair_pos_n:2d} | mean={flair_pos_mean:.4f} | std={flair_pos_std:.4f} | var={flair_pos_var:.4f}")
print(f"FLAIR- (opercular=1): n={flair_neg_n:2d} | mean={flair_neg_mean:.4f} | std={flair_neg_std:.4f} | var={flair_neg_var:.4f}")
print(f"\nVariance ratio (FLAIR+/FLAIR-): {f_ratio:.4f}")
print(f"\nStatistical Tests:")
print(f"  Levene's test: W = {levene_stat:.4f}, p = {levene_pval:.4g}")
print(f"  Bartlett's test: T = {bartlett_stat:.4f}, p = {bartlett_pval:.4g}")
print(f"  F-test (variance ratio): F = {f_stat:.4f}, p = {f_pval:.4g}")
print("="*60)

# Create visualization
fig = plt.figure(figsize=(10, 6))
ax = plt.gca()

# Create bar plot
bar_positions = np.array([0, 1])
bar_width = 0.6
means = [flair_pos_mean, flair_neg_mean]
stds = [flair_pos_std, flair_neg_std]
counts = [flair_pos_n, flair_neg_n]
labels = ['FLAIR+', 'FLAIR-']

bars = ax.bar(bar_positions, means, yerr=stds, capsize=7, 
              color=['lightcoral', 'lightblue'], edgecolor='k', width=bar_width, 
              alpha=0.8, label='Mean ± SD')

# Add individual data points with jitter
x_jitter_pos = np.random.uniform(-0.15, 0.15, size=len(flair_pos_yield))
x_jitter_neg = np.random.uniform(-0.15, 0.15, size=len(flair_neg_yield)) + 1
ax.scatter(x_jitter_pos, flair_pos_yield, color='k', alpha=0.7, s=50, zorder=10)
ax.scatter(x_jitter_neg, flair_neg_yield, color='k', alpha=0.7, s=50, zorder=10)

# Customize the plot
ax.set_xticks(bar_positions)
ax.set_xticklabels(labels)
ax.set_xlabel('FLAIR Status', fontsize=14, fontweight='bold')
ax.set_ylabel('Independent Assembly Yield', fontsize=14, fontweight='bold')
ax.set_title(f'Independent Assembly Yield by FLAIR Status\n(yield > {yield_thresh})', 
          fontsize=16, fontweight='bold')

# Add sample size annotations on top of bars
for i, (bar, count) in enumerate(zip(bars, counts)):
    height = bar.get_height()
    ax.text(bar.get_x() + bar.get_width()/2., height + stds[i] + 0.01,
             f'n={count}', ha='center', va='bottom', fontweight='bold')

# Add variance test result annotation
y_max = max([np.max(flair_pos_yield) if len(flair_pos_yield) > 0 else 0,
             np.max(flair_neg_yield) if len(flair_neg_yield) > 0 else 0])
y_start = y_max + 0.2 * y_max if y_max > 0 else 1
h = 0.05 * y_max if y_max > 0 else 0.5

# Add significance line and annotation for Levene's test
x1, x2 = bar_positions[0], bar_positions[1]
y = y_start
ax.plot([x1, x1, x2, x2], [y, y+h/4, y+h/4, y], lw=1.5, c='k')

# Add p-value annotation
if levene_pval < 0.001:
    stars = '***'
elif levene_pval < 0.01:
    stars = '**'
elif levene_pval < 0.05:
    stars = '*'
else:
    stars = 'n.s.'
ax.text((x1 + x2) / 2, y + h/6, f"Levene's test: p = {levene_pval:.3g} {stars}", 
        ha='center', va='bottom', color='k', fontsize=12, fontweight='bold')

# Add variance ratio annotation
ax.text(0.5, 1.05, f"Variance ratio (FLAIR+/FLAIR-): {f_ratio:.3f} | Levene's p = {levene_pval:.3g}", 
        ha='center', va='bottom', transform=ax.transAxes, fontsize=13, fontweight='bold')

# Add grid for better readability
ax.grid(True, alpha=0.3, axis='y')
ax.set_axisbelow(True)

# Set adjustable aspect ratio
ax.set_box_aspect(pbaspect[1]/pbaspect[0])  # height/width

# Adjust layout
plt.tight_layout()

# Save the figure as SVG
plt.savefig("independent_assembly_yield_variance_flair_comparison.svg", format="svg", dpi=300, bbox_inches='tight')

# Display the plot in notebook output
plt.show()

# Print summary statistics
print(f"\n" + "="*60)
print("SUMMARY STATISTICS")
print("="*60)
print(f"FLAIR+: n={flair_pos_n:2d} | mean={flair_pos_mean:.4f} | std={flair_pos_std:.4f} | var={flair_pos_var:.4f} | range=[{np.min(flair_pos_yield):.4f}, {np.max(flair_pos_yield):.4f}]")
print(f"FLAIR-: n={flair_neg_n:2d} | mean={flair_neg_mean:.4f} | std={flair_neg_std:.4f} | var={flair_neg_var:.4f} | range=[{np.min(flair_neg_yield):.4f}, {np.max(flair_neg_yield):.4f}]")
print("="*60)
```

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import ttest_ind, f_oneway, mannwhitneyu
import itertools

plt.rcParams['svg.fonttype'] = 'none'  # Ensures text is editable in Illustrator

# --- Adjustable aspect ratio parameter ---
pbaspect = (0.5, 1)  # (width, height) ratio; change as desired, e.g., (2,1), (1,1), etc.

# Filter data for valid entries and apply yield threshold
yield_thresh = 1
valid_for_analysis = (
    ~np.isnan(independent_assembly_yield) & 
    (np.array(yield_list) > yield_thresh)
)

independent_values = independent_assembly_yield[valid_for_analysis]
grade_values = np.array(grade_list)[valid_for_analysis]

# Get unique grades and sort them
unique_grades = sorted(np.unique(grade_values))
print(f"Grades present in data: {unique_grades}")

# Group data by grade
grade_data = []
means = []
stds = []
counts = []

for grade in unique_grades:
    grade_mask = grade_values == grade
    grade_subset = independent_values[grade_mask]
    grade_data.append(grade_subset)
    means.append(np.mean(grade_subset))
    stds.append(np.std(grade_subset))
    counts.append(len(grade_subset))

# --- ADDITIONAL PLOT: Only opercular = 0 or 2 ---
# Group grade 2 and 3 together as "2/3"

# Filter for opercular = 0 or 2
opercular_arr = np.array(opercular_list)[valid_for_analysis]
mask_opercular_0_2 = np.isin(opercular_arr, [0, 2])

independent_values_02 = independent_values[mask_opercular_0_2]
grade_values_02 = grade_values[mask_opercular_0_2]
opercular_values_02 = opercular_arr[mask_opercular_0_2]

# Map grades: group 2 and 3 together as 2
grade_values_02_grouped = np.copy(grade_values_02)
grade_values_02_grouped[(grade_values_02 == 3)] = 2

# Get unique grades present in this subset (after grouping)
unique_grades_02 = sorted(np.unique(grade_values_02_grouped))
# For display, use '2/3' instead of 2 if both 2 and 3 were present in original data
display_grades_02 = []
if 2 in unique_grades_02:
    # Check if both 2 and 3 were present in original data
    if (2 in grade_values_02) and (3 in grade_values_02):
        display_grades_02 = [g if g != 2 else '2/3' for g in unique_grades_02]
    else:
        display_grades_02 = [str(g) for g in unique_grades_02]
else:
    display_grades_02 = [str(g) for g in unique_grades_02]

print(f"\nGrades present in data (opercular=0 or 2, grouped 2/3): {display_grades_02}")

# Group data by grade for opercular=0 or 2 (with 2/3 grouped)
grade_data_02 = []
means_02 = []
stds_02 = []
counts_02 = []

for grade in unique_grades_02:
    grade_mask = grade_values_02_grouped == grade
    grade_subset = independent_values_02[grade_mask]
    grade_data_02.append(grade_subset)
    means_02.append(np.mean(grade_subset))
    stds_02.append(np.std(grade_subset))
    counts_02.append(len(grade_subset))

# --- Overall significance: One-way ANOVA across groups ---
fig2 = plt.figure(figsize=(10, 6))
ax2 = plt.gca()

if len(grade_data_02) > 1 and all(len(vals) > 0 for vals in grade_data_02):
    anova_stat_02, anova_pval_02 = f_oneway(*grade_data_02)
    print(f"\nOne-way ANOVA across grades (opercular=0 or 2, grouped 2/3): F = {anova_stat_02:.3f}, p = {anova_pval_02:.4g}")
    ax2.text(0.5, 1.05, f"One-way ANOVA: p = {anova_pval_02:.3g}", 
            ha='center', va='bottom', transform=ax2.transAxes, fontsize=13, fontweight='bold')
else:
    print("\nNot enough groups or data for ANOVA (opercular=0 or 2, grouped 2/3).")

# Create bar plot
bar_positions_02 = np.arange(len(unique_grades_02))
bar_width_02 = 0.6

bars_02 = ax2.bar(bar_positions_02, means_02, yerr=stds_02, capsize=7, 
                  color='lightgreen', edgecolor='k', width=bar_width_02, 
                  alpha=0.8, label='Mean ± SD')

# Add individual data points with jitter
for i, (grade, vals) in enumerate(zip(unique_grades_02, grade_data_02)):
    x_jitter = bar_positions_02[i] + np.random.uniform(-0.15, 0.15, size=len(vals))
    ax2.scatter(x_jitter, vals, color='k', alpha=0.7, s=50, zorder=10)

# Customize the plot
ax2.set_xticks(bar_positions_02)
ax2.set_xticklabels([f'Grade {g}' for g in display_grades_02])
ax2.set_xlabel('Grade', fontsize=14, fontweight='bold')
ax2.set_ylabel('Independent Assembly Yield', fontsize=14, fontweight='bold')
ax2.set_title(f'Independent Assembly Yield by Grade (opercular=0 or 2, 2/3 grouped)\n(yield > {yield_thresh})', 
          fontsize=16, fontweight='bold')

# Add sample size annotations on top of bars
for i, (bar, count) in enumerate(zip(bars_02, counts_02)):
    height = bar.get_height()
    ax2.text(bar.get_x() + bar.get_width()/2., height + stds_02[i] + 0.01,
             f'n={count}', ha='center', va='bottom', fontweight='bold')

# Add statistical comparisons
comparisons_02 = list(itertools.combinations(range(len(unique_grades_02)), 2))
y_max_02 = max([np.max(vals) if len(vals) > 0 else 0 for vals in grade_data_02])
y_start_02 = y_max_02 + 0.2 * y_max_02 if y_max_02 > 0 else 1
h_02 = 0.05 * y_max_02 if y_max_02 > 0 else 0.5

print(f"\nStatistical comparisons (Welch's t-test, one-sided t-tests, & Mann-Whitney U, opercular=0 or 2, 2/3 grouped):")
for idx, (i, j) in enumerate(comparisons_02):
    vals1 = grade_data_02[i]
    vals2 = grade_data_02[j]
    if len(vals1) > 0 and len(vals2) > 0:
        label1 = display_grades_02[i]
        label2 = display_grades_02[j]

        # Welch's t-test (two-sided)
        stat_t, pval_t = ttest_ind(vals1, vals2, equal_var=False)
        print(f"  Grade {label1} vs Grade {label2}: t = {stat_t:.3f}, p = {pval_t:.4g} (Welch's t-test, two-sided)")

        # Welch's t-test (one-sided: vals1 > vals2)
        try:
            stat_t_greater, pval_t_greater = ttest_ind(vals1, vals2, equal_var=False, alternative='greater')
            print(f"  Grade {label1} vs Grade {label2} (one-sided, {label1}>{label2}): p = {pval_t_greater:.4g} (t-test, one-sided, greater)")
        except TypeError:
            # For older scipy: fallback to manual one-sided adjustment
            if stat_t > 0:
                pval_t_greater = pval_t / 2
            else:
                pval_t_greater = 1 - pval_t / 2
            print(f"  Grade {label1} vs Grade {label2} (one-sided, {label1}>{label2}): p = {pval_t_greater:.4g} (manual approx)")

        # Welch's t-test (one-sided: vals1 < vals2)
        try:
            stat_t_less, pval_t_less = ttest_ind(vals1, vals2, equal_var=False, alternative='less')
            print(f"  Grade {label1} vs Grade {label2} (one-sided, {label1}<{label2}): p = {pval_t_less:.4g} (t-test, one-sided, less)")
        except TypeError:
            if stat_t < 0:
                pval_t_less = pval_t / 2
            else:
                pval_t_less = 1 - pval_t / 2
            print(f"  Grade {label1} vs Grade {label2} (one-sided, {label1}<{label2}): p = {pval_t_less:.4g} (manual approx)")

        # Mann-Whitney U test
        try:
            stat_u, pval_u = mannwhitneyu(vals1, vals2, alternative='two-sided')
            print(f"  Grade {label1} vs Grade {label2}: U = {stat_u:.3f}, p = {pval_u:.4g} (Mann-Whitney U)")
        except Exception as e:
            print(f"  Mann-Whitney U test failed for Grade {label1} vs Grade {label2}: {e}")
        
        # Add significance lines (for two-sided Welch's t-test only, keep as in original)
        y = y_start_02 + idx * h_02
        x1, x2 = bar_positions_02[i], bar_positions_02[j]
        ax2.plot([x1, x1, x2, x2], [y, y+h_02/4, y+h_02/4, y], lw=1.5, c='k')
        
        # Add significance stars according to two-sided t-test p-value
        if pval_t < 0.001:
            stars = '***'
        elif pval_t < 0.01:
            stars = '**'
        elif pval_t < 0.05:
            stars = '*'
        else:
            stars = 'n.s.'
        ax2.text((x1 + x2) / 2, y + h_02/6, stars, ha='center', va='bottom', 
                color='k', fontsize=12, fontweight='bold')

# Add grid for better readability
ax2.grid(True, alpha=0.3, axis='y')
ax2.set_axisbelow(True)

# Set adjustable aspect ratio
ax2.set_box_aspect(pbaspect[1]/pbaspect[0])  # height/width

# Adjust layout
plt.tight_layout()

# Save the figure as SVG
plt.savefig("independent_assembly_yield_by_grade_opercular_0_2_grouped_2_3.svg", format="svg", dpi=300, bbox_inches='tight')

# Display the plot in notebook output
plt.show()

# Print summary statistics
print(f"\n" + "="*60)
print("SUMMARY STATISTICS (opercular=0 or 2, 2/3 grouped)")
print("="*60)
for i, grade in enumerate(unique_grades_02):
    vals = grade_data_02[i]
    label = display_grades_02[i]
    print(f"Grade {label}: n={len(vals):2d} | mean={means_02[i]:.4f} | std={stds_02[i]:.4f} | range=[{np.min(vals):.4f}, {np.max(vals):.4f}]")
print("="*60)
print("="*60)
print("="*60)
```

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import ttest_ind, f_oneway
import itertools

plt.rcParams['svg.fonttype'] = 'none'  # Ensures text is editable in Illustrator

# --- Adjustable aspect ratio parameter ---
pbaspect = (0.5, 1)  # (width, height) ratio; change as desired, e.g., (2,1), (1,1), etc.

# Filter data for valid entries and apply yield threshold
yield_thresh = 1
valid_for_analysis = (
    ~np.isnan(independent_assembly_yield) & 
    (np.array(yield_list) > yield_thresh)
)

independent_values = independent_assembly_yield[valid_for_analysis]
path_values = np.array(path_list)[valid_for_analysis]

# Get unique paths and sort them
unique_paths = sorted(np.unique(path_values))
print(f"Paths present in data: {unique_paths}")

# Group data by path
path_data = []
means = []
stds = []
counts = []

for path in unique_paths:
    path_mask = path_values == path
    path_subset = independent_values[path_mask]
    path_data.append(path_subset)
    means.append(np.mean(path_subset))
    stds.append(np.std(path_subset))
    counts.append(len(path_subset))

# --- ADDITIONAL PLOT: opercular = 0 or 2 only ---

# Filter for opercular = 0 or 2
opercular_arr = np.array(opercular_list)[valid_for_analysis]
opercular_mask_02 = (opercular_arr == 0) | (opercular_arr == 2)

independent_values_02 = independent_values[opercular_mask_02]
path_values_02 = path_values[opercular_mask_02]

unique_paths_02 = sorted(np.unique(path_values_02))
print(f"\nPaths present in data (opercular=0 or 2): {unique_paths_02}")

# --- ADDITIONAL PLOT: opercular = 0 or 2, combine ast+oli as idh-mut (all grades), gbm as idh-wt, vasc as its own group ---

from scipy.stats import ttest_ind, kruskal

# Get the corresponding path values for opercular=0 or 2
path_values_02 = np.array(path_values_02)
independent_values_02 = np.array(independent_values_02)

# If not already available, get the corresponding grade values for opercular=0 or 2
# Assumes grade_values_02 is aligned with path_values_02 and independent_values_02
grade_values_02 = np.array(grade_values_02)  # This must be defined in the notebook context

# Define group mapping, now including vasc, and for IDH-mut include all grades
def path_to_group_with_grade(path, grade):
    p = str(path).lower()
    if p.startswith('gbm'):
        return 'IDH-wt'
    elif (p.startswith('ast') or p.startswith('oli')):
        return 'IDH-mut'
    elif p.startswith('vas'):
        return 'Vasc'
    else:
        return None

group_labels = np.array([path_to_group_with_grade(p, g) for p, g in zip(path_values_02, grade_values_02)])
valid_group_mask = np.isin(group_labels, ['IDH-mut', 'IDH-wt', 'Vasc'])

# Only keep entries with a valid group
grouped_values = independent_values_02[valid_group_mask]
grouped_labels = group_labels[valid_group_mask]

unique_groups = ['IDH-mut', 'IDH-wt', 'Vasc']
group_data = []
group_means = []
group_stds = []
group_counts = []

for group in unique_groups:
    vals = grouped_values[grouped_labels == group]
    group_data.append(vals)
    group_means.append(np.mean(vals) if len(vals) > 0 else np.nan)
    group_stds.append(np.std(vals) if len(vals) > 0 else np.nan)
    group_counts.append(len(vals))

# --- Overall significance: Kruskal-Wallis and pairwise t-tests (Welch's) between groups ---
if sum(len(vals) > 0 for vals in group_data) >= 2:
    # Kruskal-Wallis test
    nonempty_data = [vals for vals in group_data if len(vals) > 0]
    kw_stat, kw_pval = kruskal(*nonempty_data)
    print(f"\nKruskal-Wallis test across groups (opercular=0 or 2, IDH-mut: all grades): H = {kw_stat:.3f}, p = {kw_pval:.4g}")
else:
    kw_stat, kw_pval = np.nan, np.nan
    print("\nNot enough data for Kruskal-Wallis test (opercular=0 or 2, IDH-mut: all grades).")

# Pairwise Welch's t-tests (one-sided: test if group[i] > group[j])
pairwise_stats = {}
for i in range(len(unique_groups)):
    for j in range(i+1, len(unique_groups)):
        vals1 = group_data[i]
        vals2 = group_data[j]
        if len(vals1) > 0 and len(vals2) > 0:
            # Two-sided t-test
            stat, pval_two_sided = ttest_ind(vals1, vals2, equal_var=False)
            # For one-sided, decide direction: test if mean(group i) > mean(group j)
            if stat > 0:
                pval = pval_two_sided / 2
                direction = ">"
            else:
                pval = 1 - pval_two_sided/2
                direction = "<="
            pairwise_stats[(i, j)] = (stat, pval)
            print(f"{unique_groups[i]} vs {unique_groups[j]} (Welch's one-sided t-test for mean({unique_groups[i]}) > mean({unique_groups[j]})): t = {stat:.3f}, p = {pval:.4g} ({direction})")
        else:
            pairwise_stats[(i, j)] = (np.nan, np.nan)
            print(f"{unique_groups[i]} vs {unique_groups[j]}: Not enough data for t-test.")

# --- Plotting ---
fig3 = plt.figure(figsize=(7, 6))
ax3 = plt.gca()

bar_positions_group = np.arange(len(unique_groups))
bar_width_group = 0.6

bars_group = ax3.bar(bar_positions_group, group_means, yerr=group_stds, capsize=7, 
                     color=['#4daf4a', '#e41a1c', '#377eb8'], edgecolor='k', width=bar_width_group, 
                     alpha=0.8, label='Mean ± SD')

# Add individual data points with jitter
for i, vals in enumerate(group_data):
    x_jitter = bar_positions_group[i] + np.random.uniform(-0.15, 0.15, size=len(vals))
    ax3.scatter(x_jitter, vals, color='k', alpha=0.7, s=50, zorder=10)

# Customize the plot
ax3.set_xticks(bar_positions_group)
ax3.set_xticklabels(unique_groups)
ax3.set_xlabel('Group', fontsize=14, fontweight='bold')
ax3.set_ylabel('Independent Assembly Yield', fontsize=14, fontweight='bold')
ax3.set_title(f'Independent Assembly Yield by Group (opercular=0 or 2, IDH-mut: all grades)\n(yield > {yield_thresh})', 
              fontsize=16, fontweight='bold')

# Add sample size annotations on top of bars
for i, (bar, count) in enumerate(zip(bars_group, group_counts)):
    height = bar.get_height()
    ax3.text(bar.get_x() + bar.get_width()/2., height + group_stds[i] + 0.01,
             f'n={count}', ha='center', va='bottom', fontweight='bold')

# Add Kruskal-Wallis result above the plot
if not np.isnan(kw_pval):
    y_annot_group = max([np.max(vals) if len(vals) > 0 else 0 for vals in group_data]) * 1.25
    if y_annot_group == 0:
        y_annot_group = 1
    kw_text = f"Kruskal-Wallis: H = {kw_stat:.2f}, p = {kw_pval:.3g}"
    ax3.text(0.5, 1.08, kw_text, ha='center', va='bottom', fontsize=13, fontweight='bold', 
            transform=ax3.transAxes)

# Add pairwise significance lines and stars
# We'll space the lines vertically above the bars
y_base = max([np.max(vals) if len(vals) > 0 else 0 for vals in group_data]) * 1.15
if y_base == 0:
    y_base = 1
h = 0.07 * y_base if y_base > 0 else 0.07
line_offsets = [0, 1, 2]  # for stacking lines

for idx, ((i, j), (stat, pval)) in enumerate(pairwise_stats.items()):
    if not np.isnan(pval):
        y = y_base + line_offsets[idx] * h
        x1, x2 = bar_positions_group[i], bar_positions_group[j]
        ax3.plot([x1, x1, x2, x2], [y, y+h/4, y+h/4, y], lw=1.5, c='k')
        # Add significance stars
        if pval < 0.001:
            stars = '***'
        elif pval < 0.01:
            stars = '**'
        elif pval < 0.05:
            stars = '*'
        else:
            stars = 'n.s.'
        ax3.text((x1 + x2) / 2, y + h/6, stars, ha='center', va='bottom', 
                color='k', fontsize=12, fontweight='bold')

# Add grid for better readability
ax3.grid(True, alpha=0.3, axis='y')
ax3.set_axisbelow(True)

# Set adjustable aspect ratio
ax3.set_box_aspect(pbaspect[1]/pbaspect[0])  # height/width

# Adjust layout
plt.tight_layout()

# Save the figure as SVG
plt.savefig("independent_assembly_yield_by_group_opercular_0_2_with_vasc_idhmut_allgrades.svg", format="svg", dpi=300, bbox_inches='tight')

# Display the plot in notebook output
plt.show()

# Print summary statistics
print(f"\n" + "="*60)
print("SUMMARY STATISTICS (opercular=0 or 2, grouped, including Vasc, IDH-mut: all grades)")
print("="*60)
for i, group in enumerate(unique_groups):
    vals = group_data[i]
    if len(vals) > 0:
        print(f"{group}: n={len(vals):2d} | mean={group_means[i]:.4f} | std={group_stds[i]:.4f} | range=[{np.min(vals):.4f}, {np.max(vals):.4f}]")
    else:
        print(f"{group}: n= 0 | mean=nan | std=nan | range=[nan, nan]")
print("="*60)
print("="*60)
print("="*60)
```
