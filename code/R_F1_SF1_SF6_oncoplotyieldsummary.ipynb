```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import kruskal, mannwhitneyu
import pandas as pd
from scipy import stats
import statsmodels.api as sm
plt.rcParams['svg.fonttype'] = 'none'

grade_unique = sorted(set(grade_list))

# Organize yields and opercular statuses by grade
yields_by_grade = {g: [] for g in grade_unique}
operculars_by_grade = {g: [] for g in grade_unique}
for g, y, op in zip(grade_list, yield_list, opercular_list):
    yields_by_grade[g].append(y)
    operculars_by_grade[g].append(op)

# Calculate means and errors by grade (all operculars together)
mean_yields_grade = []
stderr_yields_grade = []
for g in grade_unique:
    ys = yields_by_grade[g]
    mean_yields_grade.append(np.mean(ys) if len(ys) > 0 else 0)
    stderr_yields_grade.append(np.std(ys, ddof=1)/np.sqrt(len(ys)) if len(ys) > 1 else 0)

# --- Summary table for average yield by grade ---
summary_stats = []
for g, mean, err, ys in zip(grade_unique, mean_yields_grade, stderr_yields_grade, [yields_by_grade[g] for g in grade_unique]):
    summary_stats.append({
        'Grade': g,
        'N': len(ys),
        'Mean_Yield': mean,
        'SEM': err,
        'Median': np.median(ys) if len(ys) else 0,
        'Min': np.min(ys) if len(ys) else 0,
        'Max': np.max(ys) if len(ys) else 0
    })
summary_df = pd.DataFrame(summary_stats)
print("Summary stats of average yield by grade:")
print(summary_df.to_string(index=False))
print()

# ---- One-way ANOVA across grade for yield ----
anova_df = pd.DataFrame({
    "yield_": yield_list,   # Use a column name that is NOT a Python keyword!
    "grade": grade_list
})

# Statsmodels requires a categorical type for correct ANOVA formatting:
anova_df["grade_cat"] = anova_df["grade"].astype(str)

# Fit OLS model and run ANOVA
import statsmodels.formula.api as smf
model = smf.ols('yield_ ~ C(grade_cat)', data=anova_df).fit()
anova_table = sm.stats.anova_lm(model, typ=2)
print("One-way ANOVA: effect of grade on yield")
print(anova_table.to_string(float_format="%.4g"))
print()

# --- Pairwise statistics table ---
print("Pairwise Mann-Whitney U tests (all yields, ignoring opercular):")
pairwise_results = []
for i in range(len(grade_unique)):
    for j in range(i+1, len(grade_unique)):
        g1, g2 = grade_unique[i], grade_unique[j]
        ys1, ys2 = yields_by_grade[g1], yields_by_grade[g2]
        if len(ys1) > 0 and len(ys2) > 0:
            ustat, pval = mannwhitneyu(ys1, ys2, alternative='two-sided')
            pairwise_results.append({"Grade 1": g1, "Grade 2": g2, "U": ustat, "p-value": pval})
pairwise_df = pd.DataFrame(pairwise_results)
if not pairwise_df.empty:
    pd.set_option('display.precision', 4)
    print(pairwise_df.to_string(index=False))
    print()

fig, ax = plt.subplots(figsize=(6, 4))

bar_width = 0.8

# Only show a single bar per grade (not separated by opercular)
bar_color = np.array([0.75, 0.75, 0.75])  # Neutral gray for the bar

bars = ax.bar(
    grade_unique, mean_yields_grade,
    bar_width,
    color=bar_color,
    yerr=stderr_yields_grade,
    capsize=5,
    label='Mean yield (all)',
    edgecolor='k',
    zorder=1
)

# Colors for opercular=0 and opercular=1
color_op0 = np.array([0.12, 0.47, 0.71])  # blue-ish
color_op1 = np.array([1, 0.55, 0])        # orange-ish

# Overlay each individual point, colored by opercular status
for i, g in enumerate(grade_unique):
    ys = yields_by_grade[g]
    ops = operculars_by_grade[g]
    # jitter for x
    xvals = np.random.normal(g, 0.11, size=len(ys))
    # plot opercular=0
    mask_op0 = np.array(ops) == 0
    mask_op1 = np.array(ops) == 1
    if np.any(mask_op0):
        ax.scatter(xvals[mask_op0], np.array(ys)[mask_op0], color=color_op0, alpha=0.7, s=45, zorder=10, marker='o', edgecolor='w', label=None)
    if np.any(mask_op1):
        ax.scatter(xvals[mask_op1], np.array(ys)[mask_op1], color=color_op1, alpha=0.7, s=45, zorder=10, marker='o', edgecolor='w', label=None)

# Overlay the mean and errorbar as red horizontal lines/dots
for i, g in enumerate(grade_unique):
    mean = mean_yields_grade[i]
    err = stderr_yields_grade[i]
    ax.errorbar(g, mean, yerr=err, fmt='o', color='red', capsize=6, markersize=7, zorder=15,
                label=None if i != 0 else 'Mean ± SEM')

ax.set_xlabel('Grade')
ax.set_ylabel('Yield')
ax.set_title('Yield by Grade')
ax.set_xticks(grade_unique)

# Set panel aspect
ax.set_aspect(1/80)

# Build a legend for bar, data points, and mean
from matplotlib.lines import Line2D
handles = [
    bars,
    Line2D([], [], color=color_op0, marker='o', lw=0, markersize=9, markeredgecolor='w', label='Individual (opercular=0)', alpha=0.7),
    Line2D([], [], color=color_op1, marker='o', lw=0, markersize=9, markeredgecolor='w', label='Individual (opercular=1)', alpha=0.7),
    Line2D([], [], color='red', marker='o', lw=0, markersize=9, label='Mean ± SEM'),
]
labels = [
    'Mean yield (all)',
    'Individual (opercular=0)',
    'Individual (opercular=1)',
    'Mean ± SEM',
]
ax.legend(handles, labels)

# -- Annotate ANOVA p-value on plot
anova_p = anova_table.loc['C(grade_cat)', 'PR(>F)'] if 'C(grade_cat)' in anova_table.index else np.nan
ax.text(0.98, 0.98, f"ANOVA\n$p={anova_p:.3g}$",
        transform=ax.transAxes, va='top', ha='right', fontsize=11,
        bbox=dict(boxstyle='round', facecolor='white', alpha=0.7))

# -- Pairwise stats (Mann-Whitney U) between pairs of grades (using all yields, ignoring opercular)
# Annotate on bar plot only for consecutive pairs, but print all pairwise stats in console above
for i in range(len(grade_unique)-1):
    g1 = grade_unique[i]
    g2 = grade_unique[i+1]
    ys1 = yields_by_grade[g1]
    ys2 = yields_by_grade[g2]
    if len(ys1) > 0 and len(ys2) > 0:
        ustat, pval = mannwhitneyu(ys1, ys2, alternative='two-sided')
        y_max = max(
            max(ys1) if len(ys1) else 0,
            max(ys2) if len(ys2) else 0,
            mean_yields_grade[i]+stderr_yields_grade[i],
            mean_yields_grade[i+1]+stderr_yields_grade[i+1]
        )
        y_ann = y_max + 0.03 * (max(mean_yields_grade) if mean_yields_grade else 1)
        # Only annotate if significant
        if pval < 0.05:
            ax.plot([g1, g2], [y_ann, y_ann], 'k-', lw=1)
            ax.text((g1 + g2)/2, y_ann, "*" if pval < 0.05 else "n.s.", ha='center', va='bottom', fontsize=14)

plt.tight_layout()
plt.savefig('fig1yield_grade.svg', format='svg', dpi=300, bbox_inches='tight')
plt.show()
```

```python
# LOOP FOR FRs

import fnmatch
import matplotlib as mpl
import matplotlib.pyplot as plt
import numpy as np
from scipy import stats
from sklearn.decomposition import PCA
from sklearn.decomposition import FastICA
from scipy.linalg import norm
from scipy.stats import kstest
from scipy.stats import ttest_rel
import re

def imec_key_sorter(key):
    # Extract the number after 'imec', default to a large number if not found
    match = re.search(r'imec(\d+)', key)
    return int(match.group(1)) if match else float('inf')

# ---- Behavior-duration-aware filtering (mirrors R_AIproduction.ipynb) ----
# When behavioral epochs are too short to give a stable rate / KS window,
# fall back to the full recording instead of the TaskTimes-restricted view.
MIN_BEHAVIOR_MINUTES = 0

def _behavior_duration_seconds(task_times_obj):
    starts = np.asarray(task_times_obj.start, dtype=float).ravel()
    ends   = np.asarray(task_times_obj.end,   dtype=float).ravel()
    if starts.size != ends.size:
        return np.nan
    return float(np.sum(ends - starts))

def _ks_time_bounds_full_recording(spike_times_obj):
    """Min / max spike time across all units (KS uniform null on the full recording)."""
    tmin, tmax = np.inf, -np.inf
    for u in range(len(spike_times_obj)):
        idx = spike_times_obj[u].as_series().index.values
        if len(idx):
            tmin = min(tmin, float(np.min(idx)))
            tmax = max(tmax, float(np.max(idx)))
    if np.isfinite(tmin) and np.isfinite(tmax):
        return tmin, tmax
    return np.nan, np.nan

# Initialize lists to store neuron-level data
firing_rate_all_list = []  # Average firing rate across entire recording
firing_rate_beh_list = []  # Average firing rate during behavioral epochs
path_list_neuron = []      # Pathology for each neuron
grade_list_neuron = []     # Grade for each neuron
flair_list_neuron = []     # FLAIR status for each neuron (0 or 1)

insertion = 0

# loop through all nwb files
for i in range(len(nwb_paths)):
    data = nap.load_file(nwb_paths[i])
    keys = data.keys()

    # find the spike TsGroup objects, number various based on recording apparatus
    template = "*imec*"
    keys = [key for key in keys if fnmatch.fnmatch(key, template)]
    # Preferentially select keys containing 'KS4'; if none, use all keys
    ks_keys = [key for key in keys if "KS4" in key]
    if ks_keys:
        keys = ks_keys
    # Preferentially select keys containing 'Th='; if none, use all keys
    th8_keys = [key for key in keys if "Th=8" in key]
    if th8_keys:
        keys = th8_keys
    else:
        th_keys = [key for key in keys if "Th=" in key]
        if th_keys:
            keys = th_keys
    # Drop sentgen and *_auto* shanks (match R_AIproduction)
    template_sentgen = "*sentgen*"
    template_auto    = "*_auto*"
    keys = [key for key in keys if not fnmatch.fnmatch(key, template_sentgen) and not fnmatch.fnmatch(key, template_auto)]
    # For NP137, remove 'imec1'
    if 'NP137' in str(nwb_paths[i]) or 'NP139_B2' in str(nwb_paths[i]):
        keys = [key for key in keys if 'imec1' not in key]
    # Sort keys so that imec0 is first, then imec1, then imec2, etc.
    keys = sorted(keys, key=imec_key_sorter)

    for s in range(len(keys)):
        print(keys[s])
        print(region_list[insertion])
        spike_times = data[keys[s]]
        
        # 1. filter to good neurons
        firingRates_all = spike_times.metadata["rate"]

        # task_times key may be capitalized differently across sessions
        if "TaskTimes" in data.keys():
            task_times = data["TaskTimes"]
        else:
            task_times = data["task_times"]
        start_time = task_times.start
        end_time   = task_times.end
        beh_epochs = nap.IntervalSet(start=task_times.start, end=task_times.end)

        # Decide whether behavior is long enough; if not, use full recording for filters/KS.
        beh_sec = _behavior_duration_seconds(task_times)
        beh_min = beh_sec / 60.0 if np.isfinite(beh_sec) else np.nan
        use_full_recording = (not np.isfinite(beh_min)) or (beh_min < MIN_BEHAVIOR_MINUTES)

        if use_full_recording:
            spike_times_beh = spike_times
            firingRates_beh = firingRates_all
            min_time, max_time = _ks_time_bounds_full_recording(spike_times)
            if not (np.isfinite(min_time) and np.isfinite(max_time)) or (max_time <= min_time):
                print(f"  Short behavior ({beh_min:.2f} min) but could not get recording bounds; skipping shank")
                insertion += 1
                continue
            print(f"  Behavioral duration {beh_min:.2f} min (< {MIN_BEHAVIOR_MINUTES} min) \u2014 using full recording for filters; KS window [{min_time:.3f}, {max_time:.3f}] s")
        else:
            spike_times_beh = spike_times.restrict(beh_epochs)
            firingRates_beh = spike_times_beh.metadata["rate"]
            min_time = float(np.asarray(start_time).ravel()[0])
            max_time = float(np.asarray(end_time).ravel()[-1])
            print(f"  Behavioral duration {beh_min:.2f} min (>= {MIN_BEHAVIOR_MINUTES} min) \u2014 using TaskTimes restrict; KS window [{min_time:.3f}, {max_time:.3f}] s")

        # KS test against uniform on the chosen [min_time, max_time] window
        ks_stats = np.zeros(len(spike_times))
        ks_pvals = np.zeros(len(spike_times))
        for u in range(len(spike_times)):
            test = spike_times[u].as_series().index.values
            if len(test) > 1:
                normalized_spike_times = (test - min_time) / (max_time - min_time)
                ks_result = kstest(normalized_spike_times, 'uniform')
                ks_stats[u] = ks_result.statistic
                ks_pvals[u] = ks_result.pvalue
            else:
                ks_stats[u] = np.nan
                ks_pvals[u] = np.nan

        # now do ISI calculation
        violationThreshold = 3/1000 #for 3 ms isi refractory period, what used before, middle of range that kilosort cites
        violationPct = np.zeros(len(spike_times))
        for u in range(len(spike_times)):
            unit = spike_times[u]
            unit = unit.as_series().index
            if len(unit) < 100:
                violationPct[u] = 1 
            else:
                isi = unit.diff()[1:len(unit)]
                violations = np.where(isi<violationThreshold)
                violations = np.array(violations)
                violationPct[u] = violations.size/len(isi)

        # now apply masks
        if "KSLabel" in spike_times.metadata:
            KSLabels = spike_times.metadata["KSLabel"]
        else:
            KSLabels = spike_times.metadata["quality"]
        firingRates = firingRates_beh
        mask1 = violationPct<3/100 # throwing out neurons with > 3% isi violations with isi = 3 ms; lose quite a bit with 5 ms esp when including cdf stats
        mask2 = firingRates>0.5 # firing rate > 0.5 ms during all behavior epochs
        mask3 = KSLabels!="noise" # not noise per auto label
        mask4 = ks_stats < 0.3 # KS stat < 0.3 (stability across recording)
        mask = mask1 & mask2 & mask3 & mask4
        indicesFinal = firingRates.index[mask]

        # now exclude neurons manually rejected
        indicesFinal = np.setdiff1d(indicesFinal, manual_exclude_lists[insertion])
        
        # Get path, grade, and opercular for this insertion (before incrementing insertion)
        current_path = path_list[insertion]
        current_grade = grade_list[insertion]
        current_opercular = opercular_list[insertion]
        
        # Calculate FLAIR: if opercular = 1, then FLAIR = 0; if opercular = 0, then FLAIR = 1
        current_flair = 1 - current_opercular
        
        insertion += 1
        spike_times_good = spike_times_beh[indicesFinal]
        print(f"Subject: {i}, insertion: {s}, Number of good neurons: {len(spike_times_good)}")

        # Store firing rates and metadata for each neuron
        for neuron_idx in indicesFinal:
            # A) Average firing rate across entire recording
            firing_rate_all = firingRates_all[neuron_idx]
            firing_rate_all_list.append(firing_rate_all)
            
            # B) Average firing rate during behavioral epochs
            firing_rate_beh = firingRates_beh[neuron_idx]
            firing_rate_beh_list.append(firing_rate_beh)
            
            # C) Path (pathology)
            path_list_neuron.append(current_path)
            
            # D) Grade
            grade_list_neuron.append(current_grade)
            
            # E) FLAIR status
            flair_list_neuron.append(current_flair)

# Convert to numpy arrays
firing_rate_all_array = np.array(firing_rate_all_list)
firing_rate_beh_array = np.array(firing_rate_beh_list)
path_array = np.array(path_list_neuron)
grade_array = np.array(grade_list_neuron)
flair_array = np.array(flair_list_neuron)

print(f"\n{'='*80}")
print("DATA COLLECTION COMPLETE")
print(f"{'='*80}")
print(f"Total number of neurons: {len(firing_rate_all_list)}")
print(f"Firing rate (all): mean = {np.mean(firing_rate_all_array):.3f}, std = {np.std(firing_rate_all_array):.3f}")
print(f"Firing rate (beh): mean = {np.mean(firing_rate_beh_array):.3f}, std = {np.std(firing_rate_beh_array):.3f}")
print(f"Pathology distribution: {dict(zip(*np.unique(path_array, return_counts=True)))}")
print(f"Grade distribution: {dict(zip(*np.unique(grade_array, return_counts=True)))}")
print(f"FLAIR distribution: {dict(zip(*np.unique(flair_array, return_counts=True)))}")
```

```python
# COMPARE FIRING RATES BY IDH STATUS, GRADE, AND FLAIR
# Compare IDH-mutant (ast/oli) vs IDH-wildtype (gbm), Grade 2/3 vs Grade 4, and FLAIR = 1 vs FLAIR = 0
import matplotlib.pyplot as plt
import numpy as np
from scipy import stats
from scipy.stats import mannwhitneyu
import pandas as pd
import seaborn as sns

plt.rcParams['svg.fonttype'] = 'none'
sns.set_style("whitegrid")
sns.set_palette("husl")

print("="*80)
print("FIRING RATE COMPARISONS: IDH STATUS, GRADE, AND FLAIR")
print("="*80)

# Convert path to IDH status
idh_status_list = []
for path in path_array:
    if path in ['ast', 'oli']:
        idh_status_list.append('IDH-mut')
    elif path == 'gbm':
        idh_status_list.append('IDH-wt')
    else:
        idh_status_list.append('Unknown')

idh_status_array = np.array(idh_status_list)

# ============================================================================
# IDH STATUS COMPARISONS
# ============================================================================
print("\n" + "="*80)
print("IDH STATUS COMPARISONS")
print("="*80)

# IDH-mutant vs IDH-wildtype for "all" firing rate
idh_mut_all = firing_rate_all_array[idh_status_array == 'IDH-mut']
idh_wt_all = firing_rate_all_array[idh_status_array == 'IDH-wt']

print(f"\nFiring Rate (All Recording) - IDH Status:")
median_idh_mut_all = np.median(idh_mut_all)
q1_idh_mut_all = np.percentile(idh_mut_all, 25)
q3_idh_mut_all = np.percentile(idh_mut_all, 75)
iqr_idh_mut_all = q3_idh_mut_all - q1_idh_mut_all
median_idh_wt_all = np.median(idh_wt_all)
q1_idh_wt_all = np.percentile(idh_wt_all, 25)
q3_idh_wt_all = np.percentile(idh_wt_all, 75)
iqr_idh_wt_all = q3_idh_wt_all - q1_idh_wt_all
print(f"  IDH-mutant (ast+oli): n = {len(idh_mut_all)}, mean = {np.mean(idh_mut_all):.3f} ± {np.std(idh_mut_all):.3f}, median = {median_idh_mut_all:.3f}, IQR = {iqr_idh_mut_all:.3f}")
print(f"  IDH-wildtype (gbm): n = {len(idh_wt_all)}, mean = {np.mean(idh_wt_all):.3f} ± {np.std(idh_wt_all):.3f}, median = {median_idh_wt_all:.3f}, IQR = {iqr_idh_wt_all:.3f}")

# Statistical tests
t_stat_all, p_val_all = stats.ttest_ind(idh_mut_all, idh_wt_all)
mw_stat_all, mw_p_val_all = mannwhitneyu(idh_mut_all, idh_wt_all, alternative='two-sided')
print(f"  T-test: t = {t_stat_all:.3f}, p = {p_val_all:.6f}")
print(f"  Mann-Whitney U (two-sided): U = {mw_stat_all:.1f}, p = {mw_p_val_all:.6f}")
# One-sided test: IDH-wt > IDH-mut
mw_stat_all_onesided, mw_p_val_all_onesided = mannwhitneyu(idh_wt_all, idh_mut_all, alternative='greater')
print(f"  Mann-Whitney U (one-sided, IDH-wt > IDH-mut): U = {mw_stat_all_onesided:.1f}, p = {mw_p_val_all_onesided:.6f}")

# IDH-mutant vs IDH-wildtype for "beh" firing rate
idh_mut_beh = firing_rate_beh_array[idh_status_array == 'IDH-mut']
idh_wt_beh = firing_rate_beh_array[idh_status_array == 'IDH-wt']

print(f"\nFiring Rate (Behavioral Epochs) - IDH Status:")
median_idh_mut_beh = np.median(idh_mut_beh)
q1_idh_mut_beh = np.percentile(idh_mut_beh, 25)
q3_idh_mut_beh = np.percentile(idh_mut_beh, 75)
iqr_idh_mut_beh = q3_idh_mut_beh - q1_idh_mut_beh
median_idh_wt_beh = np.median(idh_wt_beh)
q1_idh_wt_beh = np.percentile(idh_wt_beh, 25)
q3_idh_wt_beh = np.percentile(idh_wt_beh, 75)
iqr_idh_wt_beh = q3_idh_wt_beh - q1_idh_wt_beh
print(f"  IDH-mutant (ast+oli): n = {len(idh_mut_beh)}, mean = {np.mean(idh_mut_beh):.3f} ± {np.std(idh_mut_beh):.3f}, median = {median_idh_mut_beh:.3f}, IQR = {iqr_idh_mut_beh:.3f}")
print(f"  IDH-wildtype (gbm): n = {len(idh_wt_beh)}, mean = {np.mean(idh_wt_beh):.3f} ± {np.std(idh_wt_beh):.3f}, median = {median_idh_wt_beh:.3f}, IQR = {iqr_idh_wt_beh:.3f}")

# Statistical tests
t_stat_beh, p_val_beh = stats.ttest_ind(idh_mut_beh, idh_wt_beh)
mw_stat_beh, mw_p_val_beh = mannwhitneyu(idh_mut_beh, idh_wt_beh, alternative='two-sided')
print(f"  T-test: t = {t_stat_beh:.3f}, p = {p_val_beh:.6f}")
print(f"  Mann-Whitney U (two-sided): U = {mw_stat_beh:.1f}, p = {mw_p_val_beh:.6f}")
# One-sided test: IDH-wt > IDH-mut
mw_stat_beh_onesided, mw_p_val_beh_onesided = mannwhitneyu(idh_wt_beh, idh_mut_beh, alternative='greater')
print(f"  Mann-Whitney U (one-sided, IDH-wt > IDH-mut): U = {mw_stat_beh_onesided:.1f}, p = {mw_p_val_beh_onesided:.6f}")

# ============================================================================
# GRADE COMPARISONS (Grade 2/3 vs Grade 4)
# ============================================================================
print("\n" + "="*80)
print("GRADE COMPARISONS (Grade 2/3 vs Grade 4)")
print("="*80)

# Combine Grade 2 and 3 for "all" recording
grade_23_all = firing_rate_all_array[(grade_array == 2) | (grade_array == 3)]
grade_4_all = firing_rate_all_array[grade_array == 4]

print(f"\nFiring Rate (All Recording) - Grade:")
median_grade_23_all = np.median(grade_23_all)
q1_grade_23_all = np.percentile(grade_23_all, 25)
q3_grade_23_all = np.percentile(grade_23_all, 75)
iqr_grade_23_all = q3_grade_23_all - q1_grade_23_all
median_grade_4_all = np.median(grade_4_all)
q1_grade_4_all = np.percentile(grade_4_all, 25)
q3_grade_4_all = np.percentile(grade_4_all, 75)
iqr_grade_4_all = q3_grade_4_all - q1_grade_4_all
print(f"  Grade 2/3: n = {len(grade_23_all)}, mean = {np.mean(grade_23_all):.3f} ± {np.std(grade_23_all):.3f}, median = {median_grade_23_all:.3f}, IQR = {iqr_grade_23_all:.3f}")
print(f"  Grade 4: n = {len(grade_4_all)}, mean = {np.mean(grade_4_all):.3f} ± {np.std(grade_4_all):.3f}, median = {median_grade_4_all:.3f}, IQR = {iqr_grade_4_all:.3f}")

# Statistical tests for "all" firing rate
t_stat_grade_all, p_val_grade_all = stats.ttest_ind(grade_23_all, grade_4_all)
mw_stat_grade_all, mw_p_val_grade_all = mannwhitneyu(grade_23_all, grade_4_all, alternative='two-sided')
print(f"  T-test: t = {t_stat_grade_all:.3f}, p = {p_val_grade_all:.6f}")
print(f"  Mann-Whitney U (two-sided): U = {mw_stat_grade_all:.1f}, p = {mw_p_val_grade_all:.6f}")
# One-sided test: Grade 4 > Grade 2/3
mw_stat_grade_all_onesided, mw_p_val_grade_all_onesided = mannwhitneyu(grade_4_all, grade_23_all, alternative='greater')
print(f"  Mann-Whitney U (one-sided, Grade 4 > Grade 2/3): U = {mw_stat_grade_all_onesided:.1f}, p = {mw_p_val_grade_all_onesided:.6f}")

# Combine Grade 2 and 3 for "beh" epochs
grade_23_beh = firing_rate_beh_array[(grade_array == 2) | (grade_array == 3)]
grade_4_beh = firing_rate_beh_array[grade_array == 4]

print(f"\nFiring Rate (Behavioral Epochs) - Grade:")
median_grade_23_beh = np.median(grade_23_beh)
q1_grade_23_beh = np.percentile(grade_23_beh, 25)
q3_grade_23_beh = np.percentile(grade_23_beh, 75)
iqr_grade_23_beh = q3_grade_23_beh - q1_grade_23_beh
median_grade_4_beh = np.median(grade_4_beh)
q1_grade_4_beh = np.percentile(grade_4_beh, 25)
q3_grade_4_beh = np.percentile(grade_4_beh, 75)
iqr_grade_4_beh = q3_grade_4_beh - q1_grade_4_beh
print(f"  Grade 2/3: n = {len(grade_23_beh)}, mean = {np.mean(grade_23_beh):.3f} ± {np.std(grade_23_beh):.3f}, median = {median_grade_23_beh:.3f}, IQR = {iqr_grade_23_beh:.3f}")
print(f"  Grade 4: n = {len(grade_4_beh)}, mean = {np.mean(grade_4_beh):.3f} ± {np.std(grade_4_beh):.3f}, median = {median_grade_4_beh:.3f}, IQR = {iqr_grade_4_beh:.3f}")

# Statistical tests for "beh" firing rate
t_stat_grade_beh, p_val_grade_beh = stats.ttest_ind(grade_23_beh, grade_4_beh)
mw_stat_grade_beh, mw_p_val_grade_beh = mannwhitneyu(grade_23_beh, grade_4_beh, alternative='two-sided')
print(f"  T-test: t = {t_stat_grade_beh:.3f}, p = {p_val_grade_beh:.6f}")
print(f"  Mann-Whitney U (two-sided): U = {mw_stat_grade_beh:.1f}, p = {mw_p_val_grade_beh:.6f}")
# One-sided test: Grade 4 > Grade 2/3
mw_stat_grade_beh_onesided, mw_p_val_grade_beh_onesided = mannwhitneyu(grade_4_beh, grade_23_beh, alternative='greater')
print(f"  Mann-Whitney U (one-sided, Grade 4 > Grade 2/3): U = {mw_stat_grade_beh_onesided:.1f}, p = {mw_p_val_grade_beh_onesided:.6f}")

# ============================================================================
# FLAIR COMPARISONS
# ============================================================================
print("\n" + "="*80)
print("FLAIR COMPARISONS")
print("="*80)

# FLAIR = 1 vs FLAIR = 0 for "all" recording
flair_1_all = firing_rate_all_array[flair_array == 1]
flair_0_all = firing_rate_all_array[flair_array == 0]

print(f"\nFiring Rate (All Recording) - FLAIR:")
median_1_all = np.median(flair_1_all)
q1_flair_1_all = np.percentile(flair_1_all, 25)
q3_flair_1_all = np.percentile(flair_1_all, 75)
iqr_flair_1_all = q3_flair_1_all - q1_flair_1_all
median_0_all = np.median(flair_0_all)
q1_flair_0_all = np.percentile(flair_0_all, 25)
q3_flair_0_all = np.percentile(flair_0_all, 75)
iqr_flair_0_all = q3_flair_0_all - q1_flair_0_all
print(f"  FLAIR = 1: n = {len(flair_1_all)}, mean = {np.mean(flair_1_all):.3f} ± {np.std(flair_1_all):.3f}, median = {median_1_all:.3f}, IQR = {iqr_flair_1_all:.3f}")
print(f"  FLAIR = 0: n = {len(flair_0_all)}, mean = {np.mean(flair_0_all):.3f} ± {np.std(flair_0_all):.3f}, median = {median_0_all:.3f}, IQR = {iqr_flair_0_all:.3f}")

# Statistical tests for "all" firing rate
t_stat_flair_all, p_val_flair_all = stats.ttest_ind(flair_1_all, flair_0_all)
mw_stat_flair_all, mw_p_val_flair_all = mannwhitneyu(flair_1_all, flair_0_all, alternative='two-sided')
print(f"  T-test: t = {t_stat_flair_all:.3f}, p = {p_val_flair_all:.6f}")
print(f"  Mann-Whitney U (two-sided): U = {mw_stat_flair_all:.1f}, p = {mw_p_val_flair_all:.6f}")
# One-sided test: FLAIR = 1 > FLAIR = 0
mw_stat_flair_all_onesided, mw_p_val_flair_all_onesided = mannwhitneyu(flair_1_all, flair_0_all, alternative='greater')
print(f"  Mann-Whitney U (one-sided, FLAIR = 1 > FLAIR = 0): U = {mw_stat_flair_all_onesided:.1f}, p = {mw_p_val_flair_all_onesided:.6f}")
# Clarify direction: Mann-Whitney U compares distributions. Calculate which group has higher median
if median_1_all > median_0_all:
    print(f"  Direction: FLAIR = 1 has higher median ({median_1_all:.3f} vs {median_0_all:.3f})")
else:
    print(f"  Direction: FLAIR = 0 has higher median ({median_0_all:.3f} vs {median_1_all:.3f})")

# FLAIR = 1 vs FLAIR = 0 for "beh" epochs
flair_1_beh = firing_rate_beh_array[flair_array == 1]
flair_0_beh = firing_rate_beh_array[flair_array == 0]

print(f"\nFiring Rate (Behavioral Epochs) - FLAIR:")
median_1_beh = np.median(flair_1_beh)
q1_flair_1_beh = np.percentile(flair_1_beh, 25)
q3_flair_1_beh = np.percentile(flair_1_beh, 75)
iqr_flair_1_beh = q3_flair_1_beh - q1_flair_1_beh
median_0_beh = np.median(flair_0_beh)
q1_flair_0_beh = np.percentile(flair_0_beh, 25)
q3_flair_0_beh = np.percentile(flair_0_beh, 75)
iqr_flair_0_beh = q3_flair_0_beh - q1_flair_0_beh
print(f"  FLAIR = 1: n = {len(flair_1_beh)}, mean = {np.mean(flair_1_beh):.3f} ± {np.std(flair_1_beh):.3f}, median = {median_1_beh:.3f}, IQR = {iqr_flair_1_beh:.3f}")
print(f"  FLAIR = 0: n = {len(flair_0_beh)}, mean = {np.mean(flair_0_beh):.3f} ± {np.std(flair_0_beh):.3f}, median = {median_0_beh:.3f}, IQR = {iqr_flair_0_beh:.3f}")

# Statistical tests for "beh" firing rate
t_stat_flair_beh, p_val_flair_beh = stats.ttest_ind(flair_1_beh, flair_0_beh)
mw_stat_flair_beh, mw_p_val_flair_beh = mannwhitneyu(flair_1_beh, flair_0_beh, alternative='two-sided')
print(f"  T-test: t = {t_stat_flair_beh:.3f}, p = {p_val_flair_beh:.6f}")
print(f"  Mann-Whitney U (two-sided): U = {mw_stat_flair_beh:.1f}, p = {mw_p_val_flair_beh:.6f}")
# One-sided test: FLAIR = 1 > FLAIR = 0
mw_stat_flair_beh_onesided, mw_p_val_flair_beh_onesided = mannwhitneyu(flair_1_beh, flair_0_beh, alternative='greater')
print(f"  Mann-Whitney U (one-sided, FLAIR = 1 > FLAIR = 0): U = {mw_stat_flair_beh_onesided:.1f}, p = {mw_p_val_flair_beh_onesided:.6f}")
# Clarify direction: Mann-Whitney U compares distributions. Calculate which group has higher median
if median_1_beh > median_0_beh:
    print(f"  Direction: FLAIR = 1 has higher median ({median_1_beh:.3f} vs {median_0_beh:.3f})")
else:
    print(f"  Direction: FLAIR = 0 has higher median ({median_0_beh:.3f} vs {median_1_beh:.3f})")

# ============================================================================
# VISUALIZATIONS - VIOLIN PLOTS
# ============================================================================
fig = plt.figure(figsize=(14, 9))
gs = fig.add_gridspec(2, 3, hspace=0.2, wspace=0.15, left=0.06, right=0.98, top=0.92, bottom=0.1)

fig.suptitle('Firing Rate Comparisons: IDH Status, Grade, and FLAIR', fontsize=13, fontweight='bold', y=0.96)

# 1. IDH Status - All Recording (Violin plot)
ax = fig.add_subplot(gs[0, 0])
idh_df_all = pd.DataFrame({
    'Firing Rate': np.concatenate([idh_mut_all, idh_wt_all]),
    'IDH Status': ['IDH-mut'] * len(idh_mut_all) + ['IDH-wt'] * len(idh_wt_all)
})
sns.violinplot(data=idh_df_all, x='IDH Status', y='Firing Rate', ax=ax, hue='IDH Status', palette={'IDH-mut': 'blue', 'IDH-wt': 'red'}, legend=False)
sns.stripplot(data=idh_df_all, x='IDH Status', y='Firing Rate', ax=ax, color='black', alpha=0.3, size=1.5)
ax.set_ylabel('Firing Rate (Hz)', fontsize=10)
ax.set_title(f'All Recording - IDH Status\nMW-U p={mw_p_val_all:.4f}', 
             fontsize=10, fontweight='bold', pad=3)
ax.grid(True, alpha=0.3, axis='y')
ax.tick_params(labelsize=9, pad=2)
ax.margins(y=0.05, x=0.1)
# Add significance marker (using Mann-Whitney U p-value)
sig_text = '***' if mw_p_val_all < 0.001 else '**' if mw_p_val_all < 0.01 else '*' if mw_p_val_all < 0.05 else 'ns'
ax.text(0.5, 0.95, sig_text, ha='center', fontsize=12, fontweight='bold', transform=ax.transAxes)

# 2. Grade 2/3 vs Grade 4 - All Recording (Violin plot)
ax = fig.add_subplot(gs[0, 1])
grade_df_all = pd.DataFrame({
    'Firing Rate': np.concatenate([grade_23_all, grade_4_all]),
    'Grade': ['Grade 2/3'] * len(grade_23_all) + ['Grade 4'] * len(grade_4_all)
})
sns.violinplot(data=grade_df_all, x='Grade', y='Firing Rate', ax=ax, hue='Grade', palette={'Grade 2/3': 'green', 'Grade 4': 'purple'}, legend=False)
sns.stripplot(data=grade_df_all, x='Grade', y='Firing Rate', ax=ax, color='black', alpha=0.3, size=1.5)
ax.set_ylabel('')  # Remove ylabel from middle column
ax.set_title(f'All Recording - Grade\nMW-U p={mw_p_val_grade_all:.4f}', 
             fontsize=10, fontweight='bold', pad=3)
ax.grid(True, alpha=0.3, axis='y')
ax.tick_params(labelsize=9, pad=2)
ax.margins(y=0.05, x=0.1)
# Add significance marker (using Mann-Whitney U p-value)
sig_text = '***' if mw_p_val_grade_all < 0.001 else '**' if mw_p_val_grade_all < 0.01 else '*' if mw_p_val_grade_all < 0.05 else 'ns'
ax.text(0.5, 0.95, sig_text, ha='center', fontsize=12, fontweight='bold', transform=ax.transAxes)

# 3. FLAIR = 1 vs FLAIR = 0 - All Recording (Violin plot)
ax = fig.add_subplot(gs[0, 2])
flair_df_all = pd.DataFrame({
    'Firing Rate': np.concatenate([flair_1_all, flair_0_all]),
    'FLAIR': ['FLAIR = 1'] * len(flair_1_all) + ['FLAIR = 0'] * len(flair_0_all)
})
sns.violinplot(data=flair_df_all, x='FLAIR', y='Firing Rate', ax=ax, hue='FLAIR', palette={'FLAIR = 1': 'orange', 'FLAIR = 0': 'cyan'}, legend=False)
sns.stripplot(data=flair_df_all, x='FLAIR', y='Firing Rate', ax=ax, color='black', alpha=0.3, size=1.5)
ax.set_ylabel('')  # Remove ylabel from right column
ax.set_title(f'All Recording - FLAIR\nMW-U p={mw_p_val_flair_all:.4f}', 
             fontsize=10, fontweight='bold', pad=3)
ax.grid(True, alpha=0.3, axis='y')
ax.tick_params(labelsize=9, pad=2)
ax.margins(y=0.05, x=0.1)
# Add significance marker (using Mann-Whitney U p-value)
sig_text = '***' if mw_p_val_flair_all < 0.001 else '**' if mw_p_val_flair_all < 0.01 else '*' if mw_p_val_flair_all < 0.05 else 'ns'
ax.text(0.5, 0.95, sig_text, ha='center', fontsize=12, fontweight='bold', transform=ax.transAxes)

# 4. IDH Status - Behavioral Epochs (Violin plot)
ax = fig.add_subplot(gs[1, 0])
idh_df_beh = pd.DataFrame({
    'Firing Rate': np.concatenate([idh_mut_beh, idh_wt_beh]),
    'IDH Status': ['IDH-mut'] * len(idh_mut_beh) + ['IDH-wt'] * len(idh_wt_beh)
})
sns.violinplot(data=idh_df_beh, x='IDH Status', y='Firing Rate', ax=ax, hue='IDH Status', palette={'IDH-mut': 'blue', 'IDH-wt': 'red'}, legend=False)
sns.stripplot(data=idh_df_beh, x='IDH Status', y='Firing Rate', ax=ax, color='black', alpha=0.3, size=1.5)
ax.set_ylabel('Firing Rate (Hz)', fontsize=10)
ax.set_title(f'Behavioral Epochs - IDH Status\nMW-U p={mw_p_val_beh:.4f}', 
             fontsize=10, fontweight='bold', pad=3)
ax.grid(True, alpha=0.3, axis='y')
ax.tick_params(labelsize=9, pad=2)
ax.margins(y=0.05, x=0.1)
# Add significance marker (using Mann-Whitney U p-value)
sig_text = '***' if mw_p_val_beh < 0.001 else '**' if mw_p_val_beh < 0.01 else '*' if mw_p_val_beh < 0.05 else 'ns'
ax.text(0.5, 0.95, sig_text, ha='center', fontsize=12, fontweight='bold', transform=ax.transAxes)

# 5. Grade 2/3 vs Grade 4 - Behavioral Epochs (Violin plot)
ax = fig.add_subplot(gs[1, 1])
grade_df_beh = pd.DataFrame({
    'Firing Rate': np.concatenate([grade_23_beh, grade_4_beh]),
    'Grade': ['Grade 2/3'] * len(grade_23_beh) + ['Grade 4'] * len(grade_4_beh)
})
sns.violinplot(data=grade_df_beh, x='Grade', y='Firing Rate', ax=ax, hue='Grade', palette={'Grade 2/3': 'green', 'Grade 4': 'purple'}, legend=False)
sns.stripplot(data=grade_df_beh, x='Grade', y='Firing Rate', ax=ax, color='black', alpha=0.3, size=1.5)
ax.set_ylabel('')  # Remove ylabel from middle column
ax.set_title(f'Behavioral Epochs - Grade\nMW-U p={mw_p_val_grade_beh:.4f}', 
             fontsize=10, fontweight='bold', pad=3)
ax.grid(True, alpha=0.3, axis='y')
ax.tick_params(labelsize=9, pad=2)
ax.margins(y=0.05, x=0.1)
# Add significance marker (using Mann-Whitney U p-value)
sig_text = '***' if mw_p_val_grade_beh < 0.001 else '**' if mw_p_val_grade_beh < 0.01 else '*' if mw_p_val_grade_beh < 0.05 else 'ns'
ax.text(0.5, 0.95, sig_text, ha='center', fontsize=12, fontweight='bold', transform=ax.transAxes)

# 6. FLAIR = 1 vs FLAIR = 0 - Behavioral Epochs (Violin plot)
ax = fig.add_subplot(gs[1, 2])
flair_df_beh = pd.DataFrame({
    'Firing Rate': np.concatenate([flair_1_beh, flair_0_beh]),
    'FLAIR': ['FLAIR = 1'] * len(flair_1_beh) + ['FLAIR = 0'] * len(flair_0_beh)
})
sns.violinplot(data=flair_df_beh, x='FLAIR', y='Firing Rate', ax=ax, hue='FLAIR', palette={'FLAIR = 1': 'orange', 'FLAIR = 0': 'cyan'}, legend=False)
sns.stripplot(data=flair_df_beh, x='FLAIR', y='Firing Rate', ax=ax, color='black', alpha=0.3, size=1.5)
ax.set_ylabel('')  # Remove ylabel from right column
ax.set_title(f'Behavioral Epochs - FLAIR\nMW-U p={mw_p_val_flair_beh:.4f}', 
             fontsize=10, fontweight='bold', pad=3)
ax.grid(True, alpha=0.3, axis='y')
ax.tick_params(labelsize=9, pad=2)
ax.margins(y=0.05, x=0.1)
# Add significance marker (using Mann-Whitney U p-value)
sig_text = '***' if mw_p_val_flair_beh < 0.001 else '**' if mw_p_val_flair_beh < 0.01 else '*' if mw_p_val_flair_beh < 0.05 else 'ns'
ax.text(0.5, 0.95, sig_text, ha='center', fontsize=12, fontweight='bold', transform=ax.transAxes)

plt.savefig("firing_rate_comparisons_idh_grade_flair.svg", format="svg", dpi=300, bbox_inches='tight', pad_inches=0.05)
plt.show()

print(f"\n{'='*80}")
print("ANALYSIS COMPLETE")
print(f"{'='*80}")
```
