```python
# Yield (good neuron count per insertion) for all sessions in `nwb_paths`
# — same key rules + ≥10 min behavior vs full-recording KS/CDF logic as revision/R_EI.ipynb main loop.
#
# Run after: imports + `nwb_paths`, and metadata (`manual_exclude_lists` aligned per insertion).
# Requires: `nap`, `np`, `manual_exclude_lists`.

import fnmatch
import re
from scipy.stats import kstest

MIN_BEHAVIOR_MINUTES = 10.0


def imec_key_sorter(key):
    match = re.search(r"imec(\d+)", key)
    return int(match.group(1)) if match else float("inf")


def _behavior_duration_seconds(task_times_obj):
    starts = np.asarray(task_times_obj.start, dtype=float).ravel()
    ends = np.asarray(task_times_obj.end, dtype=float).ravel()
    if starts.size != ends.size:
        return np.nan
    return float(np.sum(ends - starts))


def _ks_time_bounds_full_recording(spike_times_obj):
    tmin, tmax = np.inf, -np.inf
    for u in range(len(spike_times_obj)):
        idx = spike_times_obj[u].as_series().index.values
        if len(idx):
            tmin = min(tmin, float(np.min(idx)))
            tmax = max(tmax, float(np.max(idx)))
    if np.isfinite(tmin) and np.isfinite(tmax):
        return tmin, tmax
    return np.nan, np.nan


rows = []
insertion = 0

for i in range(len(nwb_paths)):
    try:
        data = nap.load_file(nwb_paths[i])
        keys = data.keys()

        template = "*imec*"
        keys = [key for key in keys if fnmatch.fnmatch(key, template)]
        ks_keys = [key for key in keys if "KS4" in key]
        if ks_keys:
            keys = ks_keys
        th8_keys = [key for key in keys if "Th=8" in key]
        if th8_keys:
            keys = th8_keys
        else:
            th_keys = [key for key in keys if "Th=" in key]
            if th_keys:
                keys = th_keys
        template_sentgen = "*sentgen*"
        template_auto = "*_auto*"
        keys = [
            key
            for key in keys
            if not fnmatch.fnmatch(key, template_sentgen) and not fnmatch.fnmatch(key, template_auto)
        ]
        if ("NP137" in str(nwb_paths[i])) or ("NP139_B2" in str(nwb_paths[i])):
            keys = [key for key in keys if "imec1" not in key]
        keys = sorted(keys, key=imec_key_sorter)

        for s in range(len(keys)):
            spike_times = data[keys[s]]
            firingRates_all = spike_times.metadata["rate"]

            if "TaskTimes" in data.keys():
                task_times = data["TaskTimes"]
            elif "task_times" in data.keys():
                task_times = data["task_times"]
            else:
                rows.append((nwb_paths[i].stem, keys[s], np.nan, "no_task_times", np.nan))
                insertion += 1
                continue

            start_time = task_times.start
            end_time = task_times.end
            beh_epochs = nap.IntervalSet(start=task_times.start, end=task_times.end)

            beh_sec = _behavior_duration_seconds(task_times)
            beh_min = beh_sec / 60.0 if np.isfinite(beh_sec) else np.nan
            use_full_recording = (not np.isfinite(beh_min)) or (beh_min < MIN_BEHAVIOR_MINUTES)

            if use_full_recording:
                firingRates_beh = firingRates_all
                min_time, max_time = _ks_time_bounds_full_recording(spike_times)
                if not (np.isfinite(min_time) and np.isfinite(max_time)) or (max_time <= min_time):
                    rows.append((nwb_paths[i].stem, keys[s], beh_min, "full_ks_bounds_fail", np.nan))
                    insertion += 1
                    continue
                ks_mode = "full_recording"
            else:
                spike_times_beh = spike_times.restrict(beh_epochs)
                firingRates_beh = spike_times_beh.metadata["rate"]
                min_time = start_time[0]
                max_time = end_time[-1]
                ks_mode = "behavior_ge_10min"

            ks_stats = np.zeros(len(spike_times))
            ks_pvals = np.zeros(len(spike_times))
            for u in range(len(spike_times)):
                test = spike_times[u].as_series().index.values
                if len(test) > 1:
                    normalized_spike_times = (test - min_time) / (max_time - min_time)
                    ks_result = kstest(normalized_spike_times, "uniform")
                    ks_stats[u] = ks_result.statistic
                    ks_pvals[u] = ks_result.pvalue
                else:
                    ks_stats[u] = np.nan
                    ks_pvals[u] = np.nan

            violationThreshold = 3 / 1000
            violationPct = np.zeros(len(spike_times))
            for u in range(len(spike_times)):
                unit = spike_times[u].as_series().index
                if len(unit) < 100:
                    violationPct[u] = 1
                else:
                    isi = unit.diff()[1:]
                    violations = np.where(isi < violationThreshold)
                    violationPct[u] = np.array(violations).size / len(isi)

            if "KSLabel" in spike_times.metadata:
                KSLabels = spike_times.metadata["KSLabel"]
            else:
                KSLabels = spike_times.metadata["quality"]

            firingRates = firingRates_beh
            mask1 = violationPct < 3 / 100
            mask2 = firingRates > 0.5
            mask3 = KSLabels != "noise"
            mask4 = ks_stats < 0.3
            mask = mask1 & mask2 & mask3 & mask4
            indicesFinal = firingRates.index[mask]

            indicesFinal = np.setdiff1d(indicesFinal, manual_exclude_lists[insertion])
            n_yield = len(indicesFinal)
            rows.append((nwb_paths[i].stem, keys[s], beh_min, ks_mode, n_yield))
            insertion += 1

    except Exception as e:
        rows.append((nwb_paths[i].stem, "", np.nan, repr(e), np.nan))
        import traceback

        traceback.print_exc()

print(f"{'session':<14} {'key':<38} {'beh_min':>8} {'ks_mode':<18} {'yield':>6}")
print("-" * 95)
for r in rows:
    sess, key, bm, mode, y = r
    bm_s = f"{bm:8.2f}" if np.isfinite(bm) else "     nan"
    y_s = f"{int(y):6d}" if np.isfinite(y) else "   nan"
    key_disp = (key if key else "")[:36]
    print(f"{sess:<14} {key_disp:<38} {bm_s:>8} {str(mode):<18} {y_s:>6}")

print(f"\nTotal insertions listed: {len(rows)} (expected {len(manual_exclude_lists)})")
```

```python
# MAIN LOOP TO GET WAVEFORM AND DEPTH DATA
# Key filtering + KS/firing-rate windowing aligned with revision/R_EI.ipynb:
#   summed behavior duration >= 10 min  -> restrict to beh epochs, KS on task window
#   < 10 min (or invalid duration)     -> full recording for rate filter + KS on spike-time span
# Analyzer path: strip leading catgt_/mc_ from '{key}_sparse' (your prior R_yieldCheck convention).

import fnmatch
import matplotlib as mpl
import matplotlib.pyplot as plt
import numpy as np
import re
import os
from scipy import stats
from scipy.linalg import norm
from scipy.stats import kstest

MIN_BEHAVIOR_MINUTES = 10.0


def imec_key_sorter(key):
    """Extract the number after 'imec', default to a large number if not found."""
    match = re.search(r"imec(\d+)", key)
    return int(match.group(1)) if match else float("inf")


def _behavior_duration_seconds(task_times_obj):
    starts = np.asarray(task_times_obj.start, dtype=float).ravel()
    ends = np.asarray(task_times_obj.end, dtype=float).ravel()
    if starts.size != ends.size:
        return np.nan
    return float(np.sum(ends - starts))


def _ks_time_bounds_full_recording(spike_times_obj):
    tmin, tmax = np.inf, -np.inf
    for u in range(len(spike_times_obj)):
        idx = spike_times_obj[u].as_series().index.values
        if len(idx):
            tmin = min(tmin, float(np.min(idx)))
            tmax = max(tmax, float(np.max(idx)))
    if np.isfinite(tmin) and np.isfinite(tmax):
        return tmin, tmax
    return np.nan, np.nan


indicesAll = []
waveformAll = []
depthAll = []

for i in range(len(nwb_paths)):
    data = nap.load_file(nwb_paths[i])
    keys = data.keys()

    template = "*imec*"
    keys = [key for key in keys if fnmatch.fnmatch(key, template)]
    ks_keys = [key for key in keys if "KS4" in key]
    if ks_keys:
        keys = ks_keys
    th8_keys = [key for key in keys if "Th=8" in key]
    if th8_keys:
        keys = th8_keys
    else:
        th_keys = [key for key in keys if "Th=" in key]
        if th_keys:
            keys = th_keys
    template_sentgen = "*sentgen*"
    template_auto = "*_auto*"
    keys = [
        key
        for key in keys
        if not fnmatch.fnmatch(key, template_sentgen) and not fnmatch.fnmatch(key, template_auto)
    ]
    if ("NP137" in str(nwb_paths[i])) or ("NP139_B2" in str(nwb_paths[i])):
        keys = [key for key in keys if "imec1" not in key]
    keys = sorted(keys, key=imec_key_sorter)

    for s in range(len(keys)):
        print(keys[s])

        spike_times = data[keys[s]]
        firingRates_all = spike_times.metadata["rate"]

        if "TaskTimes" in data.keys():
            task_times = data["TaskTimes"]
        elif "task_times" in data.keys():
            task_times = data["task_times"]
        else:
            print("  No TaskTimes / task_times; skipping shank")
            continue

        start_time = task_times.start
        end_time = task_times.end
        beh_epochs = nap.IntervalSet(start=task_times.start, end=task_times.end)

        beh_sec = _behavior_duration_seconds(task_times)
        beh_min = beh_sec / 60.0 if np.isfinite(beh_sec) else np.nan
        use_full_recording = (not np.isfinite(beh_min)) or (beh_min < MIN_BEHAVIOR_MINUTES)

        if use_full_recording:
            firingRates_beh = firingRates_all
            min_time, max_time = _ks_time_bounds_full_recording(spike_times)
            if not (np.isfinite(min_time) and np.isfinite(max_time)) or (max_time <= min_time):
                print(f"  behavior {beh_min:.2f} min, full-recording KS bounds failed; skipping shank")
                continue
            print(
                f"  behavior {beh_min:.2f} min (< {MIN_BEHAVIOR_MINUTES}) -> full recording; KS [{min_time:.3f}, {max_time:.3f}] s"
            )
        else:
            spike_times_beh = spike_times.restrict(beh_epochs)
            firingRates_beh = spike_times_beh.metadata["rate"]
            min_time = start_time[0]
            max_time = end_time[-1]
            print(
                f"  behavior {beh_min:.2f} min (>= {MIN_BEHAVIOR_MINUTES}) -> beh window; KS [{min_time:.3f}, {max_time:.3f}] s"
            )

        ks_stats = np.zeros(len(spike_times))
        ks_pvals = np.zeros(len(spike_times))
        for u in range(len(spike_times)):
            test = spike_times[u].as_series().index.values
            if len(test) > 1:
                normalized_spike_times = (test - min_time) / (max_time - min_time)
                ks_result = kstest(normalized_spike_times, "uniform")
                ks_stats[u] = ks_result.statistic
                ks_pvals[u] = ks_result.pvalue
            else:
                ks_stats[u] = np.nan
                ks_pvals[u] = np.nan

        violationThreshold = 3 / 1000
        violationPct = np.zeros(len(spike_times))
        for u in range(len(spike_times)):
            unit = spike_times[u].as_series().index
            if len(unit) < 100:
                violationPct[u] = 1
            else:
                isi = unit.diff()[1:]
                violations = np.where(isi < violationThreshold)
                violationPct[u] = np.array(violations).size / len(isi)

        if "KSLabel" in spike_times.metadata:
            KSLabels = spike_times.metadata["KSLabel"]
        else:
            KSLabels = spike_times.metadata["quality"]

        firingRates = firingRates_beh
        mask1 = violationPct < 3 / 100
        mask2 = firingRates > 0.5
        mask3 = KSLabels != "noise"
        mask4 = ks_stats < 0.3
        mask = mask1 & mask2 & mask3 & mask4
        indicesFinal = firingRates.index[mask]
        print(indicesFinal)

        spike_times_good = spike_times[indicesFinal]
        print(f"Number of good neurons: {len(spike_times_good)}")

        kilosort_run = f"{keys[s]}_sparse"
        rec_id = os.path.basename(nwb_paths[i])
        rec_id = os.path.splitext(rec_id)[0]
        if kilosort_run.startswith("catgt_"):
            kilosort_run_path = kilosort_run.replace("catgt_", "", 1)
        elif kilosort_run.startswith("mc_"):
            kilosort_run_path = kilosort_run.replace("mc_", "", 1)
        else:
            kilosort_run_path = kilosort_run
        analyzer_path = f"/data_store2/neuropixels/nwb/old/{rec_id}/SI/{kilosort_run_path}"

        analyzer = load_sorting_analyzer(analyzer_path)

        templates = analyzer.get_extension("templates")
        avg_templates = templates.get_data(operator="average")
        print(avg_templates.shape)

        waveform_list = []
        depth_list = []
        for unit_id in indicesFinal:
            template = avg_templates[unit_id]
            max_amp_per_channel = np.max(np.abs(template), axis=0)
            max_amp_ch = np.argmax(max_amp_per_channel)
            waveform = template[:, max_amp_ch]
            waveform_list.append(waveform)

            firing_rate_val = firingRates.loc[unit_id] if unit_id in firingRates.index else None
            if firing_rate_val is not None:
                firing_rate_str = f"{firing_rate_val:.2f} Hz"
            else:
                firing_rate_str = "n/a"

            plt.figure()
            plt.plot(waveform)
            plt.title(
                f"Waveform for unit {unit_id} (channel {max_amp_ch})\nFiring rate: {firing_rate_str}"
            )
            plt.xlabel("Time (samples)")
            plt.ylabel("Amplitude")
            plt.show()

        depths = analyzer.get_extension("unit_locations").get_data()[:, 1]
        depths_good = depths[indicesFinal]
        depth_list.append(depths_good)

        indicesAll.append(indicesFinal)
        waveformAll.append(waveform_list)
        depthAll.append(depth_list)
```
