```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import kruskal, mannwhitneyu
import pandas as pd
plt.rcParams['svg.fonttype'] = 'none'

grade_unique = sorted(set(grade_list))

# Organize yields and opercular statuses by grade
yields_by_grade = {g: [] for g in grade_unique}
operculars_by_grade = {g: [] for g in grade_unique}
for g, y, op in zip(grade_list, yield_list, opercular_list):
    yields_by_grade[g].append(y)
    operculars_by_grade[g].append(op)

# Calculate means and errors by grade (all operculars together)
mean_yields_grade = []
stderr_yields_grade = []
for g in grade_unique:
    ys = yields_by_grade[g]
    mean_yields_grade.append(np.mean(ys) if len(ys) > 0 else 0)
    stderr_yields_grade.append(np.std(ys, ddof=1)/np.sqrt(len(ys)) if len(ys) > 1 else 0)

# --- Summary table for average yield by grade ---
summary_stats = []
for g, mean, err, ys in zip(grade_unique, mean_yields_grade, stderr_yields_grade, [yields_by_grade[g] for g in grade_unique]):
    summary_stats.append({
        'Grade': g,
        'N': len(ys),
        'Mean_Yield': mean,
        'SEM': err,
        'Median': np.median(ys) if len(ys) else 0,
        'Min': np.min(ys) if len(ys) else 0,
        'Max': np.max(ys) if len(ys) else 0
    })
summary_df = pd.DataFrame(summary_stats)
print("Summary stats of average yield by grade:")
print(summary_df.to_string(index=False))
print()

# --- Pairwise statistics table ---
print("Pairwise Mann-Whitney U tests (all yields, ignoring opercular):")
pairwise_results = []
for i in range(len(grade_unique)):
    for j in range(i+1, len(grade_unique)):
        g1, g2 = grade_unique[i], grade_unique[j]
        ys1, ys2 = yields_by_grade[g1], yields_by_grade[g2]
        if len(ys1) > 0 and len(ys2) > 0:
            ustat, pval = mannwhitneyu(ys1, ys2, alternative='two-sided')
            pairwise_results.append({"Grade 1": g1, "Grade 2": g2, "U": ustat, "p-value": pval})
pairwise_df = pd.DataFrame(pairwise_results)
if not pairwise_df.empty:
    # Fix: Use full option name to avoid ambiguity in set_option
    pd.set_option('display.precision', 4)
    print(pairwise_df.to_string(index=False))
    print()

fig, ax = plt.subplots(figsize=(6, 4))

bar_width = 0.8

# Only show a single bar per grade (not separated by opercular)
bar_color = np.array([0.75, 0.75, 0.75])  # Neutral gray for the bar

bars = ax.bar(
    grade_unique, mean_yields_grade,
    bar_width,
    color=bar_color,
    yerr=stderr_yields_grade,
    capsize=5,
    label='Mean yield (all)',
    edgecolor='k',
    zorder=1
)

# Colors for opercular=0 and opercular=1
color_op0 = np.array([0.12, 0.47, 0.71])  # blue-ish
color_op1 = np.array([1, 0.55, 0])        # orange-ish

# Overlay each individual point, colored by opercular status
for i, g in enumerate(grade_unique):
    ys = yields_by_grade[g]
    ops = operculars_by_grade[g]
    # jitter for x
    xvals = np.random.normal(g, 0.11, size=len(ys))
    # plot opercular=0
    mask_op0 = np.array(ops) == 0
    mask_op1 = np.array(ops) == 1
    if np.any(mask_op0):
        ax.scatter(xvals[mask_op0], np.array(ys)[mask_op0], color=color_op0, alpha=0.7, s=45, zorder=10, marker='o', edgecolor='w', label=None)
    if np.any(mask_op1):
        ax.scatter(xvals[mask_op1], np.array(ys)[mask_op1], color=color_op1, alpha=0.7, s=45, zorder=10, marker='o', edgecolor='w', label=None)

# Overlay the mean and errorbar as red horizontal lines/dots
for i, g in enumerate(grade_unique):
    mean = mean_yields_grade[i]
    err = stderr_yields_grade[i]
    ax.errorbar(g, mean, yerr=err, fmt='o', color='red', capsize=6, markersize=7, zorder=15,
                label=None if i != 0 else 'Mean ± SEM')

ax.set_xlabel('Grade')
ax.set_ylabel('Yield')
ax.set_title('Yield by Grade')
ax.set_xticks(grade_unique)

# Set panel aspect
ax.set_aspect(1/80)

# Build a legend for bar, data points, and mean
from matplotlib.lines import Line2D
handles = [
    bars,
    Line2D([], [], color=color_op0, marker='o', lw=0, markersize=9, markeredgecolor='w', label='Individual (opercular=0)', alpha=0.7),
    Line2D([], [], color=color_op1, marker='o', lw=0, markersize=9, markeredgecolor='w', label='Individual (opercular=1)', alpha=0.7),
    Line2D([], [], color='red', marker='o', lw=0, markersize=9, label='Mean ± SEM'),
]
labels = [
    'Mean yield (all)',
    'Individual (opercular=0)',
    'Individual (opercular=1)',
    'Mean ± SEM',
]
ax.legend(handles, labels)

# -- Statistical test between groups (Kruskal-Wallis for all grades, ignoring opercular)
stat, p = kruskal(*[yields_by_grade[g] for g in grade_unique if len(yields_by_grade[g]) > 0])
ax.text(0.98, 0.98, f"Kruskal-Wallis\n$p={p:.3g}$",
        transform=ax.transAxes, va='top', ha='right', fontsize=11,
        bbox=dict(boxstyle='round', facecolor='white', alpha=0.7))

# -- Pairwise stats (Mann-Whitney U) between pairs of grades (using all yields, ignoring opercular)
# Annotate on bar plot only for consecutive pairs, but print all pairwise stats in console above
for i in range(len(grade_unique)-1):
    g1 = grade_unique[i]
    g2 = grade_unique[i+1]
    ys1 = yields_by_grade[g1]
    ys2 = yields_by_grade[g2]
    if len(ys1) > 0 and len(ys2) > 0:
        ustat, pval = mannwhitneyu(ys1, ys2, alternative='two-sided')
        y_max = max(
            max(ys1) if len(ys1) else 0,
            max(ys2) if len(ys2) else 0,
            mean_yields_grade[i]+stderr_yields_grade[i],
            mean_yields_grade[i+1]+stderr_yields_grade[i+1]
        )
        y_ann = y_max + 0.03 * (max(mean_yields_grade) if mean_yields_grade else 1)
        # Only annotate if significant
        if pval < 0.05:
            ax.plot([g1, g2], [y_ann, y_ann], 'k-', lw=1)
            ax.text((g1 + g2)/2, y_ann, "*" if pval < 0.05 else "n.s.", ha='center', va='bottom', fontsize=14)

plt.tight_layout()
plt.savefig('fig1yield_grade.svg', format='svg', dpi=300, bbox_inches='tight')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
plt.rcParams['svg.fonttype'] = 'none' 

# All lists to visualize
all_lists = [
    subj_list,
    grade_list,
    path_list,
    yield_list,
    opercular_list,
    region_list,
    age_list,
    gender_list
]
list_names = [
    'subj',
    'grade',
    'path',
    'yield',
    'opercular',
    'region',
    'age',
    'gender'
]

# Prepare matrix for heatmap (numeric codes for categorical, numbers for numeric)
matrix = []
row_vmin = []
row_vmax = []
for idx, lst in enumerate(all_lists):
    # Special handling for region row
    if list_names[idx] == 'region':
        # pSTG = 2, parsOp and parsTr = 1, all others = 0
        region_numeric = []
        for r in lst:
            if r == 'pSTG':
                region_numeric.append(2)
            elif r == 'parsOp' or r == 'parsTr' or r == 'vPrCG':
                region_numeric.append(1)
            else:
                region_numeric.append(0)
        matrix.append(region_numeric)
        row_vmin.append(0)
        row_vmax.append(2)
    # Special handling for path row
    elif list_names[idx] == 'path':
        # oligo = 0, astro = 1, gbm = 2
        path_numeric = []
        for p in lst:
            if p.lower().startswith('oli'):
                path_numeric.append(0)
            elif p.lower().startswith('ast'):
                path_numeric.append(1)
            elif p.lower().startswith('gbm'):
                path_numeric.append(2)
            else:
                path_numeric.append(np.nan)  # fallback for unknown
        matrix.append(path_numeric)
        row_vmin.append(0)
        row_vmax.append(2)
    elif all(isinstance(x, (int, float, np.integer, np.floating)) for x in lst):
        matrix.append(lst)
        row_vmin.append(np.min(lst))
        row_vmax.append(np.max(lst))
    else:
        cat = pd.Categorical(lst)
        matrix.append(cat.codes)
        row_vmin.append(np.min(cat.codes))
        row_vmax.append(np.max(cat.codes))

matrix = np.array(matrix)

# Create a custom colormap from white to black
from matplotlib.colors import LinearSegmentedColormap
wb_cmap = LinearSegmentedColormap.from_list("wb", ["white", "black"])

# --- Make the heatmap and the entire plot with aspect ratio 1.5, with a single box around all cells ---

n_rows, n_cols = matrix.shape
aspect_ratio = 1
base = 8  # base size for the shorter side
if n_cols >= n_rows:
    figsize = (base * aspect_ratio, base)
else:
    figsize = (base, base * aspect_ratio)

fig, ax = plt.subplots(figsize=figsize)

# For each row, plot as a single-row heatmap (no colorbars here)
for i in range(n_rows):
    row = matrix[i][np.newaxis, :]
    ax.imshow(
        row,
        aspect='auto',
        cmap=wb_cmap,
        vmin=row_vmin[i],
        vmax=row_vmax[i],
        extent=[-0.5, n_cols-0.5, i+0.5, i-0.5]
    )

# For just the yield row, plot the number in the cell
yield_row_idx = list_names.index('yield')
for j in range(n_cols):
    val = all_lists[yield_row_idx][j]
    # Choose text color for contrast
    # If value is high (dark cell), use white; if low (light cell), use black
    norm_val = (val - row_vmin[yield_row_idx]) / (row_vmax[yield_row_idx] - row_vmin[yield_row_idx] + 1e-8)
    text_color = 'white' if norm_val > 0.5 else 'black'
    ax.text(j, yield_row_idx, str(val), ha='center', va='center', fontsize=8, color=text_color, fontweight='bold')

# Set yticks to row names
ax.set_yticks(np.arange(n_rows))
ax.set_yticklabels(list_names, fontsize=10)
ax.set_xticks(np.arange(n_cols))
ax.set_xticklabels([str(idx) for idx in range(n_cols)], rotation=90, fontsize=8)
ax.set_xlabel('Index in list (subject/sample)')

# Remove all spines
for spine in ax.spines.values():
    spine.set_visible(False)

# Draw a single box around all cells using Rectangle from matplotlib.patches
import matplotlib.patches as mpatches
rect = mpatches.Rectangle(
    (-0.5, -0.5),
    n_cols,
    n_rows,
    fill=False,
    edgecolor='black',
    linewidth=2
)
ax.add_patch(rect)

# Remove whitespace between rows
ax.set_ylim(n_rows-0.5, -0.5)
ax.set_xlim(-0.5, n_cols-0.5)

# Set the aspect ratio of the data region to 1.5
ax.set_aspect(aspect_ratio, adjustable='box')

plt.subplots_adjust(left=0.15, right=0.98, top=0.92, bottom=0.15)
plt.title('Stacked Heatmap of All Lists (white=low, black=high)', y=1.02)
plt.gcf().set_size_inches(figsize)
plt.savefig('heatmap_summary.svg', format='svg', dpi=300, bbox_inches='tight')
plt.show()

# --- Separate plot for colorbars ---
from matplotlib import gridspec

n_rows = len(all_lists)
fig_cb, axes_cb = plt.subplots(nrows=n_rows, ncols=1, figsize=(2, n_rows * 0.7), constrained_layout=True)

if n_rows == 1:
    axes_cb = [axes_cb]

for i in range(n_rows):
    # Create a dummy image for colorbar
    dummy = np.linspace(row_vmin[i], row_vmax[i], 256).reshape(1, -1)
    im = axes_cb[i].imshow(dummy, aspect='auto', cmap=wb_cmap, vmin=row_vmin[i], vmax=row_vmax[i])
    cbar = plt.colorbar(im, ax=axes_cb[i], orientation='horizontal', fraction=0.7, pad=0.2)
    cbar.ax.tick_params(labelsize=7)
    cbar.set_label(list_names[i], fontsize=8)
    axes_cb[i].axis('off')

fig_cb.suptitle('Colorbars for Each Row', fontsize=12, y=1.02)
plt.savefig('heatmap_colorbars.svg', format='svg', dpi=300, bbox_inches='tight')
plt.show()
```
